# 🔗 Correlação e Associação entre Variáveis (Contínuas e Discretas)

Entender a **relação estatística entre variáveis** é um passo fundamental em projetos de Ciência de Dados, impactando desde o pré-processamento até a modelagem e a avaliação de modelos. Dependendo do **tipo de dado** (contínuo ou categórico), diferentes métodos estatísticos são usados para mensurar **correlação** ou **associação** entre as variáveis.

---

## 📘 Conceito Teórico

- **Correlação** refere-se a uma medida de **relação linear** entre duas variáveis numéricas contínuas.
- **Associação** é um conceito mais geral, aplicável a variáveis **categóricas** ou **mistas**, e indica se duas variáveis têm relação estatística, não necessariamente linear.

---

## 🔢 Tipos de Dados Envolvidos

| Variável X       | Variável Y       | Tipo de relação            | Exemplo prático                          |
|------------------|------------------|-----------------------------|-------------------------------------------|
| Contínua         | Contínua         | Correlação linear ou não    | Altura vs Peso                            |
| Discreta         | Discreta         | Associação                  | Estado Civil vs Faixa Etária              |
| Contínua         | Discreta         | Correlação ponto-bisserial  | Idade vs Churn (0/1)                      |
| Ordinal          | Ordinal          | Correlação de Spearman      | Nível de dor (1 a 5) vs Satisfação (1 a 5) |

---

## 🧾 Principais Medidas de Correlação e Associação

### 🔹 1. Correlação de Pearson (Contínua × Contínua)

- Mede a **força e direção** da **relação linear** entre duas variáveis contínuas.

$$
\rho(X, Y) = \frac{Cov(X, Y)}{\sigma_X \sigma_Y}
$$

- Varia entre -1 (correlação negativa perfeita) e +1 (positiva perfeita). Zero implica ausência de relação linear.

✅ Ideal para:
- Explorar dependência entre variáveis numéricas normalizadas ou padronizadas.
- Eliminar colinearidade em modelos lineares.

⚠️ Armadilhas:
- Insensível a relações **não lineares**.
- Muito sensível a **outliers**.

---

### 🔹 2. Correlação de Spearman (Monótona, Ordinal ou Não-Linear)

- Baseia-se no **posto (ranking)** dos dados e captura relações **monótonas** (não necessariamente lineares).

$$
\rho_s = \text{Correlação de Pearson dos ranks}
$$

✅ Ideal para:
- Dados **não normais**.
- Relações não lineares mas **monótonas** (ex: logarítmica, exponencial).
- Variáveis **ordinais**.

---

### 🔹 3. Correlação de Kendall Tau (Ordinal, robusta)

- Mede o grau de concordância entre duas classificações **ordinais**.

- Mais robusta que Spearman para **pequenos conjuntos** ou muitos empates.

✅ Ideal para:
- Análise robusta de rankings.
- Pequenos datasets ou com empates.

---

### 🔹 4. Correlação Ponto-Bisserial (Contínua × Binária)

- Variante da correlação de Pearson usada quando uma variável é contínua e a outra é **binária (0/1)**.

$$
r_{pb} = \frac{\overline{X}_1 - \overline{X}_0}{s_X} \cdot \sqrt{\frac{n_1 n_0}{n^2}}
$$

✅ Ideal para:
- Relação entre uma métrica (idade, salário) e um desfecho binário (churn, fraude).

---

### 🔹 5. Qui-quadrado de Independência (Discreta × Discreta)

- Testa se há **associação estatística** entre variáveis categóricas (nominais ou discretas).

$$
\chi^2 = \sum \frac{(O_{ij} - E_{ij})^2}{E_{ij}}
$$

- Onde \( O_{ij} \) é a frequência observada e \( E_{ij} \) é a frequência esperada na célula da tabela de contingência.

✅ Ideal para:
- Testar dependência entre duas variáveis discretas.

---

### 🔹 6. Cramér's V (Associação Normalizada)

- Varia de 0 (sem associação) a 1 (associação perfeita).
- Calculado a partir do teste qui-quadrado:

$$
V = \sqrt{ \frac{\chi^2}{n \cdot (k - 1)} }
$$

- Onde \( k \) é o menor número de categorias entre as variáveis.

✅ Ideal para:
- Quantificar a força da associação entre variáveis **categóricas nominais**.

---

### 🔹 7. Theil’s U (Informação Condicional)

- Métrica **assimétrica** baseada em **entropia**, muito útil em associação de categóricas.

$$
U(Y|X) = \frac{H(Y) - H(Y|X)}{H(Y)}
$$

✅ Ideal para:
- Casos em que se deseja saber **o quanto X reduz a incerteza sobre Y**.
- Melhor que Cramér’s V para casos assimétricos (ex: "Produto" → "Categoria").

## 🔀 Diferenças entre Correlação e Associação

| Aspecto             | Correlação                         | Associação                             |
|---------------------|-------------------------------------|----------------------------------------|
| Tipo de variável    | Contínua (em geral)                | Categórica ou mista                    |
| Tipo de relação     | Linear (ou monótona)               | Qualquer dependência estatística       |
| Medidas principais  | Pearson, Spearman, Kendall         | Qui-quadrado, Cramér’s V, Theil’s U    |
| Simetria            | Simétrica                          | Pode ser assimétrica (ex: Theil’s U)   |

---

## ✅ Vantagens do Uso

- 📈 Detecta variáveis **redundantes ou colineares**
- 🔍 Ajuda a identificar **relações implícitas** nos dados
- ⚠️ Previne **problemas de multicolinearidade** em modelos lineares
- 🧠 Gera insights exploratórios úteis na criação de variáveis derivadas

---

## ⚠️ Limitações e Armadilhas

- Pearson e Spearman assumem dados numéricos e contínuos
- Qui-quadrado requer **frequências esperadas mínimas** adequadas
- Categóricas com muitos níveis podem distorcer a associação
- Correlação **não implica causalidade**
- Relações **não-lineares complexas** podem não ser capturadas

---

## 🔍 Interpretação Prática para Modelagem

### 💡 Modelos impactados por correlação:

- **Regressão Linear / Logística**: multicolinearidade afeta estabilidade dos coeficientes.
- **KNN, SVM, Árvores**: menos sensíveis, mas podem ser otimizados com seleção de variáveis.

### 📉 Impacto na performance:

- Correlação alta entre features pode causar **overfitting**.
- Variáveis não associadas ao target geram **ruído** e reduzem capacidade preditiva.

## 🧠 Estratégias de Análise Recomendadas
- Usar Pearson ou Spearman entre variáveis numéricas para eliminar colinearidade.

- Aplicar qui-quadrado, Cramér's V ou Theil's U para entender relações entre categóricas.

- Avaliar associação variável explicativa × target antes da modelagem.

- Com variáveis mistas (contínua × discreta), usar análise de variância (ANOVA), mutual information, ou modelos de árvore.

- Validar a significância estatística com p-valores e testes de hipóteses.
