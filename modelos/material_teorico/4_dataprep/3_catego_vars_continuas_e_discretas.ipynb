# 🧮 Categorização de Variáveis Contínuas e Discretas (Data Preparation)

A **categorização (ou discretização)** de variáveis contínuas e discretas é uma técnica de **pré-processamento de dados** que converte uma variável numérica em uma variável categórica, dividindo seu domínio em faixas ou grupos.

Essa abordagem é comum em projetos de ciência de dados para facilitar a modelagem, **aumentar interpretabilidade** ou lidar com **não-linearidades** em algoritmos sensíveis à escala ou distribuição.

---

## 🧠 Conceito Teórico

Categorização significa **agrupar valores contínuos ou discretos em categorias ou *bins***. Em vez de lidar com uma variável como um número contínuo (ex: idade = 37), passamos a tratá-la como uma classe (ex: "30-40 anos").

---

## 🧾 Características Principais e Tipos de Binning

### 🔹 1. **Binning Uniforme (Equal-Width Binning)**
- Divide o intervalo da variável em faixas de **mesma largura**.
- Exemplo: [0-10), [10-20), [20-30)

```python
pd.cut(df['idade'], bins=5)
```

### 🔹 2. **Binning por Frequência (Equal-Frequency / Quantile Binning)**
- Cada faixa contém aproximadamente o **mesmo número de observações**.
- Exemplo: tercis, quartis, decis.

```python
pd.qcut(df['idade'], q=4)
```

### 🔹 3. **Binning Personalizado**
- Criado com base em domínio ou negócio.
- Ex: Faixas salariais: "baixa", "média", "alta".

### 🔹 4. **Categorização via Árvores de Decisão**
- Utiliza algoritmos como **CHAID** ou **Decision Trees** para encontrar pontos de corte com base em variáveis-alvo (ex: churn, inadimplência).
- Mais eficazes em capturar **relações não-lineares**.

---

## 📊 Aplicações Práticas em Ciência de Dados

| Caso de Uso | Exemplo | Estratégia |
|-------------|---------|------------|
| Modelagem de crédito | Renda, idade, score de crédito | Bins com base no GINI por faixa |
| Segmentação de clientes | Faturamento, volume de compras | Quartis ou clustering |
| Modelos de churn | Tempo de contrato | Árvores de decisão para cortes ótimos |
| Scorecard (score de risco) | Transformação de variáveis contínuas em bins monotônicos | WOE binning |

---

## 🔁 Diferenças entre Conceitos Relacionados

| Conceito | Descrição |
|----------|-----------|
| Categorização | Converte variável contínua em categórica |
| Normalização / Padronização | Escala variável contínua, mas **não altera tipo** |
| One-hot Encoding | Converte variável categórica para dummies |
| Label Encoding | Converte categorias para inteiros (sem binning) |

---

## ✅ Vantagens

- **Facilita a modelagem interpretável** (logística, scorecards, etc.)
- Ajuda a lidar com **não-linearidades**
- Reduz sensibilidade a **outliers**
- Permite análise de risco ou comportamento por faixas
- Útil quando modelos não assumem relação funcional exata

---

## ⚠️ Limitações e Armadilhas

- **Perda de informação**: discretizar reduz variabilidade e precisão
- **Escolha ruim de cortes**: bins mal escolhidos induzem ruído
- **Pode introduzir artefatos**: fronteiras artificiais criam descontinuidades
- **Problemas em modelos complexos**: pode dificultar uso em redes neurais ou regressões com penalizações

---

## 🧠 Implicações na Modelagem

| Modelo | Impacto da Categorização |
|--------|---------------------------|
| Regressão Logística | Melhora interpretação (ex: Odds por faixa) |
| Árvores de Decisão | Pode ser redundante (árvores já categorizam) |
| Redes Neurais | Pode **reduzir performance** se não for embutido |
| Scorecards | Essencial para monotonicidade e cálculo de WOE/IV |
| Modelos baseados em distâncias (KNN) | Requer codificação apropriada pós-categorização |

## 🎯 Exemplos de Estratégias Reais

### 📌 Exemplo 1: Faixas de Idade em um Modelo de Risco

```python
bins = [18, 25, 35, 45, 60, 100]
labels = ['18-25', '26-35', '36-45', '46-60', '60+']
df['idade_faixa'] = pd.cut(df['idade'], bins=bins, labels=labels)
```

- Posteriormente codificada como `OneHot` para regressão logística.

---

### 📌 Exemplo 2: WOE Binning supervisionado (usado em crédito)

Usa algoritmos de binning supervisionado para maximizar **monotonicidade** com a variável de interesse (`y`, ex: inadimplência).

```python
from optbinning import BinningProcess
bp = BinningProcess(variable_names=['idade'], categorical_variables=[])
X = df[['idade']]
y = df['inadimplente']
bp.fit(X, y)
```

## 🔎 Comparação entre Estratégias

| Estratégia | Quando usar | Vantagem | Limitação |
|------------|-------------|----------|-----------|
| Equal-width | Distribuições uniformes | Simples | Pode gerar bins desequilibrados |
| Quantílica (qcut) | Dados assimétricos | Proporções balanceadas | Perde relação com valores reais |
| Personalizada | Casos de negócio | Interpretável | Subjetiva |
| Árvores de decisão | Classificação supervisionada | Otimizada | Requer target (não serve para unsupervised) |

---

## 📌 Conclusão

Categorização é uma ferramenta poderosa, mas que **deve ser usada com critério estatístico e contextual**. Em ciência de dados, é ideal quando:

- Há necessidade de **interpretação clara**
- Os dados apresentam **relações não-lineares**
- O domínio exige **segmentação por faixas**

> ⚠️ Em modelos modernos (boosting, redes neurais), **pode ser contraproducente**, mas em modelos tabulares clássicos, **pode melhorar estabilidade e explicação**.

> 🎯 **Categorização é tão útil quanto perigosa. O segredo está na escolha do corte e no entendimento do problema.**


# 🧮 Continuação: Codificação e Outras Formas de Categorização de Variáveis

## 🔤 Codificação de Variáveis Categóricas

Após a **categorização** de variáveis contínuas ou discretas, é comum transformar as categorias em formatos numéricos para alimentar modelos de machine learning. As principais técnicas são:

---

### 1. One-Hot Encoding (Codificação Dummy)

- **Conceito:** Cria uma nova variável binária para cada categoria, indicando presença (1) ou ausência (0).
- **Exemplo:**
  
  Variável "Cor": {Vermelho, Azul, Verde}

  | Vermelho | Azul | Verde |
  |----------|-------|-------|
  | 1        | 0     | 0     |
  | 0        | 1     | 0     |
  | 0        | 0     | 1     |

- **Aplicações:**
  - Algoritmos que não lidam bem com variáveis ordinais (ex: regressão logística, SVM).
- **Vantagens:**
  - Não impõe ordem entre categorias.
- **Limitações:**
  - Pode gerar alta dimensionalidade (problema em variáveis com muitas categorias = *high cardinality*).
- **Ferramentas:**
  - `pandas.get_dummies()`
  - `sklearn.preprocessing.OneHotEncoder`

---

### 2. Label Encoding (Codificação Ordinal)

- **Conceito:** Atribui um número inteiro único para cada categoria.

  Ex: {Vermelho=0, Azul=1, Verde=2}

- **Aplicações:**
  - Quando a variável categórica é ordinal (ex: nível educacional).
  - Modelos baseados em árvores, que interpretam esses números apenas como rótulos.
- **Vantagens:**
  - Simples e eficiente em memória.
- **Limitações:**
  - Pode induzir ordens artificiais em variáveis nominais, distorcendo modelos lineares.
- **Ferramentas:**
  - `sklearn.preprocessing.LabelEncoder`

---

### 3. Target Encoding (Codificação por Média do Target)

- **Conceito:** Substitui cada categoria pela média da variável resposta associada (ex: média de inadimplência).

- **Aplicações:**
  - Quando há muitas categorias e relação forte com variável alvo.
- **Vantagens:**
  - Pode melhorar performance em modelos lineares e boosting.
- **Limitações:**
  - Risco de *data leakage* (vazamento de informação do target) se usado incorretamente.
- **Cuidados:**
  - Usar validação cruzada ou *out-of-fold* para cálculo.
- **Ferramentas:**
  - Bibliotecas como `category_encoders`

---

### 4. Binary Encoding / Hash Encoding

- Técnicas intermediárias que reduzem a dimensionalidade de One-Hot usando representações binárias ou hash.
- Úteis para variáveis com alto cardinalidade.

---

## 🔄 Outras Técnicas de Categorização de Variáveis Contínuas e Discretas

Além dos bins tradicionais e codificações, existem técnicas mais sofisticadas para lidar com variáveis numéricas:

### 1. **Binning Supervisionado (WOE Binning)**

- Descrito anteriormente, transforma variáveis em faixas baseadas no impacto no target, útil em scorecards.

### 2. **Transformações Baseadas em Estatísticas Locais**

- Ex: agrupamento por clusters k-means e transformação das variáveis contínuas em rótulos dos clusters.

### 3. **Discretização Dinâmica**

- Técnica que ajusta os bins conforme dados novos chegam (streaming).

### 4. **Embeddings para Categorias**

- Variáveis categóricas podem ser transformadas em vetores densos aprendidos via redes neurais (ex: embeddings em NLP).

- Aplicação em variáveis discretas de alta cardinalidade para modelos complexos.

---

## 🔧 Relação com Modelagem, Pré-processamento e Avaliação

| Técnica de Codificação / Categorização | Modelos Indicados | Vantagens | Limitações |
|---------------------------------------|-------------------|-----------|------------|
| One-Hot Encoding                      | Regressão, SVM, Redes Neurais | Intuitiva, não impõe ordem | Dimensionalidade alta |
| Label Encoding                       | Árvores, Random Forest | Simples, compacto | Pode induzir ordem errada |
| Target Encoding                     | GBDT, regressão linear | Explora relação com target | Risco de vazamento |
| Binning por Clusters                 | Modelos interpretáveis | Agrupamento baseado em estrutura | Complexidade computacional |
| Embeddings                         | Deep Learning | Captura relações complexas | Requer dados grandes |

## 🧠 Implicações Práticas

  - Sobreajuste
  - Viés de modelo
  - Dificuldade de interpretação

- **Categorização e codificação também influenciam a validade dos testes estatísticos e métricas de avaliação.**

## 📌 Conclusão

Categorização e codificação são passos **críticos** na pipeline de ciência de dados para transformar variáveis numéricas e categóricas em formatos adequados para modelagem. A escolha da técnica deve considerar:

- Natureza da variável (ordinal, nominal, alta cardinalidade)
- Tipo do modelo
- Tamanho e complexidade do dataset
- Necessidade de interpretabilidade

> ⚠️ Uma abordagem equivocada pode comprometer a qualidade e generalização do modelo, enquanto uma bem planejada pode maximizar o desempenho e facilitar a interpretação.
