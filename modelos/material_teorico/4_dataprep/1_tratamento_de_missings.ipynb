# 🧹 Tratamento de Dados Faltantes (Missing Values) em Ciência de Dados

Dados faltantes — também chamados de *missing values* ou *dados ausentes* — são uma realidade comum em praticamente todos os conjuntos de dados reais. Saber tratá-los de forma correta é **essencial para garantir qualidade, robustez e interpretabilidade** dos modelos estatísticos e preditivos.

---

## 🧠 Conceito Teórico

Um dado faltante ocorre quando **o valor de uma variável está ausente** para uma ou mais observações. Formalmente, dados faltantes são representados como `NaN`, `NULL`, ou valores ausentes específicos (ex: -999, 'NA').

### 🔎 Tipos de mecanismos de dados faltantes (Rubin, 1976)

1. **MCAR – Missing Completely At Random**  
   A ausência é completamente aleatória, **não depende de nenhuma variável observada ou não observada**.

2. **MAR – Missing At Random**  
   A ausência depende **apenas de variáveis observadas**.  
   Ex: Idade ausente em mulheres jovens, mas conhecida para outras faixas etárias.

3. **MNAR – Missing Not At Random**  
   A ausência depende **da própria variável faltante**.  
   Ex: Pessoas com renda mais alta não informam a própria renda.

> ⚠️ Importante: O tipo de mecanismo afeta diretamente **como podemos tratar os dados** sem introduzir viés.

---

## 🧾 Características Principais

- Preservar ou imputar os dados faltantes influencia:
  - **Distribuição estatística das variáveis**
  - **Correlação entre atributos**
  - **Performance e estabilidade do modelo**
- Algoritmos como árvores (XGBoost, LightGBM) toleram `NaN`; outros (KNN, SVM, redes neurais) **não**.

---

## 🧰 Estratégias Comuns de Tratamento

### 1. 🔍 Diagnóstico e Visualização

Antes de tratar, é fundamental **detectar e entender o padrão** de ausência.

Ferramentas:
- `df.isnull().sum()` (pandas)
- `missingno.matrix(df)` ou `missingno.heatmap(df)`
- `sns.heatmap(df.isnull())`

---

### 2. 🧼 Exclusão de Dados (Deleção)

#### a) Exclusão de Linhas (`listwise deletion`)
Remove as linhas com valores ausentes.

- ✅ Quando os dados estão **MCAR** e em **baixa proporção**.
- ⚠️ Pode introduzir **viés** ou **reduzir poder estatístico**.

#### b) Exclusão de Colunas
Remove variáveis com muitos `NaN`.

- ⚠️ Pode eliminar variáveis importantes para o modelo.

---

### 3. 🧩 Imputação de Valores

#### a) Imputação com Estatísticas Simples
- **Média**: `df['x'].fillna(df['x'].mean())`
- **Mediana**: menos sensível a outliers
- **Moda**: útil para variáveis categóricas

> 🔁 Preferível para **MAR** ou **MCAR** quando a ausência é pequena.

#### b) Imputação Condicional ou Avançada

- **KNN Imputer**: média dos vizinhos mais próximos
- **MICE (Multivariate Imputation by Chained Equations)**: regressão multivariada iterativa
- **MissForest**: florestas aleatórias para imputação
- **Regressão**: previsão do valor ausente com base nas outras variáveis

#### c) Imputação por Categorias
- "Desconhecido", "Sem Resposta", "Faltante"
- Cuidado: pode induzir **clusters artificiais**.

---

### 4. 🏷️ Criação de Flags de Missing

Adicionar variáveis binárias indicando a ausência:

```python
df['x_missing'] = df['x'].isnull().astype(int)
```

- Muito usado em **modelos de risco, crédito e saúde**.
- Ajuda o modelo a aprender com **a própria ausência como informação**.

---

### 5. 🧠 Modelagem Direta com NaNs (quando permitido)

Algoritmos como:
- **XGBoost / LightGBM / CatBoost** lidam nativamente com NaNs
- Árvores de decisão e Random Forest (em algumas implementações)

> Pode ser vantajoso manter os `NaN` quando a **ausência tem valor informativo**.

---

## 🎯 Aplicações Práticas em Projetos

| Exemplo | Aplicação | Estratégia comum |
|--------|-----------|------------------|
| Churn de clientes | Informações de perfil incompletas | Imputação com média + flag |
| Crédito bancário | Renda ou ocupação ausente | Modelo de imputação + variável indicadora |
| Saúde | Exames não realizados | Imputação condicional (MICE) |
| IoT / Sensores | Leituras ausentes em série temporal | Forward-fill, interpolação ou exclusão local |

---

## ⚠️ Diferenças vs Estratégias Relacionadas

| Conceito | Diferença Principal |
|---------|----------------------|
| Imputação x Deleção | Imputação **preserva registros**, deleção os remove |
| Imputação Simples x MICE | MICE considera **relações entre variáveis**, simples não |
| Imputar x Usar `NaN` no modelo | Nem todos os algoritmos aceitam `NaN` — depende do modelo |

---

## ✅ Vantagens e Armadilhas

### ✅ Vantagens de imputação:
- **Mantém volume de dados**
- Reduz viés de seleção
- Permite uso de algoritmos que exigem dados completos

### ⚠️ Armadilhas:
- Imputação mal feita pode **criar padrões artificiais**
- Ignorar o tipo de ausência (MCAR/MAR/MNAR) pode levar a **conclusões erradas**
- Variáveis criadas para indicar ausência podem ser **mal interpretadas** em modelos lineares

---

## 🧪 Impacto nos Modelos Estatísticos

- A **magnitude do erro** de imputação afeta:
  - **Bias** do modelo (erro sistemático)
  - **Variância** (instabilidade da predição)
  - **Inferência estatística** (valores de p incorretos)

- Dados faltantes são uma fonte **crítica de leakage** se tratados incorretamente antes da divisão treino/teste.

---

## 🔁 Relação com Modelagem e Avaliação

| Etapa | Impacto dos Missings |
|-------|-----------------------|
| Pré-processamento | Necessário para pipelines consistentes |
| Feature engineering | A ausência pode ser **informação útil** |
| Modelagem | Alguns algoritmos toleram `NaN`, outros não |
| Validação | Nunca imputar **após split** para evitar leakage |
| Interpretação | Erros de imputação podem afetar SHAP, coeficientes, etc |

---

## 🧠 Conclusão

O tratamento de missings deve ser **cuidadoso, contextual e estatisticamente informado**. A ausência de um dado pode carregar **tanta ou mais informação que a presença**, e sua manipulação impacta diretamente:

- Performance preditiva
- Robustez do modelo
- Confiabilidade da inferência
- Reprodutibilidade em produção

🔍 **Conhecer o porquê dos dados estarem faltando é tão importante quanto saber como imputá-los.**
