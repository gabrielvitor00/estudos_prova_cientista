# 🧠 PCA (Principal Component Analysis) – Análise de Componentes Principais



A **Análise de Componentes Principais (PCA)** é uma técnica estatística amplamente utilizada em ciência de dados para **redução de dimensionalidade**, compressão de dados, **exploração de variabilidade** e pré-processamento para modelagem.



---

## 📘 Conceito Teórico

O PCA transforma um conjunto de variáveis originais **possivelmente correlacionadas** em um novo conjunto de variáveis **não correlacionadas**, chamadas **componentes principais**.

Cada componente principal é uma **combinação linear das variáveis originais**, onde:

- O **primeiro componente** captura a **maior variância possível**.
- O **segundo componente** é ortogonal ao primeiro e captura a **segunda maior variância**, e assim por diante.

PCA é um método **não supervisionado** e **linear**, usado amplamente para explorar padrões e simplificar datasets com muitas variáveis.

---


## 🧾 Características Formais



### 🎯 Definições:



- Seja $X$ uma matriz de dados centrada com $n$ amostras e $p$ variáveis.

- O PCA encontra vetores ortogonais $\mathbf{w}_1, \mathbf{w}_2, \ldots, \mathbf{w}_p$ que maximizam a variância dos dados projetados:



  $$
  \mathbf{z}_1 = X \cdot \mathbf{w}_1 \quad \text{(com maior variância)}
  $$



- A matriz de covariância $S = \frac{1}{n-1}X^TX$ é diagonalizada para encontrar **autovalores** (variância explicada) e **autovetores** (direções principais).



### 📊 Variância Explicada:



- Cada componente explica uma **proporção da variância total**.



  $$
  \text{Variância explicada do componente } k = \frac{\lambda_k}{\sum_{j=1}^{p} \lambda_j}
  $$



---


## 🧮 Conceitos Relacionados Fundamentais

### 🎯 Colinearidade vs Multicolinearidade

- **Colinearidade**: duas variáveis altamente correlacionadas linearmente.
- **Multicolinearidade**: uma variável explicada como combinação linear de duas ou mais outras variáveis. Causa instabilidade em modelos como Regressão Linear.

✅ O PCA ajuda a eliminar **multicolinearidade** ao transformar variáveis correlacionadas em componentes **ortogonais (não correlacionadas)**.

## 🛠️ Aplicações Práticas em Ciência de Dados



| Aplicação                        | Descrição                                                                 |
|:----------------------------------|:-------------------------------------------------------------------------|
| 🌐 Redução de dimensionalidade    | Reduz colinearidade e melhora eficiência computacional                   |
| 🧪 Pré-processamento para modelos | Reduz ruído e melhora desempenho de algoritmos sensíveis à dimensionalidade |
| 🔍 Visualização de dados          | PCA 2D ou 3D para visualização de dados em alta dimensão                 |

---


## ✅ Vantagens



- 📉 Reduz dimensionalidade com **mínima perda de informação**
- 📐 Remove **correlação** entre variáveis
- ⚙️ Melhora desempenho e convergência de modelos
- 🔎 Facilita **interpretação visual** (ex: clusters)
- 📈 Útil para detectar padrões ocultos e redundância



---


## ⚠️ Limitações e Armadilhas



- ❌ Difícil interpretar os **componentes** (são combinações lineares)
- ⚠️ Requer que os dados sejam **padronizados** (especialmente se variáveis têm escalas diferentes)
- ❗ Pode **descartar variabilidade útil** (em busca de máxima variância)
- 📉 Ineficaz com dados não lineares (nesse caso, preferir Kernel PCA ou Autoencoders)



---


## 🔍 Interpretação Prática e Modelagem



### Quando usar PCA:



- Quando há **muitas variáveis correlacionadas**
- Antes de algoritmos como:



  - Regressão Linear
  - KNN (muito sensível a dimensionalidade)
  - SVM
- Quando variáveis têm **alta multicolinearidade**



### Quando NÃO usar PCA:



- Quando é essencial **interpretar os coeficientes originais**
- Com variáveis categóricas sem transformação adequada
- Com dados não padronizados (pode enviesar os componentes)

## 🧠 Técnicas Avançadas Relacionadas



- **Kernel PCA** – versão não linear do PCA
- **Sparse PCA** – aplica regularização para gerar componentes mais esparsos
- **Truncated SVD** – PCA otimizado para dados esparsos (usado em NLP, TF-IDF)
- **t-SNE / UMAP** – melhores para visualização, mas não interpretáveis como PCA
