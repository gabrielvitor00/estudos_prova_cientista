# ⚡️ LGBM Regressor – Regressão com Gradient Boosting de Alta Performance

O **LightGBM Regressor** é uma implementação de gradient boosting baseada em árvores, altamente otimizada para eficiência computacional, escalabilidade e acurácia. Desenvolvido pela Microsoft, o LightGBM tornou-se uma das ferramentas mais utilizadas em competições de machine learning e projetos de regressão complexos no mundo real.

---

## 📘 Conceito Teórico

O LightGBM Regressor pertence à classe de algoritmos **ensemble** baseados em **boosting**, especificamente **gradient boosting**, onde modelos fracos (árvores de decisão de baixa profundidade) são treinados **sequencialmente** para corrigir os erros residuais do modelo anterior.

O objetivo é minimizar uma função de perda (tipicamente o MSE, mas pode ser MAE, Huber, quantile, etc.) por meio de **gradientes negativos**, que indicam a direção da maior redução da perda.

O diferencial do LightGBM é que ele:

- Usa a estratégia de crescimento de árvore **leaf-wise** (com depth constraints)

- Utiliza **histogramas discretizados** para acelerar os cálculos e reduzir consumo de memória

- Suporta paralelização e GPU nativamente

---

## 🧱 Características Principais e Definições Formais

| Elemento                        | Descrição                                                                 |
|---------------------------------|---------------------------------------------------------------------------|
| **Boosting**                   | Aprendizado sequencial, onde cada modelo corrige os erros do anterior      |
| **Leaf-wise growth**           | A árvore cresce escolhendo o nó folha com maior ganho de informação        |
| **Histogram-based split**      | Reduz precisão das features para acelerar splits                          |
| **GOSS**                       | Gradient-based One Side Sampling – prioriza instâncias com maior gradiente |
| **Exclusive Feature Bundling (EFB)** | Agrupa features esparsas para reduzir dimensionalidade             |

### 🔣 Função Objetivo Genérica

$$
L(y, \hat{y}) = \sum_{i=1}^{n} \ell(y_i, F_M(x_i))
$$

Onde:

- $\ell$: função de perda (ex: MSE)

- $F_M(x)$: modelo final composto por $M$ árvores

## 🔄 Diferenças com Modelos Semelhantes

| Modelo              | Diferença-chave                                                                |
|---------------------|-------------------------------------------------------------------------------|
| **XGBoost**         | Mais estável com regularização forte, mas mais lento                          |
| **CatBoost**        | Suporte nativo superior para categóricas, mas mais pesado computacionalmente  |
| **Random Forest**   | Usa bagging, não boosting – mais robusto, mas menos preciso em geral          |
| **Regressão Linear**| Assume linearidade, enquanto LGBM modela relações complexas                   |

## ⚠️ Limitações e Armadilhas Comuns

| Limitação         | Descrição                                                                   |
|-------------------|----------------------------------------------------------------------------|
| **Overfitting**   | Leaf-wise + profundidade alta pode levar a sobreajuste                      |
| **Tuning sensível**| Muitos hiperparâmetros interdependentes (learning_rate, num_leaves, etc.)  |
| **Exigência de validação** | Necessário monitorar early stopping ou usar validação cruzada           |

---

## 🧠 Interpretação Prática

Embora o LightGBM seja um modelo "caixa preta", existem formas de interpretá-lo:

- **Importância das features**: baseada no ganho cumulativo ou número de splits

- **SHAP Values**: permitem explicar localmente as predições

- **Partial Dependence Plots (PDPs)**: mostram efeito médio de uma feature

## 🛠️ Pré-processamento, Modelagem e Validação

- **Codificação**: suporta `categorical_feature` diretamente

- **Normalização**: não é necessária

- **Validação**: usar `early_stopping_rounds` com `eval_set`

- **Hiperparâmetros**:

  - `learning_rate`: passo do gradiente – menor → melhor generalização

  - `num_leaves`: número de folhas por árvore – mais folhas → mais complexidade

  - `max_depth`: profundidade máxima da árvore – evita overfitting

  - `min_data_in_leaf`: regularização para evitar folhas com poucos dados
