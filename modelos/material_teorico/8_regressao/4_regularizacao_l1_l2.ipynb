# 🧮 Regularização L1 e L2 – Fundamentos Avançados para Ciência de Dados

A regularização L1 e L2 são técnicas fundamentais para **evitar overfitting** e **controlar a complexidade** de modelos estatísticos e de machine learning. Elas introduzem **penalidades na função de custo**, forçando o modelo a aprender padrões mais generalizáveis e robustos, especialmente em contextos de **alta dimensionalidade** ou **colinearidade**.

---

## 📘 Conceito Teórico

Regularização é o processo de adicionar um **termo de penalização** à função de perda de um modelo, com o objetivo de **limitar os valores dos coeficientes** (pesos) aprendidos. A ideia central é **reduzir a variância sem aumentar excessivamente o viés**.

Duas formas principais de regularização:

- **L1 (Lasso)**: soma dos valores absolutos dos coeficientes
- **L2 (Ridge)**: soma dos quadrados dos coeficientes

Ambas alteram a **função objetivo** de regressão ou classificação, adicionando um termo de penalidade aos coeficientes.

---

## 📐 Definições Formais

### 🔢 Regressão Linear Clássica:

$$
J(\beta) = \frac{1}{n} \sum_{i=1}^{n} (y_i - X_i \cdot \beta)^2
$$

### 🔒 Regularização L2 (Ridge):

$$
J_{L2}(\beta) = \frac{1}{n} \sum_{i=1}^{n} (y_i - X_i \cdot \beta)^2 + \lambda \sum_{j=1}^{p} \beta_j^2
$$

- Penaliza grandes coeficientes suavemente
- **Não zera** os coeficientes
- Preserva todas as variáveis

### 🔓 Regularização L1 (Lasso):

$$
J_{L1}(\beta) = \frac{1}{n} \sum_{i=1}^{n} (y_i - X_i \cdot \beta)^2 + \lambda \sum_{j=1}^{p} |\beta_j|
$$

- **Promove esparsidade**: zera coeficientes irrelevantes
- Executa **seleção de variáveis** implicitamente

---

## 🧪 Aplicações Práticas em Ciência de Dados

| Cenário                        | Aplicação                                                                      |
|-------------------------------|--------------------------------------------------------------------------------|
| Dados com alta dimensionalidade| Reduz overfitting em datasets com muitos atributos (ex: genômica, texto)       |
| Multicolinearidade            | Ridge controla instabilidade causada por variáveis correlacionadas              |
| Seleção de variáveis          | Lasso é útil quando se deseja identificar as features mais relevantes           |
| Engenharia de features        | L1 pode ser usada como filtro de feature selection automático                   |
| Modelos lineares (reg ou class)| Logistic Regression + L1/L2 para classificação regularizada                    |

---

## 🔄 Diferenças entre L1 e L2

| Aspecto                 | L1 (Lasso)                          | L2 (Ridge)                          |
|-------------------------|-------------------------------------|-------------------------------------|
| Penalização             | Soma dos valores absolutos          | Soma dos quadrados                  |
| Efeito nos coeficientes | Pode zerar coeficientes             | Apenas encolhe coeficientes         |
| Seleção de variáveis    | Sim (explícita)                     | Não (mantém todas)                  |
| Solução                 | Não diferenciável em zero ($|\beta|$) | Diferenciável (convexo e suave)     |
| Preferível quando       | Muitas variáveis irrelevantes        | Muitas variáveis correlacionadas     |
| Interpretação           | Modelo mais simples e interpretável  | Modelo mais estável e robusto        |

---

## 📈 Implicações para Modelos Estatísticos e Preditivos

- Controlam **complexidade do modelo** → evitam overfitting
- L1 pode **remover ruído** e **aumentar interpretabilidade**
- L2 pode **melhorar estabilidade numérica**
- Crucial para **regressão logística**, **redes neurais**, **SVMs** e **modelos lineares em geral**
- Em problemas multivariados, ajudam a evitar **explosão de variância** em pequenas amostras

## ⚠️ Vantagens e Limitações

| Vantagem                              | Limitação                                         |
|---------------------------------------|---------------------------------------------------|
| Reduz overfitting                     | Exige tuning fino do hiperparâmetro $\lambda$     |
| Aumenta generalização                 | Lasso pode ser instável com features correlacionadas|
| Ajuda em seleção de features (L1)     | Ridge não faz seleção explícita (L2)               |
| Facilmente integrável a modelos lineares| L1 é computacionalmente mais difícil (não diferenciável) |

### 🧠 Regularização na Prática: Viés vs Variância

| $\lambda$ pequeno | $\lambda$ intermediário | $\lambda$ grande |
|-------------------|------------------------|------------------|
| Pouca regularização → overfitting | Bom equilíbrio | Underfitting: modelo muito simples |
| Alta variância | Baixo viés e variância | Alto viés, baixa variância |

### 🧬 Hiperparâmetro $\lambda$
- Controla a intensidade da penalização
- Deve ser ajustado com validação cruzada
- Pode ser otimizado com GridSearch, RandomizedSearch ou Optuna

## 🧩 Como a Regularização Afeta Modelos: Compreensão Conceitual e Prática

### 🔍 Intuição Geométrica e Estatística

Quando aplicamos regularização, estamos alterando o espaço de busca da solução ótima, **forçando o modelo a preferir soluções mais simples ou estáveis**. A penalização nos coeficientes atua como uma **restrição geométrica** sobre os pesos que o modelo pode assumir:

- **L2 (Ridge)** impõe uma restrição esférica (bola Euclidiana): tende a **encolher os coeficientes** para perto de zero, mas raramente os zera.
- **L1 (Lasso)** impõe uma restrição em formato de diamante (norma Manhattan): **cria cantos** no espaço de solução onde alguns coeficientes se tornam exatamente zero → **sparsidade**.

Essa geometria influencia diretamente a solução do modelo:

- Ridge: mais suave e estável
- Lasso: mais agressivo, com potencial de excluir variáveis

---

### 🧠 O que Acontece na Prática com o Modelo?

| Efeito Prático da Regularização         | Implicação para o Modelo                       |
|:----------------------------------------|:-----------------------------------------------|
| Coeficientes menores                    | Modelo mais conservador, evita overfitting     |
| Redução da variância                    | Melhora a generalização para novos dados       |
| Eliminação de variáveis irrelevantes (L1)| Automatiza seleção de features                 |
| Solução mais estável (L2)               | Menos sensível a pequenas variações no dado    |
| Piora da acurácia em treino             | Esperada e necessária para melhorar em teste   |
| Evita explosão de coeficientes          | Crucial em dados colineares ou mal escalados   |

---

### 📉 Interpretação Estatística

A regularização L2 pode ser interpretada sob a ótica bayesiana como a **inclusão de uma prior normal com média zero** nos coeficientes (regularização bayesiana). Já a L1 equivale a uma **prior Laplaciana**, que favorece coeficientes exatamente zero.

| Tipo de Regularização | Prior Bayesiana (Interpretação Probabilística) |
|----------------------|------------------------------------------------|
| L2 (Ridge)           | Distribuição normal centrada em 0              |
| L1 (Lasso)           | Distribuição Laplaciana centrada em 0          |

---

### 🧮 Efeitos em Função da Magnitude de $\lambda$

- $\lambda \to 0$: sem penalização, o modelo tende a overfitting
- $\lambda$ moderado: equilíbrio ideal entre erro de treino e generalização
- $\lambda$ muito alto: underfitting severo, o modelo ignora padrões reais

**Visualmente**, $\lambda$ controla o **tamanho do espaço onde buscamos a solução**.

---

### 🧠 Exemplo Numérico (Intuitivo):

Imagine uma regressão linear simples:

$$
y = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \beta_3 x_3
$$

Suponha que:

- $x_1$ é muito relevante
- $x_2$ é moderadamente relevante
- $x_3$ é puro ruído

Sem regularização:
- O modelo pode dar pesos altos até para $x_3$ para ajustar ruído

Com **L1**:
- O modelo provavelmente zera o coeficiente de $x_3$

Com **L2**:
- O modelo ainda usa $x_3$, mas com peso muito pequeno

| Modelos com milhares de variáveis| Regularização é **obrigatória** para evitar overfitting         |
