# 📈 Regressão Linear – Guia Ultra-Robusto para Ciência de Dados

A **regressão linear** é uma das técnicas mais clássicas e fundamentais em estatística e ciência de dados, utilizada para **modelar relações entre variáveis contínuas**. Apesar de sua simplicidade, continua sendo uma ferramenta poderosa para **interpretação, inferência e previsão** em contextos reais.

---

## 📘 Conceito Teórico

A regressão linear busca modelar a relação entre:

- Uma **variável dependente** $y$ (output, contínua)
- Uma ou mais **variáveis independentes** $x_1, x_2, ..., x_p$ (features)

A ideia central é ajustar uma **função linear** que minimize o erro de predição de $y$ dado $X$:

$$
y = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \ldots + \beta_p x_p + \epsilon
$$

Onde:

- $\beta_0$ é o intercepto
- $\beta_i$ são os coeficientes de regressão
- $\epsilon \sim N(0, \sigma^2)$ é o erro aleatório (resíduo)

---

## 🔧 Características Principais e Definições Formais

| Componente                      | Descrição                                                                 |
|---------------------------------|---------------------------------------------------------------------------|
| **Linearidade**                 | A relação entre preditores e resposta é **linear nos parâmetros**         |
| **Erro quadrático médio (MSE)** | Métrica típica de otimização: minimiza $\sum (y_i - \hat{y}_i)^2$        |
| **Estimadores OLS**             | Coeficientes obtidos por **Mínimos Quadrados Ordinários (Ordinary Least Squares)** |
| **R² (Coeficiente de Determinação)** | Percentual da variância de $y$ explicado pelo modelo                  |
| **Assunções**                   | Linearidade, independência, homocedasticidade, normalidade dos resíduos   |

---

## 🔬 Aplicações Reais em Ciência de Dados

| Setor         | Problema                                                        |
|---------------|-----------------------------------------------------------------|
| Finanças      | Previsão de lucro, custo, retorno sobre investimento            |
| Saúde         | Estimativa de tempo de internação, custo de cirurgia            |
| Varejo        | Modelagem de demanda com base em preço, campanhas, sazonalidade |
| Logística     | Previsão de tempo de entrega ou consumo de combustível          |
| Mídia digital | Estimar cliques ou impressões com base em variáveis de campanha |

---

## 🧠 Diferenças com Modelos Correlatos

| Modelo                      | Diferença-chave                                                                         |
|-----------------------------|----------------------------------------------------------------------------------------|
| **Regressão Logística**     | Output binário (classificação), usa função logística                                    |
| **Regressão Polinomial**    | Adiciona termos não-lineares (ex: $x^2, x^3$), mas ainda linear nos parâmetros          |
| **Árvores de Regressão**    | Captura relações não-lineares sem pressupostos estatísticos                             |
| **Lasso/Ridge/ElasticNet**  | Regularizações da regressão linear para lidar com overfitting e multicolinearidade      |

---

## ✅ Vantagens

- Altamente interpretável (coeficientes indicam direção e magnitude do efeito)
- Rápida de treinar, mesmo com grandes datasets
- Permite inferência estatística (intervalos, testes de hipótese)
- Fácil integração com pipelines de engenharia de atributos

---

## ⚠️ Limitações e Armadilhas

| Limitação                         | Detalhes técnicos                                                                |
|-----------------------------------|----------------------------------------------------------------------------------|
| **Multicolinearidade**            | Preditores altamente correlacionados distorcem estimativas dos $\beta_i$         |
| **Heterocedasticidade**           | Variância dos resíduos não é constante → viola assunções → erros não confiáveis   |
| **Outliers e pontos influentes**  | Impactam fortemente a inclinação da reta → considerar robustez ou remoção         |
| **Relações não lineares**         | Não modela bem dados com curvatura ou interação complexa                          |

---

## 🔍 Interpretação Prática

- Cada coeficiente $\beta_i$ representa a **variação esperada em $y$** para **uma unidade de aumento em $x_i$**, mantendo os outros $x_j$ constantes.
- O intercepto $\beta_0$ é o valor esperado de $y$ quando todos os $x_i = 0$.
- O R² indica o **grau de explicação** do modelo: $R^2 = 1$ significa predição perfeita, $R^2 = 0$ significa nenhuma explicação.

## 📊 Visualizações e Ferramentas Comuns

|------------------------------------|----------------------------------------------------------------------------|

---

## 🛠️ Integração com Pré-processamento e Modelagem

- **Normalização ou padronização** (ex: `StandardScaler`) pode ser útil se variáveis estiverem em escalas muito diferentes
- **Seleção de variáveis:** análise de significância (p-valor), VIF (Variance Inflation Factor), `SelectKBest`, etc.
- **Interações:** criar variáveis como $x_1 \cdot x_2$, $x^2$, etc., para capturar efeitos não lineares
- **Regularização:** Lasso (L1) ou Ridge (L2) se houver risco de overfitting ou muitas variáveis

---

## 📐 Exemplo Numérico com sklearn

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
import pandas as pd

# Dataset simples
df = pd.DataFrame({
    'idade': [22, 25, 30, 45, 50],
    'salario': [2000, 2500, 4000, 9000, 11000]
})

X = df[['idade']]
y = df['salario']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = LinearRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
print("Coef:", model.coef_)
print("Intercepto:", model.intercept_)
print("R²:", r2_score(y_test, y_pred))
```

---

## 🧪 Avaliação de Modelo

| Métrica         | Descrição                                                        |
|-----------------|------------------------------------------------------------------|
| **MSE / RMSE**  | Penaliza erros grandes fortemente, comum em treino/teste         |
| **MAE**         | Erro absoluto médio, mais robusto a outliers                     |
| **R² Score**    | Explica quanta variabilidade do target foi explicada             |
| **Adjusted R²** | Penaliza inclusão de variáveis irrelevantes                      |
