# 📐 Modelos Lineares Generalizados (GLM) – Guia Ultra-Robusto para Ciência de Dados

Os **Modelos Lineares Generalizados (GLMs)** são uma extensão dos modelos lineares tradicionais que permitem lidar com **distribuições não normais** da variável resposta, tornando-se extremamente úteis em problemas de regressão além do domínio contínuo-gaussiano.

---

## 📘 Conceito Teórico

GLMs generalizam a regressão linear clássica ao introduzir:

1. **Função de ligação (link function)** que conecta a **média da variável resposta** ao **previsor linear**.
2. **Distribuições da família exponencial** como base para modelar a variável dependente.

**Forma geral:**

$$
g(\mu_i) = \eta_i = \beta_0 + \beta_1 x_{i1} + \dots + \beta_p x_{ip}
$$

Onde:

- $\mu_i = \mathbb{E}(Y_i)$: média da variável resposta

- $g(\cdot)$: função de ligação

- $\eta_i$: preditor linear

- $Y_i \sim \text{Distribuição da família exponencial}$

---

## 🧬 Características Principais

| Componente              | Descrição                                                                |
|-------------------------|--------------------------------------------------------------------------|
| **Distribuição**        | A variável resposta segue uma **distribuição da família exponencial**    |
| **Link function**       | Função que transforma a média da resposta para o espaço linear           |
| **Previsor linear**     | Combinação linear dos preditores $\beta^\top X$                        |
| **Estimativa por MLE**  | Os parâmetros são ajustados via **Máxima Verossimilhança**               |
| **Família exponencial** | Inclui Normal, Binomial, Poisson, Gamma, Inversa-Gaussiana etc.         |

---

## 🧪 Distribuições Típicas e Casos de Uso

| Distribuição   | Variável resposta | Função de ligação                        | Aplicação típica                        |
|---------------|-------------------|------------------------------------------|-----------------------------------------|
| Normal        | Contínua          | Identidade $g(\mu)=\mu$                 | Regressão linear                        |
| Binomial      | Binária           | Logit $g(\mu)=\log(\mu/(1-\mu))$       | Regressão logística                     |
| Poisson       | Contagem          | Log $g(\mu)=\log(\mu)$                  | Ocorrência de eventos (ex: chamadas)    |
| Gamma         | Contínua $> 0$    | Inversa $g(\mu)=1/\mu$                  | Modelos de tempo/custo com assimetria   |
| Inversa Gauss.| Contínua $> 0$    | Inversa quadrada                         | Modelos de tempo em processos físicos   |

---

## 📈 Aplicações Reais em Ciência de Dados

- **Poisson GLM** para prever número de compras por cliente

- **Logístico (Binomial)** para churn, ad-click, fraude

- **Gamma GLM** para modelar valores de sinistros em seguros

- **GLMs customizados** em modelagem atuarial, medicina, engenharia de confiabilidade

GLMs são amplamente usados em setores regulados (seguros, saúde, governo) por sua **transparência, interpretabilidade e solidez estatística**.

---

## 🔍 Diferenças vs Conceitos Relacionados

| Modelo                      | Diferença principal                                            |
|-----------------------------|---------------------------------------------------------------|
| Regressão Linear            | Supõe erro normal com variância constante                      |
| Regressão Logística         | Caso específico de GLM com distribuição binomial               |
| Rede Neural                 | Flexível, mas não garante interpretabilidade ou estrutura      |
| XGBoost / LGBM              | Ensemble não paramétrico, melhor preditivo, menos interpretável|
| Modelos Lineares Ridge/Lasso| GLM pode usar penalizações, mas tem foco estatístico rígido    |

---

## ✅ Vantagens

- 📏 **Modelagem direta da distribuição da resposta** (sem precisar transformar a variável)

- 🧠 **Interpretável**: os coeficientes mantêm interpretação clara (ex: log-odds)

- 🔬 **Rigor estatístico**: uso de teoria de estimação via máxima verossimilhança

- 🧱 **Modular**: permite encaixar diferentes famílias/link functions com facilidade

- 🔄 Suporte nativo a técnicas de regularização (GLM + L1/L2)

---

## ⚠️ Limitações e Armadilhas

- ❌ Assume **independência entre observações**

- ❌ Pode falhar com dados ruidosos ou altamente não-lineares

- 📉 Requer especificação correta da **distribuição e da função de ligação**

- 🚫 Sensível a multicolinearidade (como qualquer modelo linear)

- ⚠️ Diagnóstico estatístico exige atenção: resíduos, deviance, overdispersion

---

## 🧠 Interpretação Prática e Implicações

- Coeficiente $\beta_j$ representa:

  - Em link identidade: variação esperada em $Y$ por unidade de $x_j$

  - Em logit: efeito multiplicativo nos odds

  - Em log: efeito percentual esperado

### Importante:

A função de ligação **transforma a escala da média da resposta**, mas não transforma os dados diretamente.

---

## 🧮 Pré-processamento e Modelagem

| Etapa                  | Relevância para GLM                                                    |
|------------------------|-----------------------------------------------------------------------|
| Escalonamento          | Recomenda-se para melhorar convergência em MLE                        |
| Codificação categórica | Usar one-hot; modelos lineares não tratam variáveis categóricas nativas|
| Feature Engineering    | Termos quadráticos ou interações devem ser explicitamente modelados    |
| Regularização          | Pode-se aplicar L1 (Lasso) ou L2 (Ridge) para controle de overfitting |

---

## 🔍 Diagnóstico e Validação

- 📉 **Resíduos deviance e Pearson**

- 📊 **Gráfico de resíduos vs preditores** (verifica forma funcional)

- 🔎 **Teste de overdispersion** em Poisson e Binomial

- 🔧 **Pseudo-$R^2$** e AIC/BIC como métricas comparativas

## 🔄 GLM + Regularização

GLMs podem ser estendidos com **penalidades (L1, L2)** para reduzir sobreajuste:

$$
\text{Loss} = -\log L(\beta) + \lambda \cdot \|\beta\|_p
$$

- **L1** → Sparsidade (feature selection automática)

- **L2** → Coeficientes suavizados

- **ElasticNet** → Combina ambas

Modelos Lineares Generalizados são **fundamentais para ciência de dados aplicada com fundamentos estatísticos rigorosos**. Eles permitem modelar variáveis respostas com distribuições diversas, mantendo **transparência, explicabilidade e robustez inferencial**.
