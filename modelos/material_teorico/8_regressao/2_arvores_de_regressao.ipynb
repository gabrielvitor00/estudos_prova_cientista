# 🌲 Árvore de Regressão – Guia Ultra-Robusto para Ciência de Dados

As **árvores de regressão** são modelos de aprendizado supervisionado utilizados para prever **variáveis contínuas**, estruturando a tomada de decisão por meio de **divisões recursivas** dos dados com base em critérios de minimização de erro. São altamente interpretáveis e aplicáveis a uma ampla gama de problemas de regressão.

---

## 📘 Conceito Teórico

A árvore de regressão é uma **estrutura hierárquica** onde cada **nó interno** representa uma divisão (split) em uma variável de entrada (feature), cada **ramo** representa um resultado de teste, e cada **folha** representa uma **predição numérica** (geralmente a média dos valores observados naquele subconjunto).

Essas árvores são construídas **recursivamente** usando o critério de divisão que **minimiza o erro de predição** (geralmente o **Erro Quadrático Médio** – MSE).

---

## 🔧 Características Principais e Definições Formais

| Elemento                   | Descrição                                                                 |
|----------------------------|---------------------------------------------------------------------------|
| **Divisão (Split)**        | Regras binárias do tipo $x \leq t$ para segmentar os dados                |
| **Critério de divisão**    | Minimiza uma função de perda (ex: MSE, MAE, Poisson deviance)             |
| **Folha (Leaf)**           | Predição = média (ou mediana) dos valores da variável target              |
| **Árvore Recursiva**       | Construída top-down (divisão até critério de parada)                      |
| **Profundidade (Depth)**   | Nível máximo de ramificações da árvore                                    |
| **Poda (Pruning)**         | Técnica para remover ramos que não melhoram generalização                 |

---

## 🎯 Critérios de Divisão – Funções de Perda

### 🔹 Erro Quadrático Médio (MSE)

$$
\text{MSE} = \frac{1}{N} \sum_{i=1}^N (y_i - \bar{y})^2
$$

- Penaliza severamente outliers.
- Otimiza a média como predição.
- Reforça o ajuste em regiões com alta variância.

### 🔹 Erro Absoluto Médio (MAE)

$$
\text{MAE} = \frac{1}{N} \sum_{i=1}^N |y_i - \text{med}(y)|
$$

- Mais robusto a outliers.
- Otimiza a mediana como predição.
- Pode gerar árvores menos instáveis em presença de ruído assimétrico.

### 🔹 Desvio de Poisson (Poisson Deviance)

- Usado em tarefas onde a variável resposta é uma contagem.
- Penaliza desvios em escala logarítmica.
- Frequentemente usada com perda:

$$
L(y, \hat{y}) = \hat{y} - y \cdot \log(\hat{y})
$$

### 🧠 Como a divisão é escolhida?

Para cada possível split, a árvore avalia o ganho de informação:

$$
\Delta \text{Loss} = L_{pai} - \left(\frac{n_{esq}}{n} L_{esq} + \frac{n_{dir}}{n} L_{dir}\right)
$$

## 🔍 Diferenças com Modelos Correlacionados

| Modelo                      | Diferença-chave                                                                 |
|-----------------------------|--------------------------------------------------------------------------------|
| **Árvore de Classificação** | Output é categórico, usa critérios como Gini/Entropia                          |
| **Regressão Linear**        | Assume linearidade e distribuições normais                                     |
| **Random Forest Regressor** | Conjunto de árvores para maior robustez e menor variância                      |
| **Gradient Boosting**       | Árvore combinada sequencialmente para reduzir viés                             |

---

## ✅ Vantagens

- Não requer suposições sobre a distribuição dos dados
- Fácil de interpretar e visualizar
- Capta interações e relações não-lineares
- Naturalmente realiza seleção de variáveis
- Suporta variáveis categóricas e contínuas

---

## ⚠️ Limitações e Armadilhas Comuns

| Problema                   | Descrição                                                                 |
|----------------------------|---------------------------------------------------------------------------|
| **Overfitting**            | Árvores profundas se ajustam demais aos dados de treino                    |
| **Instabilidade**          | Pequenas mudanças nos dados geram árvores muito diferentes                 |
| **Divisões não suaves**    | Predições são feitas em "degraus" (piecewise constant)                    |
| **Predições ruidosas**     | Ruído nos dados pode levar a divisões irrelevantes                         |

➡️ **Soluções**: Poda (pruning), limitar profundidade, usar ensemble (Random Forest/Boosting)

---

## 🎯 Viés e Variância

| Aspecto            | Explicação                                                                 |
|--------------------|----------------------------------------------------------------------------|
| **Viés**           | Baixo. Árvores conseguem modelar relações não lineares com boa precisão.   |
| **Variância**      | Alta. Mudanças pequenas nos dados causam grandes mudanças na árvore.       |
| **Trade-off**      | Árvores simples têm viés maior e variância menor; árvores profundas, o oposto |

➡️ Estratégias como Random Forest e Bagging reduzem a variância sem aumentar viés.

---

## 🧠 Interpretação Prática e Implicações

- As divisões mostram **quais variáveis são mais importantes** na explicação da variabilidade de $y$
- As folhas retornam a **média local** da variável target (ou mediana, dependendo da perda)
- Permite **interpretabilidade local** (ex: por que a predição foi 102?)

### Exemplo de interpretação:
> Se `preço < R$30` e `número de avaliações > 100`, então valor predito da venda é R$450.
