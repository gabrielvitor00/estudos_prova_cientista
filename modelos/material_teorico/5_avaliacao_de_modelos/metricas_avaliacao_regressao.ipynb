# 📉 Métricas de Avaliação de Modelos de Regressão — Explicações Avançadas

Modelos de regressão predizem valores **contínuos**. Avaliar corretamente essas predições exige escolher **métricas de erro e desempenho** que reflitam diferentes aspectos de qualidade: erro absoluto, erro quadrático, variabilidade explicada, robustez a outliers e estabilidade numérica.

---

## ⚙️ Fundamentos: Erros de Previsão

Antes das métricas, defina:

- $y_i$: valor real do $i$-ésimo exemplo

- $\hat{y}_i$: valor predito pelo modelo

- $n$: número de observações

- $\bar{y}$: média dos valores reais

O **erro de predição** é:

$$
e_i = y_i - \hat{y}_i
$$

---

## 📐 1. MAE — Mean Absolute Error (Erro Absoluto Médio)

**Conceito:**

Média das diferenças absolutas entre os valores reais e preditos:

$$
\text{MAE} = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|
$$

**Interpretação:**

- Mede o erro médio real cometido pelo modelo.

- Está na mesma unidade da variável resposta.

**Vantagens:**

- Robusto a outliers (mais que RMSE).

- Fácil de interpretar: "erro médio de X unidades".

**Limitações:**

- Não penaliza fortemente grandes erros.

- Não diferenciável em zero (impacta alguns otimizadores).

**Aplicação prática:**

- Use quando o erro médio é o foco (ex: previsão de vendas, demanda, consumo).

---

## 📐 2. MSE — Mean Squared Error (Erro Quadrático Médio)

**Conceito:**

Média dos quadrados dos erros:

$$
\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$

**Interpretação:**

- Penaliza fortemente erros grandes devido ao quadrado.

**Vantagens:**

- Diferenciável e suavemente derivável (ótimo para otimização).
- Função de perda padrão em regressão linear e redes neurais.

**Limitações:**

- Sensível a outliers.
- Valor na unidade ao quadrado da variável-alvo.

**Aplicação prática:**

- Ideal quando grandes erros devem ser fortemente penalizados.

---

## 📐 3. RMSE — Root Mean Squared Error (Raiz do Erro Quadrático Médio)

**Conceito:**

Raiz quadrada do MSE:

$$
\text{RMSE} = \sqrt{ \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 }
$$

**Interpretação:**

- Erro médio ponderado (com ênfase em erros grandes), com unidade da variável.

**Vantagens:**

- Mais interpretável que o MSE (mesma unidade de $y$).
- Útil quando grandes erros são inaceitáveis.

**Limitações:**

- Muito influenciado por outliers.
- Menos robusto que MAE.

**Aplicação prática:**

- Modelos onde erros grandes impactam diretamente o negócio (ex: estimativa de custo, perdas).

---

## 📐 4. $R^2$ — Coeficiente de Determinação

**Conceito:**

Proporção da variância de $y$ explicada pelo modelo:

$$
R^2 = 1 - \frac{ \sum (y_i - \hat{y}_i)^2 }{ \sum (y_i - \bar{y})^2 }
$$

**Interpretação:**

- $R^2 = 1$: predição perfeita.
- $R^2 = 0$: o modelo não melhora em nada em relação à média.
- Pode ser negativo se o modelo for pior que a média.

**Vantagens:**

- Interpretação estatística direta.
- Compara com modelo nulo (só a média).

**Limitações:**

- Não indica se o modelo é bom ou ruim isoladamente.
- Pode ser enganoso com overfitting (modelo muito complexo).

**Aplicação prática:**
- Indica quão bem o modelo captura padrões gerais dos dados.

---

## 📐 5. $R^2$ Ajustado — R² Ajustado

**Conceito:**

Versão penalizada do $R^2$ que considera o número de variáveis independentes:

$$
R^2_{ajustado} = 1 - \left( \frac{(1 - R^2)(n - 1)}{n - p - 1} \right)
$$

Onde:
- $p$ é o número de preditores

**Interpretação:**

- Evita inflar o $R^2$ artificialmente com variáveis irrelevantes.

**Vantagens:**

- Mais confiável em modelos com muitas features.
- Permite comparar modelos de complexidade diferente.

---


## 📐 6. MAPE — Mean Absolute Percentage Error

**Conceito:**

Erro percentual médio absoluto:

$$
\text{MAPE} = \frac{100\%}{n} \sum_{i=1}^{n} \left| \frac{y_i - \hat{y}_i}{y_i} \right|
$$

**Interpretação:**
- Mostra o erro relativo (%) em relação ao valor verdadeiro.

**Vantagens:**
- Escala independente da unidade (fácil de comparar entre datasets).

**Limitações:**
- Instável para valores reais próximos de zero.
- Pode superestimar erro em valores pequenos.

**Aplicação prática:**
- Ideal em forecasting financeiro, demanda, séries temporais.

## 📊 Visualizações Comuns

- **Gráfico de resíduos:** $\hat{y}$ vs $y - \hat{y}$

- **QQ-plot dos resíduos** (normalidade)

- **Distribuição dos erros (histogramas)**

- **Scatter plot** entre $y$ e $\hat{y}$

- **Boxplot de erros por grupo** (ex: por região, faixa etária etc.)

## ⚠️ Armadilhas e Boas Práticas

- **Não use $R^2$ isoladamente**: pode ser alto mesmo com erros grandes.

- **Evite MAPE com zeros ou valores pequenos**: distorce o erro.

- **MAE e RMSE não devem ser comparados entre datasets com escalas diferentes.**

- **Considere o domínio do problema**: erro percentual vs erro absoluto pode implicar decisões diferentes.

---

## ✅ Conclusão

A escolha da métrica correta depende de:

- **Objetivo do modelo** (minimizar erro, explicar variância, prever com precisão relativa)

- **Impacto dos erros grandes** no negócio

- **Formato e escala dos dados**

- **Compreensão estatística e operacional da equipe**

🎯 Avaliar modelos de regressão exige mais do que uma métrica: o cientista de dados deve **analisar múltiplas perspectivas** de erro e interpretar suas **implicações preditivas e estratégicas**.
