# 🧪 Métricas de Avaliação de Modelos de Classificação — Detalhamento Profundo

Avaliar corretamente um modelo de classificação é uma etapa crítica em **ciência de dados aplicada**, especialmente quando o objetivo envolve tomada de decisão baseada em risco, retorno financeiro ou impacto humano. Aqui detalhamos as principais métricas de avaliação com profundidade.

---

## 🎯 1. Acurácia (Accuracy)

### 📘 Conceito:
Proporção de predições corretas (positivas e negativas) entre todas as predições feitas.

$$
\text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN}
$$

### 💡 Interpretação:
Indica o quão "frequentemente" o modelo acerta. **Útil em problemas com classes balanceadas.**

### 📉 Exemplo prático:
Se 95% dos dados são da classe 0 e o modelo prevê sempre 0, a acurácia será 95% — **mas o modelo é inútil**.

### ⚠️ Limitações:
- **Altamente enganosa em datasets desbalanceados.**
- **Ignora o custo dos erros (FP vs FN).**

### ✅ Use quando:
- Classes estiverem **bem balanceadas**.
- Erros **FP e FN tiverem o mesmo peso**.

---

## 🔍 2. Precision (Precisão)

### 📘 Conceito:
Entre as instâncias **classificadas como positivas**, quantas **realmente são positivas**?

$$
\text{Precision} = \frac{TP}{TP + FP}
$$

### 💡 Interpretação:
Avalia o grau de "confiança" ao **afirmar positividade**. Alta precisão = **baixo FP**.

### 📉 Exemplo prático:
Em detecção de fraude, alta precisão = a maioria dos alertas são realmente fraudes.

### ⚠️ Limitações:
- **Ignora FN**: pode ser alta mesmo se perder muitos casos positivos.

### ✅ Use quando:
- Custo de **falsos positivos é alto** (ex: detecção de câncer → menos falsos alarmes).

---

## 🔍 3. Recall (Sensibilidade / TPR)

### 📘 Conceito:
Entre todas as instâncias **realmente positivas**, quantas foram **corretamente identificadas**?

$$
\text{Recall} = \frac{TP}{TP + FN}
$$

### 💡 Interpretação:
Capacidade do modelo de **capturar todos os casos relevantes**. Alta recall = **baixo FN**.

### 📉 Exemplo prático:
Num modelo de diagnóstico de doenças, recall alto significa **não deixar casos positivos passarem despercebidos**.

### ⚠️ Limitações:
- Pode ser **altamente inflado à custa da precisão**.

### ✅ Use quando:
- Custo de **falsos negativos é muito alto** (ex: perder um cliente churn, deixar passar uma fraude).

---

## ⚖️ 4. F1-Score

### 📘 Conceito:
Média harmônica entre precision e recall. Penaliza extremos.

$$
F1 = 2 \cdot \frac{Precision \cdot Recall}{Precision + Recall}
$$

### 💡 Interpretação:
Compromisso entre **capturar casos positivos (recall)** e **acertar nas predições positivas (precision)**.

### 📉 Exemplo prático:
Em classificação de spam, F1 alto significa que o modelo identifica a maioria dos spams e raramente marca e-mails legítimos como spam.

### ⚠️ Limitações:
- **Não reflete o custo real de FP e FN.**
- Não captura bem casos com **interesse desbalanceado entre precision e recall.**

### ✅ Use quando:
- Deseja **balancear** FP e FN.
- Problema com **classes desbalanceadas**.

---

## 📈 5. AUC-ROC (Área Sob a Curva ROC)

### 📘 Conceito:
Avalia a **capacidade do modelo de rankear corretamente** as classes, independentemente do threshold.

- ROC: curva TPR vs FPR
- AUC: área sob essa curva

$$
AUC = \int_0^1 TPR(FPR) \, d(FPR)
$$

### 💡 Interpretação:
AUC é a **probabilidade de o modelo rankear um positivo acima de um negativo**.

- AUC = 1: classificação perfeita
- AUC = 0.5: aleatório

### 📉 Exemplo prático:
Em classificação de crédito, um AUC de 0.85 significa que há 85% de chance do modelo rankear um bom pagador acima de um inadimplente.

### ⚠️ Limitações:
- Não considera **threshold operacional**.
- Pode esconder problemas com **calibração de probabilidade**.

### ✅ Use quando:
- Quer **avaliar modelo em diferentes thresholds**.
- Precisa medir **capacidade de discriminação**, especialmente em modelos com score contínuo.

---

## 📊 6. Gini Coefficient

### 📘 Conceito:
Índice derivado diretamente do AUC:

$$
\text{Gini} = 2 \cdot AUC - 1
$$

### 💡 Interpretação:
Gini mede **desigualdade de rankeamento** entre classes. Usado extensivamente em modelos de risco de crédito.

- Gini = 1: separação perfeita
- Gini = 0: sem poder preditivo

### 📉 Exemplo prático:
Um modelo com Gini = 0.60 é **bem melhor que um modelo aleatório** e com boa capacidade de ranquear bons/mau pagadores.

### ⚠️ Limitações:
- **Menos intuitivo** que o AUC.
- Pouco usado fora de domínios regulados.

### ✅ Use quando:
- Em **modelos de score regulados** (bancos, seguros).
- Quando deseja uma métrica robusta ao threshold.

---

## 📊 7. KS-Score (Kolmogorov-Smirnov)

### 📘 Conceito:
Distância máxima entre as curvas de distribuição acumulada de scores para as classes 1 e 0.

$$
KS = \max |F_1(x) - F_0(x)|
$$

Onde:
- $F_1(x)$ = CDF dos scores dos positivos
- $F_0(x)$ = CDF dos scores dos negativos

### 💡 Interpretação:
Quanto maior o KS, **melhor a separação** entre as distribuições de classes.

- KS > 0.4: excelente
- KS ~ 0.2–0.3: aceitável

### 📉 Exemplo prático:
Em churn, se KS = 0.52, o modelo separa muito bem clientes que vão sair dos que vão ficar.

### ⚠️ Limitações:
- Sensível a dados ruidosos e **outliers nos scores**.
- Requer análise visual adicional (KS-plot).

### ✅ Use quando:
- Trabalha com **modelos de scoring**.
- Deseja avaliar **poder discriminatório real** com base em probabilidade.

## 📘 Resumo Comparativo das Métricas

| Métrica     | Foco Principal        | Threshold Depende? | Robusto a Desbalanceamento? | Interpretação Direta? |
|-------------|-----------------------|---------------------|------------------------------|------------------------|
| Accuracy    | % total correto       | ✅ Sim              | ❌ Não                       | ✅ Sim                 |
| Precision   | Evitar FP             | ✅ Sim              | ❌ Não                       | ✅ Sim                 |
| Recall      | Evitar FN             | ✅ Sim              | ❌ Não                       | ✅ Sim                 |
| F1-Score    | Balancear FP/FN       | ✅ Sim              | ✅ Sim                       | ✅ Sim                 |
| AUC         | Rankeamento           | ❌ Não              | ✅ Sim                       | ✅ Sim                 |
| Gini        | Separabilidade        | ❌ Não              | ✅ Sim                       | ❌ Pouco intuitivo     |

## 🛠️ Visualizações Recomendadas

- `sklearn.metrics.roc_curve`, `roc_auc_score`
- `precision_recall_curve`
- `seaborn.heatmap` (confusion matrix)
- **KS-plot**: duas curvas acumuladas com o ponto de maior separação
- **Lift e Cumulative Gain Charts**

---

## ✅ Conclusão Final

- Use **múltiplas métricas**: cada uma responde a uma pergunta diferente.
- Considere **impacto de negócio** ao priorizar métricas.
- **Para classes desbalanceadas**, evite acurácia como métrica principal.
- **Para modelos de score**, AUC, Gini e KS são cruciais.
- **Para classificação direta**, Precision, Recall e F1 são preferíveis.

💡 O domínio completo dessas métricas é **fundamental para ajustar modelos, reportar resultados com confiança** e tomar **decisões baseadas em dados com responsabilidade**.
