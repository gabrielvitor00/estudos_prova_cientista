# 🧪 Testes de Hipóteses — Fundamento Estatístico Aplicado à Ciência de Dados

Testes de hipóteses são **procedimentos estatísticos formais** usados para **tomar decisões** com base em dados. Eles avaliam se uma **afirmação (hipótese)** sobre uma população é compatível com os dados observados.

---

## 🎯 1. Conceito Teórico

### ✅ O que é um Teste de Hipótese?

Um **teste de hipótese** consiste em:
- Formular uma **hipótese nula** ($H_0$): suposição de status quo (ex: "não há diferença").
- Formular uma **hipótese alternativa** ($H_1$ ou $H_a$): o que se deseja testar ou provar.
- Calcular uma **estatística de teste** a partir da amostra.
- Avaliar se os dados são suficientemente incompatíveis com $H_0$ para rejeitá-la.

---

## 📐 2. Estrutura Formal

### ✅ Hipóteses
- $H_0$: Não há efeito, diferença, ou relação.
- $H_1$: Existe um efeito, diferença, ou relação.

### ✅ Nível de significância ($\alpha$)
- Probabilidade máxima de rejeitar $H_0$ quando ela é verdadeira (erro tipo I).
- Valor comum: $\alpha = 0{,}05$

### ✅ Estatística de teste (T)
- Valor calculado da amostra que será comparado com uma distribuição teórica (ex: Normal, t, F, Qui-quadrado)

### ✅ Valor-p (p-value)
- Probabilidade de obter um resultado tão extremo quanto o observado, assumindo que $H_0$ é verdadeira.
- Interpretação:
  - $p < \alpha$ → rejeita-se $H_0$ (resultado **estatisticamente significativo**)
  - $p \geq \alpha$ → não se rejeita $H_0$

---

## ⚠️ 3. Tipos de Erro

| Tipo de Erro     | Significado                                     |
|------------------|--------------------------------------------------|
| **Erro tipo I**  | Rejeitar $H_0$ quando ela é verdadeira         |
| **Erro tipo II** | Não rejeitar $H_0$ quando ela é falsa          |
| **Poder do teste** | $1 - \beta$: capacidade de detectar um efeito real |

---

## 🔬 4. Principais Testes Estatísticos

### ✅ Teste t (Student)

- **Objetivo**: comparar médias
- **Tipos**:
  - Amostra única: $\mu = \mu_0$
  - Duas amostras independentes
  - Duas amostras pareadas
- **Exemplo (Python):**

```python
from scipy.stats import ttest_ind
t_stat, p_val = ttest_ind(grupo1, grupo2)
```

---

### ✅ Teste de proporção (z-test)

- Comparação entre proporções binárias.
- Ex: taxa de conversão em campanhas A e B.

---

### ✅ Teste Qui-quadrado ($\chi^2$)

- Testa independência entre variáveis categóricas ou ajuste a distribuições esperadas.

---

### ✅ Teste de normalidade

- Verifica se os dados seguem distribuição normal.
  - **Shapiro-Wilk**: amostras pequenas
  - **Kolmogorov-Smirnov**: geral
  - **Anderson-Darling**: sensível em caudas

---

### ✅ Teste de Mann-Whitney (U)

- Alternativa não paramétrica ao t-test
- Compara **distribuições** entre dois grupos (não exige normalidade)

---

## 🧠 5. Aplicações Reais em Ciência de Dados

| Cenário Prático                                   | Teste Usado                              |
|---------------------------------------------------|-------------------------------------------|
| Testar diferença entre taxas de conversão (A/B)   | Teste z para proporções                  |
| Validar se os resíduos de um modelo são normais   | Shapiro-Wilk, Anderson-Darling           |
| Comparar métricas de modelos com e sem feature    | Teste t pareado                          |
| Avaliar mudança de comportamento antes/depois     | Teste t pareado ou Wilcoxon              |
| Verificar independência entre variáveis categóricas | Teste Qui-quadrado                       |
| Testar se uma feature explica a variável resposta | Teste F ou ANOVA                         |

---

## 🔄 6. Diferenças com Conceitos Correlatos

| Conceito              | Testes de Hipóteses                              | Intervalos de Confiança               |
|------------------------|--------------------------------------------------|----------------------------------------|
| Finalidade             | Decisão: rejeita $H_0$ ou não                  | Estima faixa plausível para parâmetro  |
| Output                 | p-valor                                          | Intervalo de valores                   |
| Relacionamento         | Complementares                                   | Intervalo que não contém $H_0$ → rejeita $H_0$ |

---

## 📊 7. Interpretação em Modelos Preditivos

- **Modelos lineares (regressão, GLM)**:
  - Cada coeficiente tem p-valor associado → testa se o impacto da variável é estatisticamente significativo
- **Seleção de variáveis**:
  - A partir da significância dos coeficientes ou testes F globais
- **Validação de suposições**:
  - Normalidade dos resíduos, homocedasticidade, independência

## ✅ 9. Vantagens, Limitações e Armadilhas

| Aspecto           | Vantagens                                         | Limitações / Armadilhas                     |
|-------------------|---------------------------------------------------|---------------------------------------------|
| Decisão objetiva  | Base em estatística formal                        | Sensível ao tamanho amostral (p ≠ impacto)  |
| Flexível          | Adapta-se a muitos contextos                      | Suposições nem sempre válidas (normalidade, homocedasticidade) |
| Popular           | Fácil de interpretar quando bem aplicado          | P-valor pode ser mal interpretado           |

---

## 🧠 Dicas para Ciência de Dados

- ⚠️ **P-valor pequeno ≠ efeito relevante** (analisar tamanho de efeito)
- 🔄 Combine com **análises gráficas** (boxplot, densidade, violin)
- 📏 Use **intervalos de confiança** como complemento interpretativo
- ✅ Sempre teste suposições antes de aplicar testes paramétricos
