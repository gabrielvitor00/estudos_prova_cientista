# 📈 Principais Distribuições de Probabilidade (Foco em Ciência de Dados)

As distribuições de probabilidade descrevem como os valores de uma variável aleatória estão distribuídos. Conhecer suas propriedades e aplicações é essencial em projetos de **modelagem estatística**, **inferência**, **simulações** e **avaliação de modelos** em Ciência de Dados.

---

## 🔷 1. Distribuição Normal (Gaussiana)

### 🧠 Conceito:
Distribuição **contínua e simétrica** em torno da média. É a mais usada na estatística por sua presença natural em fenômenos reais (ex: altura, peso, ruído).

### 📐 Características formais:
- Função densidade (PDF):

$$
f(x) = \frac{1}{\sigma \sqrt{2\pi}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right )
$$

- Parâmetros:
  - $\mu$: média
  - $\sigma^2$: variância
- Simétrica, média = mediana = moda

### 📊 Aplicações:
- Modelagem de **erros residuais** em regressão.
- Testes de hipótese (ex: t-test, z-test).
- Detecção de outliers.
- Padronização (z-score).

### ✅ Vantagens:
- Muitos algoritmos assumem normalidade.
- Facilidade de interpretação e cálculo.

### ⚠️ Armadilhas:
- Muitos dados reais **não seguem** distribuição normal.
- Outliers distorcem fortemente média e variância.

### 🔧 Ferramentas:
- `scipy.stats.norm`
- `seaborn.histplot`, `sns.kdeplot`

---

## 🔷 2. Distribuição Bernoulli

### 🧠 Conceito:
Distribuição **discreta** que modela um experimento com dois resultados possíveis: sucesso (1) ou fracasso (0).

### 📐 Características formais:
- Probabilidade de sucesso: $p$
- Probabilidade de fracasso: $1 - p$
- Esperança: $E[X] = p$
- Variância: $Var(X) = p(1 - p)$

### 📊 Aplicações:
- Classificação binária.
- Modelagem de eventos booleanos: clique ou não, churn ou não.

### 🔧 Ferramentas:
- `scipy.stats.bernoulli`

---

## 🔷 3. Distribuição Binomial

### 🧠 Conceito:
Modela o número de sucessos em $n$ ensaios independentes de Bernoulli.

### 📐 Características formais:
- Parâmetros: $n$, $p$
- Função de massa:

$$
P(X = k) = \binom{n}{k} p^k (1 - p)^{n-k}
$$

- Esperança: $E[X] = np$
- Variância: $Var(X) = np(1 - p)$

### 📊 Aplicações:
- Quantidade de acertos em testes A/B.
- Contagem de compras ou cliques em campanhas.

### 🔧 Ferramentas:
- `scipy.stats.binom`

---

## 🔷 4. Distribuição Uniforme

### 🧠 Conceito:
Todos os valores em um intervalo têm **probabilidades iguais**.

### 📐 Características formais:
**Contínua (uniforme real):**

- $a \leq X \leq b$
- PDF:

$$
f(x) = \frac{1}{b - a}
$$

**Discreta:**
- $P(X = x) = \frac{1}{n}$, para $n$ valores distintos.

### 📊 Aplicações:
- Simulações aleatórias.
- Inicialização de pesos em redes neurais.
- Geração de ruído.

### 🔧 Ferramentas:
- `numpy.random.uniform`
- `scipy.stats.uniform`

---

## 🔷 5. Distribuição de Poisson

### 🧠 Conceito:
Modela o número de eventos em um intervalo de tempo ou espaço **fixo**, quando esses eventos ocorrem **independentemente**.

### 📐 Características formais:
- Parâmetro: $\lambda$ (média e variância)
- PMF:

$$
P(X = k) = \frac{e^{-\lambda} \lambda^k}{k!}
$$

### 📊 Aplicações:
- Modelagem de chamadas em call centers.
- Detecção de anomalias (eventos raros).
- Contagem de acessos, fraudes, falhas de sistema.

### 🔧 Ferramentas:
- `scipy.stats.poisson`

---

## 🔷 6. Distribuição Geométrica

### 🧠 Conceito:
Modela o número de tentativas até o **primeiro sucesso** em ensaios de Bernoulli.

### 📐 Características formais:
- Parâmetro: $p$
- PMF:

$$
P(X = k) = (1 - p)^{k - 1} p
$$

### 📊 Aplicações:
- Previsão de número de interações até conversão.
- Número de chamadas até resolver um problema.
- Análise de churn com *survival models*.

### 🔧 Ferramentas:
- `scipy.stats.geom`

---

## 🔷 7. Distribuição Log-Normal

### 🧠 Conceito:
Distribuição de uma variável cujo logaritmo tem distribuição **normal**. É **assimétrica** e positiva.

### 📐 Características formais:
- $Y = \log(X) \sim \mathcal{N}(\mu, \sigma^2)$
- Usada quando os dados são multiplicativos ou com cauda longa.

### 📊 Aplicações:
- Modelagem de **salários**, **renda**, **tempo de vida**.
- Análise de dados **altamente assimétricos**.
- Modelagem de riscos financeiros.

### 🔧 Ferramentas:
- `scipy.stats.lognorm`

---

## 🔁 Comparações e Relações

| Distribuição   | Tipo      | Aplicação típica                          | Correlação com outras |
|----------------|-----------|-------------------------------------------|------------------------|
| Normal         | Contínua  | Modelos de erro, residuais                | Base de muitos testes  |
| Bernoulli      | Discreta  | Eventos binários                          | Caso especial de Binomial (n = 1) |
| Binomial       | Discreta  | Contagem de sucessos                      | Soma de Bernoullis     |
| Uniforme       | Contínua/Discreta | Simulações, inicializações       | Base para gerar outras distribuições |
| Poisson        | Discreta  | Contagem de eventos raros                 | Limite da Binomial (n \to \infty, p \to 0) |
| Geométrica     | Discreta  | Número de tentativas até sucesso          | Análoga à Exponencial  |
| Log-Normal     | Contínua  | Valores positivos com cauda longa         | Exponencial multiplicativo |

---

## ✅ Conclusão

Conhecer as **distribuições de probabilidade** é essencial para a prática da Ciência de Dados, pois elas:
- Guiam a escolha e avaliação de **modelos estatísticos**.
- Fundamentam **testes de hipótese**, **simulações** e **inferência bayesiana**.
- Permitem entender a **estrutura dos dados** e selecionar transformações apropriadas.

A má escolha da distribuição pode comprometer toda a análise — por isso, entender **quando e como usar cada uma** é uma habilidade fundamental para um cientista de dados.
