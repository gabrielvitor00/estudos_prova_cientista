# 📊 Propriedades de Distribuições: Médias, Medianas, Quartis, Moda, Variância, etc.  
*(Foco total em Ciência de Dados)*

As propriedades estatísticas de uma distribuição **resumem e descrevem** a forma, dispersão, tendência central e características de seus dados. Compreender essas propriedades é essencial para a **escolha de modelos, pré-processamento, engenharia de atributos e avaliação de hipóteses** em Ciência de Dados.

---

## 📌 1. Tendência Central: Média, Mediana e Moda

### ✅ **Média (Esperança Matemática)**

#### 💡 Conceito:
Valor médio de uma variável. Representa o **centro de gravidade** da distribuição.

#### 📐 Definição:
Para variável discreta:  
$$
\mu = E[X] = \sum_{i} x_i P(x_i)
$$  
Para variável contínua:  
$$
\mu = E[X] = \int_{-\infty}^{\infty} x f(x) dx
$$

#### 📊 Aplicações:
- Avaliação de performance média (ex: score médio de modelo)
- Feature engineering: substituição de nulos (cuidado!)
- Ponto de referência para z-score e padronização

#### ⚠️ Limitações:
- **Muito sensível a outliers e assimetrias**
- Pode **não representar bem** dados assimétricos (ex: log-normal)

---

### ✅ **Mediana**

#### 💡 Conceito:
Valor que **divide a amostra ao meio** (50% abaixo, 50% acima).  
Robusta a outliers.

#### 📐 Cálculo:
- Ordena os valores: se ímpar, valor central; se par, média dos dois centrais.

#### 📊 Aplicações:
- Avaliação de tendências centrais em distribuições assimétricas (ex: tempo de resposta, salários)
- Robustez em datasets com ruído ou caudas pesadas

#### ⚠️ Limitações:
- Menos eficiente que a média em distribuições simétricas

---

### ✅ **Moda**

#### 💡 Conceito:
Valor **mais frequente** em uma distribuição.

#### 📊 Aplicações:
- Análise de **variáveis categóricas**
- Detectar valores dominantes em classificações

#### ⚠️ Armadilhas:
- Pode não existir (ou haver múltiplas modas)
- Menos informativa em variáveis contínuas

---

## 📌 2. Medidas de Dispersão: Variância, Desvio Padrão, Amplitude, Intervalos Interquartis

### ✅ **Variância (σ²)**

#### 💡 Conceito:
Mede a **dispersão dos dados** em torno da média.

#### 📐 Fórmula:
$$
Var(X) = E[(X - \mu)^2]
$$  
A unidade é o **quadrado da unidade da variável original**.

---

### ✅ **Desvio Padrão (σ)**

#### 💡 Conceito:
Raiz quadrada da variância — indica **quão longe, em média, os dados estão da média**.  
Tem **mesma unidade dos dados**, facilitando a interpretação.

#### 📊 Aplicações práticas:
- Padronização com Z-score
- Regras da normal: 68%, 95%, 99.7%
- Avaliação de risco em modelos preditivos e decisões

---

### ✅ **Amplitude (Range)**

$$
\text{Range} = \text{máximo} - \text{mínimo}
$$

- Simples, mas **extremamente sensível a outliers**

---

### ✅ **Intervalo Interquartil (IQR)**

$$
\text{IQR} = Q3 - Q1
$$

- Abrange o intervalo central de 50% dos dados
- Base para detectar outliers (valores fora de $[Q1 - 1.5 \cdot IQR,\ Q3 + 1.5 \cdot IQR]$)

---

## 📌 3. Medidas de Forma: Assimetria (Skewness) e Curtose (Kurtosis)

### ✅ **Assimetria (Skewness)**

#### 💡 Conceito:
Indica a **simetria da distribuição**.

- $\text{Skewness} = 0$: distribuição simétrica
- $\text{Skewness} > 0$: cauda à direita (ex: salários)
- $\text{Skewness} < 0$: cauda à esquerda

#### 📊 Implicações:
- Modelos lineares sofrem com variáveis muito assimétricas
- Pode-se aplicar transformações log, Box-Cox, Yeo-Johnson

---

### ✅ **Curtose (Kurtosis)**

#### 💡 Conceito:
Mede a **concentração de valores nos extremos** (caudas) comparado à normal.

- $\text{Kurtosis} = 3$: normal
- $\text{Kurtosis} > 3$: mais caudas (leptocúrtica)
- $\text{Kurtosis} < 3$: menos caudas (platicúrtica)

#### 📊 Aplicações:
- Avaliação de probabilidade de valores extremos
- Avaliação de risco de eventos raros (ex: fraudes)

---

## 🔁 Comparações e Interpretações Conjuntas

| Métrica        | Interpretação                           | Impacto em Modelagem                    |
|----------------|------------------------------------------|-----------------------------------------|
| Média          | Tendência central sensível a outliers    | Referência para normalização            |
| Mediana        | Tendência central robusta                | Útil com dados assimétricos             |
| Variância/σ    | Dispersão em torno da média              | Avalia risco e instabilidade            |
| Skewness       | Assimetria dos dados                     | Indica necessidade de transformação     |
| IQR            | Dispersão resistente a outliers          | Fundamental na detecção de anomalias    |
| Moda           | Valor mais comum                         | Análise de frequência para categóricos  |

---

## 🧠 Aplicações em Ciência de Dados

| Aplicação prática           | Propriedades usadas                      |
|----------------------------|------------------------------------------|
| **Modelagem preditiva**     | Verificação de normalidade, variância   |
| **Feature Engineering**     | Transformações baseadas em skewness     |
| **Detecção de outliers**    | Z-score, IQR, boxplot                   |
| **Modelos lineares**        | Pressupõem média e variância constantes |
| **Normalização/padronização** | Média e desvio padrão usados            |
| **Teste de hipóteses**      | Média, variância, curtose envolvidas    |

## ✅ Conclusão

As propriedades estatísticas de uma distribuição são **muito mais do que simples métricas descritivas**: elas **orientam decisões analíticas, guiam transformações, afetam resultados de modelos e testes**. Compreender cada medida em profundidade permite atuar de forma mais estratégica em todas as etapas de um pipeline de ciência de dados.

```python
# Diagnóstico completo de uma variável
col = df["feature"]

print("Média:", col.mean())
print("Mediana:", col.median())
print("Desvio padrão:", col.std())
print("Assimetria:", col.skew())
print("Curtose:", col.kurtosis())
```
