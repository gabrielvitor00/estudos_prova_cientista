# 🔍 Função Densidade de Probabilidade (PDF) e Distribuição Acumulada (CDF)
*Com foco em aplicações de Ciência de Dados*

As funções de densidade (PDF) e distribuição acumulada (CDF) são ferramentas fundamentais para descrever o **comportamento probabilístico de variáveis aleatórias** — contínuas ou discretas. Elas sustentam o funcionamento de modelos probabilísticos, inferência estatística e validações baseadas em suposições de distribuição.

---

## 📌 1. Conceito Teórico

### ✅ Função Densidade de Probabilidade (PDF)

#### 💡 Definição:
A **PDF** descreve a **densidade de probabilidade em torno de cada valor possível de uma variável aleatória contínua**. Diferentemente da probabilidade de um valor específico, que é zero em variáveis contínas, a PDF permite calcular **probabilidades em intervalos**.

#### 📐 Formulação:
Seja $X$ uma variável contínua com PDF $f(x)$, então:

- A probabilidade de $X$ estar entre $a$ e $b$ é:

$$
P(a \leq X \leq b) = \int_a^b f(x)\, dx
$$

- Propriedades:
  - $f(x) \geq 0$ para todo $x$
  - $\int_{-\infty}^{\infty} f(x)\, dx = 1$ (área total = 1)

---

### ✅ Função de Distribuição Acumulada (CDF)

#### 💡 Definição:
A **CDF** (Cumulative Distribution Function) de uma variável aleatória representa a **probabilidade de que a variável assuma um valor menor ou igual a um ponto específico**.

#### 📐 Formulação:
- Para variável contínua:

$$
F(x) = P(X \leq x) = \int_{-\infty}^{x} f(t) dt
$$

- Para variável discreta:

$$
F(x) = \sum_{x_i \leq x} P(X = x_i)
$$

#### Propriedades:
- **Monótona crescente:** $F(x_1) \leq F(x_2)$ se $x_1 < x_2$
- Limites:
  - $\lim_{x \to -\infty} F(x) = 0$
  - $\lim_{x \to +\infty} F(x) = 1$

## 🧠 Aplicação Prática em Ciência de Dados

### 📊 PDF – Casos de uso:
- Estimar distribuição dos **erros/resíduos** em regressões
- Detectar **anormalidades** e **caudas longas** (análise de risco, fraude)
- Criar **modelos generativos probabilísticos** (ex: Naive Bayes contínuo)
- Aplicar transformações baseadas na forma da densidade (ex: log, box-cox)

### 📈 CDF – Casos de uso:
- Avaliação de **quantis**, **percentis**, **quartis**
- Construção de **funções de perda baseadas em quantis** (ex: pinball loss)
- Visualização de **domínio de influência de outliers**
- Aplicação de **transformações baseadas em distribuição empírica** (ex: quantile normalization)

---

### 🧪 Exemplos práticos

Abaixo, visualizamos a PDF e a CDF de uma distribuição normal padrão ($\mu=0$, $\sigma=1$):

```python
from scipy.stats import norm
import numpy as np
import matplotlib.pyplot as plt

# Gerando dados
x = np.linspace(-4, 4, 1000)
pdf = norm.pdf(x, loc=0, scale=1)
cdf = norm.cdf(x, loc=0, scale=1)

# Configurando o gráfico
plt.figure(figsize=(12,5))

# Subgráfico 1 - PDF
plt.subplot(1,2,1)
plt.title("PDF - Distribuição Normal")
plt.plot(x, pdf, label="PDF")
plt.fill_between(x, pdf, alpha=0.3)
plt.xlabel("Valores")
plt.ylabel("Densidade de Probabilidade")
plt.legend()
plt.grid(True)

# Subgráfico 2 - CDF
plt.subplot(1,2,2)
plt.title("CDF - Distribuição Normal")
plt.plot(x, cdf, label="CDF", color="green")
plt.xlabel("Valores")
plt.ylabel("Probabilidade Acumulada")
plt.legend()
plt.grid(True)

plt.show()
```

---

## 📊 Diferenças Entre PDF e CDF

| Aspecto                  | PDF                                        | CDF                                      |
|--------------------------|--------------------------------------------|------------------------------------------|
| Tipo de função           | Derivada da CDF                            | Integral da PDF                          |
| Interpretação            | Densidade instantânea de probabilidade     | Probabilidade acumulada até x            |
| Aplicação em contínuas   | $$P(a \leq X \leq b) = \int_a^b f(x)dx$$   | $$F(b) - F(a)$$                          |
| Unidade                  | Densidade (não é probabilidade direta)     | Probabilidade (de 0 a 1)                 |
| Forma visual             | Curva tipo sino (ex: normal)               | Curva sigmoide (crescente)               |
| Uso em thresholds        | Não diretamente                            | Sim (usado em ROC, percentis, outliers) |

## ✅ Vantagens, Limitações e Armadilhas

| Propriedade         | Vantagens                                      | Limitações / Armadilhas                               |
|---------------------|-----------------------------------------------|--------------------------------------------------------|
| PDF                 | Altamente informativa sobre forma da variável | Não dá probabilidade direta de valores exatos          |
|                     | Útil para detectar multimodalidade e caudas   | Requer suavização (ex: KDE) em amostras pequenas       |
| CDF                 | Probabilidade acumulada intuitiva             | Pouco sensível à variação local da densidade           |
|                     | Ideal para comparar distribuições             | Pode ocultar detalhes de caudas ou assimetrias         |

## 📐 Interpretação Prática em Modelagem

- 📦 **Modelos bayesianos** utilizam PDFs como funções de verossimilhança
- 📉 **Transformações não lineares** podem ser guiadas pela PDF/CDF para tornar os dados mais normais ou simétricos
- 🧮 **Análise de quantis** (usando a CDF) é crítica para:
  - Modelos quantílicos
  - Modelos de score de crédito
  - Avaliação de fairness e distribuição preditiva
