# 📍 KNN – K-Nearest Neighbors (Classificação) – Guia Ultra-Robusto para Ciência de Dados

O **K-Nearest Neighbors (KNN)** é um algoritmo de classificação (e regressão) baseado em **instâncias**. Seu funcionamento parte de uma premissa simples e poderosa: **observações similares estão próximas no espaço de atributos**.

Em essência, **KNN não constrói um modelo paramétrico**, mas retém o conjunto de treinamento e faz predições com base na **distância** entre a nova instância e os exemplos conhecidos.

---

## 📘 Conceito Teórico

O algoritmo **classifica uma nova observação** considerando os **k exemplos mais próximos** (vizinhos) no conjunto de treino. A **classe mais comum entre os vizinhos** é atribuída à nova instância.

$$
\hat{y}(x) = \arg\max_{c \in \mathcal{C}} \sum_{i=1}^{k} \mathbf{1}(y_i = c)
$$

- $\hat{y}(x)$: classe predita para o ponto $x$
- $k$: número de vizinhos considerados
- $\mathbf{1}$: função indicadora
- $y_i$: rótulo da $i$-ésima amostra mais próxima
- $\mathcal{C}$: conjunto de classes possíveis

---

## ⚙️ Características Principais

| Característica           | Detalhes                                                        |
|-------------------------|-----------------------------------------------------------------|
| **Tipo de modelo**      | Classificador **não paramétrico**, **preguiçoso** (lazy learner) |
| **Fase de treinamento** | Quase inexistente (apenas armazena os dados)                    |
| **Fase de predição**    | Custo computacional alto (busca por vizinhos em tempo real)      |
| **Função de distância** | Euclidiana, Manhattan, Minkowski, Coseno, etc.                  |
| **Classificação**       | Votação majoritária entre os vizinhos                           |
| **Regressão**           | Média (ou mediana) dos valores dos vizinhos                     |
| **Vizinhos ponderados** | Possível atribuir maior peso a vizinhos mais próximos           |

---

## 🧠 Aplicações em Ciência de Dados (Classificação)

| Caso de uso         | Aplicação prática                                                        |
|---------------------|--------------------------------------------------------------------------|
| Finanças            | Classificação de transações como fraudulentas vs legítimas               |
| Saúde               | Diagnóstico baseado em similaridade com casos anteriores                 |
| Recomendação        | Agrupamento de usuários ou produtos por perfis similares                 |
| Visão computacional | Reconhecimento de dígitos manuscritos (ex: MNIST), faces, imagens        |
| NLP                 | Classificação de sentimentos com vetores semânticos (ex: word embeddings)|

---

## ✅ Vantagens

- 🔍 **Simples** de entender e implementar.
- 🧠 **Alta capacidade de modelar padrões complexos**, sem necessidade de assumir forma funcional.
- 📊 **Classificador adaptativo**: regiões de decisão podem ser arbitrariamente não-lineares.
- 🔁 Funciona bem com **dados atualizados dinamicamente** (streaming).
- 🔧 Não precisa de treinamento intensivo → ideal para sistemas dinâmicos de recomendação.

---

## ❌ Limitações e Armadilhas

| Limitação                  | Detalhamento                                                                 |
|---------------------------|------------------------------------------------------------------------------|
| ❌ Custo computacional alto| Predição lenta: exige cálculo de distância com todas as instâncias           |
| ❌ Sensível à dimensionalidade | Em espaços com muitas variáveis (curse of dimensionality), distâncias perdem significado |
| ❌ Sensível a outliers     | Um ponto ruidoso pode atrair votos indevidos                                 |
| ❌ Requer normalização     | Variáveis com magnitudes diferentes distorcem o cálculo de distância         |
| ❌ Classes desbalanceadas  | Maioria pode dominar votação → uso de ponderação ou ajuste de prior          |

---

## 📏 Distâncias Usadas em KNN

| Distância    | Fórmula                                                                 | Aplicabilidade                |
|--------------|-------------------------------------------------------------------------|-------------------------------|
| Euclidiana   | $\sqrt{\sum (x_i - x_j)^2}$                                            | Padrão em dados contínuos     |
| Manhattan    | $\sum |x_i - x_j|$                                                     | Mais robusta a outliers       |
| Minkowski    | $\left(\sum |x_i - x_j|^p\right)^{1/p}$, generaliza Euclidiana ($p=2$) e Manhattan ($p=1$) | Parâmetro $p$ controlável     |
| Coseno       | $1 - \frac{x \cdot y}{\|x\|\|y\|}$                                  | Dados esparsos, vetores texto |

## 🔢 Estratégias de Escolha do $k$

- Pequeno $k$: modelo mais **complexo**, com maior variância (overfitting)
- Grande $k$: modelo mais **simples**, maior viés (underfitting)

💡 Dica: sempre usar **validação cruzada** para escolha do melhor $k$

## 🧪 Implicações para Modelagem

- 📌 Não aprende um modelo explícito → ideal quando os dados são ricos e próximos da fronteira.
- 📈 Ideal para problemas em que a proximidade semântica é relevante (ex: embeddings).
- ❗ Necessita de pré-processamento robusto:
    - Escalonamento com `StandardScaler` ou `MinMaxScaler`
    - Redução de dimensionalidade com `PCA` ou `UMAP`
    - Binarização ou embeddings para variáveis categóricas

---

## 🔬 Validação e Avaliação

| Técnica de Validação      | Aplicação em KNN                                              |
|--------------------------|---------------------------------------------------------------|
| Hold-out ou Cross Validation | Avaliação da performance global (ex: accuracy, F1, AUC)    |
| Curva de erro vs. $k$    | Escolha do hiperparâmetro ideal (trade-off viés x variância)  |
| Matriz de Confusão       | Análise de erros entre classes                                |
| AUC-ROC / Precision-Recall | Avaliação probabilística (se `weights='distance'` usado)    |

## 📌 Conclusão

O KNN é um classificador poderoso, simples e não paramétrico, ideal para problemas em que a semelhança entre exemplos carrega informação crítica. Porém, seu custo computacional e sensibilidade à dimensionalidade exigem cuidados.

🧠 Seu uso é mais indicado quando:

- O dataset é pequeno ou moderado
- A estrutura local (vizinhança) tem valor preditivo alto
- Deseja-se flexibilidade sem modelagem paramétrica rígida

🎯 Dominar KNN é essencial para entender modelos baseados em instâncias, comportamento de dados em espaços vetoriais e técnicas de aprendizado não paramétrico.
