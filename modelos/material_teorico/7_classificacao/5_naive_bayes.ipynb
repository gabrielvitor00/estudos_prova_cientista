# 📘 Naive Bayes (Gaussiano e Bernoulli) – Guia Ultra-Robusto para Classificação em Ciência de Dados

O algoritmo **Naive Bayes** é uma **família de classificadores probabilísticos** baseados no **Teorema de Bayes**, com uma **hipótese forte (e irrealista) de independência entre atributos**.

Apesar da simplicidade, sua eficácia em tarefas de classificação é surpreendente, especialmente em **contextos com alta dimensionalidade** (como texto e bioinformática), onde modelos complexos tendem a sobreajustar.

---

## 📚 Conceito Teórico

Baseado no **Teorema de Bayes**:

$$
P(C_k \mid \mathbf{x}) = \frac{P(\mathbf{x} \mid C_k) \cdot P(C_k)}{P(\mathbf{x})}
$$

Onde:

- $P(C_k \mid \mathbf{x})$: probabilidade a posteriori da classe $C_k$ dado o vetor de atributos $\mathbf{x}$
- $P(\mathbf{x} \mid C_k)$: verossimilhança dos atributos dado a classe
- $P(C_k)$: probabilidade a priori da classe
- $P(\mathbf{x})$: probabilidade marginal dos dados

A simplificação “naive” assume **independência condicional** entre os atributos:

$$
P(\mathbf{x} \mid C_k) = \prod_{i=1}^{n} P(x_i \mid C_k)
$$

Essa suposição **reduz a complexidade computacional** drasticamente, permitindo cálculo eficiente mesmo em vetores de atributos extensos.

---

## 🧪 Variedades do Naive Bayes

| Variante           | Tipo de variável            | Distribuição assumida             |
|--------------------|-----------------------------|-----------------------------------|
| **Gaussiano**      | Contínua (ex: idade, renda) | Distribuição normal por atributo  |
| **Bernoulli**      | Binária (0 ou 1)            | Distribuição de Bernoulli         |
| Multinomial        | Inteira/contagem            | Distribuição multinomial          |
| Complementar       | Texto desbalanceado         | Variante melhor em desbalanceios  |

---

## 📐 Definições Formais – Classificador Gaussiano

Cada atributo $x_i$ é modelado como uma **distribuição normal** dentro de cada classe $C_k$:

$$
P(x_i \mid C_k) = \frac{1}{\sqrt{2\pi\sigma^2_{ki}}} \cdot \exp\left( -\frac{(x_i - \mu_{ki})^2}{2\sigma^2_{ki}} \right)
$$

- $\mu_{ki}$: média do atributo $x_i$ na classe $C_k$
- $\sigma^2_{ki}$: variância do atributo $x_i$ na classe $C_k$

## 🧠 Diferenças entre Variantes do Naive Bayes

| Característica      | **Bernoulli**                                | **Gaussiano**                                |
|---------------------|-----------------------------------------------|-----------------------------------------------|
| Tipo de variável    | Binária (0 ou 1)                              | Contínua                                      |
| Exemplo de dado     | Presença de palavras em um email              | Idade, salário, densidade óssea               |
| Requisitos          | Dados já binarizados                          | Dados numéricos contínuos                     |
| Verossimilhança     | $P(x_i \mid C_k) = p^{x_i} (1 - p)^{1 - x_i}$ | Normal por atributo dentro da classe          |
| Ponto forte         | Classificação de texto com vetores binários   | Dados reais onde a normalidade é razoável     |

## 🔄 Vantagens, ❌ Limitações e ⚠️ Armadilhas

### ✅ Vantagens

- ⚡ **Extremamente rápido** para treinar e predizer
- 🧠 **Interpretação probabilística** transparente
- 🏗️ Escala bem com **dados de alta dimensionalidade**
- 🔒 Robusto a **dados ruidosos e irrelevantes**
- 🧮 Estimativas simples, sem necessidade de tuning complexo

### ❌ Limitações

- 🚫 **Suposição de independência condicional** raramente válida
- 🚫 Desempenho reduzido quando há **forte correlação entre atributos**
- 🚫 Modela mal distribuições multimodais ou não gaussianas (caso Gaussiano)
- 🚫 Binários com muitos zeros tornam Bernoulli sensível à ausência de sinal

### ⚠️ Armadilhas comuns

| Problema                   | Solução recomendada                                 |
|---------------------------|-----------------------------------------------------|
| Atributos correlacionados | Redução de dimensionalidade (ex: PCA)               |
| Probabilidade zero        | Suavização de Laplace (add-one smoothing)           |
| Variância nula            | Regularização das variâncias                        |
| Dados não normalizados    | Padronização (`StandardScaler`) melhora desempenho  |

---

## 🔍 Interpretação Preditiva

- A **classe predita** é aquela com **maior probabilidade a posteriori**:

$$
\hat{y} = \arg\max_{C_k} P(C_k) \cdot \prod_{i=1}^{n} P(x_i \mid C_k)
$$

- A **probabilidade predita** permite:
    - Curvas ROC
    - Cálculo de entropia
    - Avaliação de calibragem probabilística

### 🔢 Probabilidades com suavização (Laplace):

$$
P(x_i \mid C_k) = \frac{N_{ik} + 1}{N_k + 2}
$$

Evita que categorias com zero ocorrência dominem a multiplicação com resultado 0.
