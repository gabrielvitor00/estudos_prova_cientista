# 🌳 Árvore de Classificação – Guia Ultra-Robusto para Modelos de Decisão em Ciência de Dados

As **árvores de classificação** são modelos preditivos amplamente utilizados por sua **intuitividade, capacidade de lidar com diferentes tipos de dados** e facilidade de interpretação. Fundamentadas na **teoria da entropia e impureza de dados**, elas particionam recursivamente o espaço amostral com base em regras de decisão **hierárquicas**, formando uma estrutura em árvore que classifica novas observações com base em perguntas sequenciais.

---

## 📘 Conceito Teórico

Uma **árvore de decisão** é um **modelo supervisionado** que mapeia entradas $X \in \mathbb{R}^d$ para uma variável de saída categórica $Y \in \{c_1, ..., c_k\}$ por meio de divisões binárias (ou multinárias) do espaço de atributos.

A lógica é:
- Em cada **nó de decisão**, escolhe-se a **melhor divisão** de um atributo $X_j$ segundo uma métrica de **impureza**.
- A recursão continua até se atingir um critério de parada (ex: profundidade, número mínimo de amostras, pureza total).

A estrutura é composta por:
- **Raiz (root)**: o nó inicial que contém todo o conjunto de treinamento.
- **Nós internos**: pontos onde decisões são tomadas com base nos atributos.
- **Folhas (leaves)**: nós terminais que atribuem uma **classe**.

---

## 🔧 Métricas de Divisão e Definições Formais

### 1. Gini Impurity (usado por padrão no CART)

$$
Gini(S) = 1 - \sum_{i=1}^{k} p_i^2
$$
Onde $p_i$ é a proporção de classe $i$ em um subconjunto $S$. Quanto menor, mais puro o nó.

### 2. Entropia (usado em ID3 e C4.5)

$$
H(S) = - \sum_{i=1}^{k} p_i \log_2(p_i)
$$

### 3. Information Gain:

$$
IG(S, A) = H(S) - \sum_{v \in \text{valores}(A)} \frac{|S_v|}{|S|} H(S_v)
$$

### 4. Gain Ratio (C4.5)

$$
GR = \frac{IG(S, A)}{IV(A)}
$$
Com $IV(A)$ sendo a entropia da distribuição dos valores do atributo $A$ — penaliza splits com muitos ramos.

## 🔍 Comparação Avançada das Métricas de Divisão em Árvores de Classificação

A **escolha da métrica de divisão** afeta **profundamente a estrutura da árvore**, o comportamento do algoritmo e a **forma como os dados são particionados**. Estas métricas determinam **como avaliar a qualidade de um split (divisão)** em um nó. Abaixo está um detalhamento conceitual e prático das métricas mais comuns: **Gini Impurity**, **Entropia + Information Gain** e **Gain Ratio**.

---

### 📌 Objetivo de uma métrica de divisão

Selecionar a variável e o ponto de corte que **maximizam a "pureza" dos subconjuntos** gerados a partir do split. Isso significa:
- Criar grupos que contenham, preferencialmente, **exemplares de uma única classe**.
- Minimizar a **impureza** (ou maximizar o ganho de informação) após a divisão.

---

## 🔢 1. Gini Impurity (usado por padrão no CART)

**Definição Formal:**

$$
Gini(S) = 1 - \sum_{i=1}^{k} p_i^2
$$

- Onde $p_i$ é a proporção de instâncias da classe $i$ no nó $S$.
- Varia de **0** (puro) a **1 - 1/k** (máxima impureza com k classes balanceadas).

**Interpretação:**
- Mede a **probabilidade de classificar incorretamente uma instância escolhida aleatoriamente**, se ela fosse rotulada de acordo com a distribuição do nó.

**Características:**
- Computacionalmente **mais eficiente** que entropia (evita logaritmos).
- Favorece **classes majoritárias** nos splits iniciais (mais agressivo).
- Utilizado no **algoritmo CART** (Classification and Regression Trees).

---

## 🔢 2. Entropia (usado no ID3, C4.5)

**Definição Formal:**

$$
Entropy(S) = -\sum_{i=1}^{k} p_i \cdot \log_2(p_i)
$$

- Mede o grau de **desordem** ou **incerteza** em um nó.
- Vai de **0** (completamente puro) até **log₂(k)** (máxima entropia, classes uniformes).

**Intuição:**
- Do ponto de vista da **teoria da informação**, representa o **número médio de bits necessários para codificar a classe de uma instância**.

**Informação Adicional — Information Gain (IG):**

$$
IG(S, A) = Entropy(S) - \sum_{v \in valores(A)} \frac{|S_v|}{|S|} \cdot Entropy(S_v)
$$

- Mede **quanto a entropia é reduzida** após dividir o nó com base na variável $A$.
- Serve como **critério de seleção de variáveis** nos algoritmos ID3 e C4.5.

---

## 🔢 3. Gain Ratio (usado no C4.5)

**Problema com Information Gain:**
- O **IG tende a favorecer atributos com muitos valores distintos** (ex: CPF, ID), mesmo que esses splits **não sejam generalizáveis**.

**Solução — Gain Ratio:**

$$
GainRatio(S, A) = \frac{InformationGain(S, A)}{SplitInformation(A)}
$$

- Onde:
$$
SplitInformation(A) = -\sum_{v \in valores(A)} \frac{|S_v|}{|S|} \cdot \log_2\left(\frac{|S_v|}{|S|}\right)
$$

**Interpretação:**
- Penaliza splits que **criam muitos subconjuntos pequenos** (ex: um por valor).
- Favorece atributos com **divisões mais equilibradas** e **informativas**.

## ⚖️ Comparação Prática Entre as Métricas

| Métrica           | Base Teórica             | Quando usar                                      | Pontos Fortes                              | Pontos Fracos                                |
|-------------------|--------------------------|--------------------------------------------------|---------------------------------------------|-----------------------------------------------|
| **Gini Impurity** | Probabilidade condicional| Árvores rápidas e pouco profundas (CART)         | Rápido, simples, favorece pureza local       | Menos sensível a mudanças de distribuição     |
| **Entropia**      | Teoria da informação     | Alta pureza desejada, compreensão de incerteza   | Mais sensível a mudanças, melhor para dados desbalanceados | Mais caro computacionalmente                 |
| **Information Gain** | Redução de entropia  | Escolha de variável para split (ID3, C4.5)       | Clareza estatística                          | Pode favorecer variáveis com muitos valores   |
| **Gain Ratio**    | Normaliza o IG           | Dados com variáveis altamente cardinais          | Neutraliza viés do IG, mais estável          | Pode penalizar atributos relevantes com alta cardinalidade legítima |

---

## 🧪 Exemplos Ilustrativos

**Exemplo 1 – Nó com 50% da classe A e 50% da classe B:**

| Métrica        | Valor |
|----------------|-------|
| Gini           | $1 - (0.5^2 + 0.5^2) = 0.5$ |
| Entropia       | $-[0.5 \log_2 0.5 + 0.5 \log_2 0.5] = 1.0$ |
| IG             | Depende da entropia do pai e da distribuição dos filhos |
| Gain Ratio     | Depende da entropia do split                          |

**Exemplo 2 – Nó puro (100% da classe A):**

| Métrica        | Valor |
|----------------|-------|
| Gini           | 0     |
| Entropia       | 0     |
| IG             | Máximo possível (se o pai era impuro) |
| Gain Ratio     | Elevado (se split foi eficaz)          |

---

## 📌 Conclusão

A escolha da métrica de divisão **não afeta apenas o desempenho final**, mas **também a complexidade, a forma e a interpretabilidade da árvore**.

🎯 **Recomendações práticas:**
- Use **Gini** para performance e simplicidade.
- Use **Entropia + Information Gain** se valoriza interpretação probabilística ou sensibilidade a desbalanceamento.
- Use **Gain Ratio** quando tiver atributos com muitos valores distintos ou quando utilizar o algoritmo C4.5.

🔄 Em muitos casos, os resultados práticos entre Gini e Entropia são semelhantes. O mais importante é aplicar poda e validação adequada para evitar **overfitting**, independentemente da métrica usada.

## ✅ Vantagens

- **Interpretável**: estrutura de decisão facilmente visualizada.
- **Não paramétrica**: não assume distribuições dos dados.
- **Funciona com variáveis categóricas e contínuas**.
- **Captura interações e não linearidades naturalmente**.
- **Rápido para inferência** (tempo de predição $O(\log n)$).

---

## ❌ Limitações e ⚠️ Armadilhas

| Problema                            | Descrição                                                                 |
|-------------------------------------|---------------------------------------------------------------------------|
| Sobreajuste (Overfitting)          | Árvores muito profundas se ajustam a ruídos e outliers                    |
| Alta variância                     | Pequenas mudanças nos dados geram árvores muito diferentes                |
| Fronteiras rígidas                 | Fronteiras de decisão ortogonais podem não representar bem o espaço      |
| Dados desbalanceados               | Pode favorecer classes majoritárias se não ajustado com pesos ou critérios |
| Falta de generalização             | Árvores únicas tendem a ter baixa performance fora da amostra             |

**Mitigação:**
- Poda (pruning)
- Limite de profundidade
- Aumento de dados
- Uso de ensembles (Random Forest, Boosting)

## 🧮 Interpretação Prática

**Como interpretar a saída:**
- Cada **caminho da raiz até a folha** representa uma **regra lógica de classificação**.
- A **profundidade da árvore** reflete a **complexidade da regra**.
- Pode-se extrair regras do tipo:
  > Se `idade < 30` e `salário > 3k`, então `não é inadimplente`.

**Feature Importance:**
A importância de uma variável é medida pela **redução total da impureza** que ela provoca nos splits onde participa.

$$
FI(j) = \sum_{nó~t:~split~em~X_j} \frac{N_t}{N} \cdot \Delta I_t
$$

## 🔬 Técnicas Relacionadas à Validação e Pré-processamento

| Técnica                       | Relevância para Árvores de Classificação                                   |
|-------------------------------|------------------------------------------------------------------------------|
| One-Hot Encoding              | **Nem sempre necessário**: árvores lidam diretamente com variáveis categóricas |
| Escalonamento de variáveis   | **Desnecessário**: splits são baseados em ordenamento, não magnitude         |
| **Stratified k-Fold**         | Preserva distribuição de classes em cross-validation                         |
| Cost-sensitive learning       | Ajuste de pesos para lidar com desbalanceamento                             |
| Grid Search em hiperparâmetros | Profundidade máxima, mínimo de amostras, tipo de split                      |
| Pruning                       | Eliminação de ramos com ganho marginal baixo                                 |

---

## 🧠 Hiperparâmetros Críticos

| Hiperparâmetro          | Efeito                                                                 |
|-------------------------|------------------------------------------------------------------------|
| `max_depth`             | Controla complexidade da árvore (evita overfitting)                    |
| `min_samples_split`     | Número mínimo para fazer nova divisão                                 |
| `min_samples_leaf`      | Tamanho mínimo permitido em folhas                                     |
| `max_features`          | Número máximo de atributos avaliados por split                         |
| `criterion`             | Métrica de impureza (`gini`, `entropy`)                               |
| `class_weight`          | Compensação por classes desbalanceadas                                |

## 📈 Avaliação de Performance

Usa-se métricas de **classificação supervisionada** padrão:

- **Acurácia**
- **Precision / Recall / F1**
- **Matriz de Confusão**
- **ROC AUC** (quando possível)
- **Log-loss** (se usar probabilidades)

---

## 📌 Conclusão

As **árvores de classificação** são ferramentas extremamente poderosas para **tomada de decisão supervisionada**. Elas combinam **simplicidade operacional com uma expressividade lógica rica**, tornando-se especialmente úteis em contextos de:

- Interpretação de modelos
- Prototipagem rápida
- Ambientes regulatórios

🎯 Em pipelines robustos de ciência de dados, as árvores puras são frequentemente usadas como **base para ensembles mais sofisticados** (como Random Forest e XGBoost), mas sua compreensão é **fundamental para domínio pleno de algoritmos preditivos**.
