# 🔢 Matrizes e Vetores — Fundamentos de Álgebra Linear para Ciência de Dados

Vetores e matrizes são estruturas centrais na matemática aplicada e são **pilares fundamentais de algoritmos em ciência de dados, machine learning e deep learning**. Eles representam dados, modelos e operações numéricas de forma eficiente e vetorializada.

---

## 📌 1. Conceito Teórico

### ✅ Vetores

Um **vetor** é uma sequência ordenada de números, representando um ponto ou direção em um espaço n-dimensional.

- Vetor linha: $\mathbf{x} = [x_1, x_2, ..., x_n]$
- Vetor coluna:
  $$
  \mathbf{x} = 
  \begin{bmatrix}
  x_1 \\
  x_2 \\
  \vdots \\
  x_n
  \end{bmatrix}
  $$

### ✅ Matrizes

Uma **matriz** é uma **tabela de números organizada em linhas e colunas**, usada para representar **sistemas de equações, transformações lineares e dados multidimensionais**.

- Matriz $A_{m \times n}$: m linhas e n colunas
  $$
  A = \begin{bmatrix}
  a_{11} & a_{12} & \dots & a_{1n} \\
  a_{21} & a_{22} & \dots & a_{2n} \\
  \vdots & \vdots & \ddots & \vdots \\
  a_{m1} & a_{m2} & \dots & a_{mn}
  \end{bmatrix}
  $$

---

## ⚙️ 2. Características e Definições Formais

| Conceito               | Definição Matemática                                 |
|------------------------|-------------------------------------------------------|
| **Transposta**         | $A^T$ → troca linhas por colunas                    |
| **Multiplicação de matrizes** | $A_{m \times n} \cdot B_{n \times p} = C_{m \times p}$ |
| **Produto escalar**    | $\mathbf{u} \cdot \mathbf{v} = \sum u_i v_i$        |
| **Norma de vetor**     | $\|\mathbf{x}\|_2 = \sqrt{\sum x_i^2}$ (norma Euclidiana) |
| **Matriz identidade**  | $I_n$ → matriz com 1s na diagonal, resto 0s         |
| **Determinante**       | $\det(A)$ → escala de transformação linear          |
| **Inversa**            | $A^{-1}$ tal que $A A^{-1} = I$ (se $\det(A) \ne 0$) |

---

## 🧠 3. Aplicação Prática em Ciência de Dados

### 🔷 Representação de Dados

- Um dataset tabular com $n$ observações e $d$ variáveis é uma **matriz $X_{n \times d}$**
- Cada linha: vetor de características de uma amostra
- Cada coluna: vetor de uma feature ao longo dos exemplos

---

### 🔷 Modelagem (Regressão Linear)

A regressão linear multivariada é formulada matricialmente como:

$$
\mathbf{y} = X \beta + \varepsilon
$$

- $\mathbf{y}$: vetor de saídas (n × 1)
- $X$: matriz de atributos (n × d)
- $\beta$: vetor de coeficientes (d × 1)

Estimativa dos coeficientes via **mínimos quadrados**:

$$
\hat{\beta} = (X^T X)^{-1} X^T \mathbf{y}
$$

---

### 🔷 PCA (Análise de Componentes Principais)

- Baseia-se em **autovalores e autovetores** de matrizes de covariância
- Reduz a dimensionalidade preservando a variância

---

### 🔷 Redes Neurais e Deep Learning

- Camadas de redes são **multiplicações matriciais** seguidas de não linearidades
- Dados, pesos e ativações são vetores/matrizes

---

## 🔁 4. Diferenças e Relações com Conceitos Relacionados

| Conceito         | Vetor                                   | Matriz                                    |
|------------------|------------------------------------------|--------------------------------------------|
| Estrutura        | Lista ordenada de números (1D)           | Tabela bidimensional de números (2D)       |
| Aplicação comum  | Representar uma observação ou variável   | Representar múltiplas observações/variáveis |
| Operações        | Produto escalar, norma, ângulo           | Produto, transposição, decomposição        |

---

## ✅ 5. Vantagens e Limitações no Contexto Aplicado

| Aspecto             | Vantagens                                             | Limitações / Cuidados                          |
|---------------------|------------------------------------------------------|------------------------------------------------|
| Eficiência           | Computação vetorizada → alta performance             | Pode consumir muita memória com grandes matrizes |
| Expressividade       | Permite formulações matemáticas concisas             | Complexidade algébrica aumenta em grandes modelos |
| Base de frameworks   | Numpy, pandas, TensorFlow, PyTorch usam essas estruturas | Cálculo de inversas pode ser instável numericamente |

## 📈 6. Visualizações e Ferramentas Comuns

### 🔧 Python/Numpy

```python
import numpy as np

A = np.array([[1, 2], [3, 4]])
x = np.array([1, 0])

A.T         # transposta
np.dot(A, x)  # multiplicação matriz-vetor
np.linalg.inv(A)  # inversa
np.linalg.norm(x)  # norma
```

## 📚 Conclusão

Vetores e matrizes são **estruturas essenciais e onipresentes na ciência de dados moderna**. Elas permitem:

- Eficiência computacional por vetorização
- Formulação matemática clara de modelos
- Implementações robustas e escaláveis

⚠️ Um cientista de dados que domina álgebra linear terá maior facilidade em:
- Entender modelos complexos (PCA, regressão, redes neurais)
- Diagnosticar problemas numéricos
- Otimizar pipelines analíticos

```python
# Exemplo: Resolvendo regressão linear por álgebra matricial
X = np.array([[1, 1], [1, 2], [1, 3]])
y = np.array([1, 2, 3])

# β = (XᵀX)^(-1)Xᵀy
beta_hat = np.linalg.inv(X.T @ X) @ X.T @ y
```
