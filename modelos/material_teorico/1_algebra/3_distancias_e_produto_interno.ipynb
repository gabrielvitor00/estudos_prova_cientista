# 📏 Distâncias e Produto Interno — Álgebra Aplicada à Ciência de Dados

Em ciência de dados, **distâncias** e **produto interno** (ou escalar) são ferramentas essenciais para **medir similaridade ou diferença entre observações**, realizar **projeções vetoriais** e alimentar algoritmos de **agrupamento**, **classificação**, **redução de dimensionalidade**, entre outros.

---

## 📌 1. Conceito Teórico

### ✅ Distância

É uma **função que mede o grau de dissimilaridade** entre dois vetores no espaço vetorial. Para ser considerada uma "distância válida", deve satisfazer os seguintes critérios (métrica):

- $d(x, y) \geq 0$ (não-negatividade)
- $d(x, y) = 0 \iff x = y$ (identidade)
- $d(x, y) = d(y, x)$ (simetria)
- $d(x, z) \leq d(x, y) + d(y, z)$ (desigualdade triangular)

### ✅ Produto Interno (ou Escalar)

O **produto interno** entre dois vetores $x$ e $y$ mede a **projeção** de um vetor sobre o outro, refletindo **semelhança na direção**:

$$
\langle x, y \rangle = \sum_{i=1}^n x_i y_i = x^T y
$$

Também pode ser interpretado como:

$$
\langle x, y \rangle = \|x\| \cdot \|y\| \cdot \cos(\theta)
$$

Onde $\theta$ é o ângulo entre os vetores.

## 📐 2. Principais Distâncias (com fórmulas)

| Nome                      | Fórmula                                                                 | Aplicações Típicas                                  |
|---------------------------|-------------------------------------------------------------------------|-----------------------------------------------------|
| **Euclidiana (L2)**       | $\|x - y\|_2 = \sqrt{ \sum (x_i - y_i)^2 }$                           | KNN, clustering, modelos geométricos                 |
| **Manhattan (L1)**        | $\|x - y\|_1 = \sum |x_i - y_i|$                                       | LASSO, caminhos ortogonais                          |
| **Minkowski (L_p)**       | $\|x - y\|_p = \left( \sum |x_i - y_i|^p \right)^{1/p}$                | Generalização da Euclidiana e Manhattan             |
| **Coseno (semelhança)**   | $\cos(\theta) = \frac{x^T y}{\|x\| \cdot \|y\|}$                       | NLP, recomendação, vetores de embeddings            |
| **Jaccard**               | $J(A, B) = \frac{|A \cap B|}{|A \cup B|}$                              | Dados binários, análise de conjuntos                |
| **Hamming**               | Nº de posições diferentes em vetores binários                          | Erros em transmissão, strings codificadas           |
| **Mahalanobis**           | $\sqrt{(x - \mu)^T \Sigma^{-1} (x - \mu)}$                            | Outlier detection, dados com correlação entre features |

## 🧠 3. Aplicações em Ciência de Dados

| Tarefa                             | Uso de Distâncias / Produto Interno                                |
|-----------------------------------|---------------------------------------------------------------------|
| **KNN / KMeans**                  | Baseado na distância entre amostras (geralmente Euclidiana)        |
| **Redes neurais (embeddings)**    | Similaridade por produto interno ou cosseno                        |
| **PCA / SVD**                     | Produto interno define projeções e decomposição                    |
| **Sistemas de recomendação**      | Similaridade cosseno entre vetores de usuário/item                 |
| **Clustering hierárquico**        | Cálculo repetido de distâncias entre clusters                      |
| **Análise de texto (TF-IDF)**     | Similaridade vetorial com cosseno                                  |
| **Análise de outliers**           | Distância de Mahalanobis, distância ao centro                      |

## 🔁 4. Diferenças entre Distância e Produto Interno

| Aspecto                 | Distância                                   | Produto Interno                                |
|-------------------------|---------------------------------------------|------------------------------------------------|
| Mede                   | Dissimilaridade                             | Alinhamento / semelhança                       |
| Valor baixo indica     | Semelhança                                  | Fraca projeção / ortogonalidade                |
| Valor alto indica      | Diferença maior                              | Alinhamento forte (cos(θ) próximo de 1)        |
| Escalável diretamente? | Sim (com normalização)                      | Não, sensível a escala                         |
| Comum em               | Clustering, KNN, outliers                   | PCA, word embeddings, análise de correlação    |

## ✅ 5. Vantagens e Limitações no Contexto Aplicado

| Propriedade                | Vantagens                                            | Limitações / Armadilhas                                  |
|----------------------------|-----------------------------------------------------|-----------------------------------------------------------|
| **Distância Euclidiana**   | Intuitiva, geométrica                               | Sensível à escala → requer normalização                   |
| **Distância de Mahalanobis** | Leva correlação em conta                          | Requer inversa da matriz de covariância (pode ser instável) |
| **Cosseno**                | Robusta à magnitude absoluta                        | Insensível a diferenças absolutas (boas em texto, ruins em escala absoluta) |
| **Produto Interno**        | Projeção interpretável, base de SVD, PCA            | Não mede distância, pode ser alto apenas por magnitude     |

## 📐 7. Relação com Pré-processamento e Modelagem

| Etapa / Técnica              | Uso relacionado com Distâncias / Produto Interno           |
|------------------------------|-------------------------------------------------------------|
| **Padronização (z-score)**   | Essencial antes de usar distâncias Euclidianas             |
| **PCA**                      | Usa matriz de covariância (produto interno)                |
| **Feature Selection**        | Remoção de features redundantes → melhora cálculo de distâncias |
| **Validadores de Clustering**| Ex: silhouette score → depende de distância intra/intergrupo |
| **Word2Vec / BERT embeddings** | Produto interno entre vetores representa similaridade semântica |

---

## 🧠 8. Dicas Práticas

- Sempre **normalize ou padronize os dados** antes de calcular distâncias.
- Para **dados esparsos ou com diferentes magnitudes**, prefira **distância cosseno ou Mahalanobis**.
- Em sistemas de recomendação, usar **similaridade cosseno** ou **produto escalar** é padrão.
- Em **modelos baseados em vizinhos ou distância**, verifique se a métrica escolhida está alinhada com o comportamento esperado dos dados.
