# 📊 K-Means – Guia Ultra-Robusto de Clusterização para Ciência de Dados

O **K-Means** é um algoritmo de **clusterização não supervisionada** amplamente utilizado para agrupar observações com base em sua similaridade. Ele se destaca por sua simplicidade computacional e eficiência em grandes volumes de dados. Apesar disso, **sua fundamentação matemática e suas implicações práticas** exigem atenção rigorosa para aplicação correta em projetos de ciência de dados.

---

## 📘 Conceito Teórico

O K-Means visa particionar um conjunto de dados em **K clusters distintos**, minimizando a **distância intra-cluster** (distância entre os pontos e o centro do cluster) e maximizando a **separação inter-cluster**.

O conceito central é o **centróide**, que representa o **"centro de massa"** (média vetorial) de um cluster. O centróide **não é necessariamente um ponto real dos dados**, mas sim um vetor médio das observações atribuídas ao cluster.

### 🧠 O que é um centróide?

Seja um cluster $C_i$ com $n_i$ pontos em $\mathbb{R}^d$, o centróide $\mu_i$ é definido como:

$$
\mu_i = \frac{1}{n_i} \sum_{x_j \in C_i} x_j
$$

Ou seja, cada componente do centróide é a **média aritmética** da respectiva coordenada de todos os pontos no cluster.

---

## 🔧 Características Principais e Definições Formais

### 🧮 Objetivo do Algoritmo

Minimizar a **função de custo (distância quadrática total)**:

$$
J = \sum_{i=1}^{K} \sum_{x_j \in C_i} \| x_j - \mu_i \|^2
$$

- $J$: função de distorção (inércia total)
- $x_j$: ponto de dado
- $\mu_i$: centróide do cluster $C_i$
- $\| \cdot \|$: norma Euclidiana (L2)

### 🔁 Passos do Algoritmo K-Means

1. **Inicialização**:
   - Escolher K centróides iniciais (aleatórios ou via KMeans++)
2. **Atribuição**:
   - Atribuir cada ponto $x_j$ ao cluster $C_i$ com centróide mais próximo:

     $$
     C_i = \{ x_j : \|x_j - \mu_i\|^2 \leq \|x_j - \mu_l\|^2, \forall l \}
     $$

3. **Atualização**:
   - Recalcular centróides como média dos pontos atribuídos:

     $$
     \mu_i = \frac{1}{|C_i|} \sum_{x_j \in C_i} x_j
     $$

4. **Critério de parada**:
   - Convergência dos centróides (sem mudanças relevantes) ou máximo de iterações

### 📌 Hiperparâmetros

| Parâmetro   | Descrição                                                                 |
|-------------|---------------------------------------------------------------------------|
| `K`         | Número de clusters desejado                                                |
| `init`      | Método de inicialização (`random`, `k-means++`)                            |
| `max_iter`  | Número máximo de iterações                                                 |
| `tol`       | Tolerância para mudança mínima nos centróides                              |
| `n_init`    | Número de vezes que o algoritmo será rodado com diferentes inicializações  |

## ✅ Vantagens, ❌ Limitações e ⚠️ Armadilhas Comuns

### ✅ Vantagens

- Simples, intuitivo e interpretável
- Eficiente em grandes bases (linear em n)
- Rápido com K pequeno e features normalizadas
- Compatível com paralelização e mini-batches

### ❌ Limitações

- Assume clusters esféricos e de mesma densidade
- Requer número de clusters definido a priori
- Converge para **mínimos locais** (resultado pode variar por inicialização)
- Extremamente sensível a outliers e a **features mal escaladas**

### ⚠️ Armadilhas Comuns

| Erro Comum                       | Consequência Prática                                            |
|----------------------------------|-----------------------------------------------------------------|
| Usar sem padronização (scaling)  | Variáveis com maior escala dominam distância Euclidiana         |
| Escolher K arbitrário            | Clusterização com pouca relevância semântica                    |
| Ignorar múltiplas inicializações | Pode escolher mínimo local ruim                                 |
| Interpretar como classificação   | K-Means não generaliza para novos dados sem reexecução completa |
| Aplicar em dados não vetoriais   | Requer representação numérica e métrica de distância adequada   |

---

## 🔍 Interpretação Prática e Implicações Estatísticas

- **Centróide ≠ ponto real**, mas representa a média vetorial do grupo
- A **distância ao centróide** pode ser usada como indicador de similaridade ou anomalia
- Clusters podem ser usados como:
  - **Rótulo pseudo-supervisionado**
  - **Feature categórica (ex: `cluster_id`)**
- Se mal definido, pode causar **overfitting**, especialmente se os clusters forem usados como feature sem validação
- Avaliação da **estabilidade dos clusters** é essencial: clusters variam com diferentes seeds, subsets ou normalizações

|----------------------|-------------------------------------------------------------------------|

| **Gap Statistic**    | Compara inércia observada com esperada por dados aleatórios              |
| **BIC/AIC com GMM**  | Alternativa probabilística que ajuda na escolha de K                     |

## 🎯 K-Means com e sem Variável Target: Clusterização Não Supervisionada vs Supervisionada

Embora o **K-Means** seja, por definição, um algoritmo **não supervisionado**, ele pode ser incorporado de maneira estratégica em pipelines **supervisionados** para enriquecer modelos preditivos. A distinção entre esses dois contextos — **com ou sem variável target** — é fundamental para entender **quando e como aplicar K-Means corretamente**.

---

### 🔍 1. K-Means Puro (Não Supervisionado)

#### ✅ Características:
- **Não há variável target (rótulo ou saída)**.
- O objetivo é **descobrir padrões ou estruturas latentes** nos dados.
- Agrupamentos são baseados apenas nas **features independentes**.

#### 📌 Exemplo:
```python
from sklearn.cluster import KMeans

X = df[["tempo_site", "qtd_compras", "gasto_total"]]
kmeans = KMeans(n_clusters=4)
df["cluster"] = kmeans.fit_predict(X)
```

**Aplicações:**
- Segmentação de clientes
- Agrupamento de produtos
- Compressão de imagens
- Análise exploratória inicial
- Redução de ruído ou outliers

**⚠️ Implicações:**
- Sem validação direta com métrica supervisionada (ex: accuracy, AUC)
- Clusters podem não ter relevância com a variável alvo
- Ideal para entendimento da estrutura dos dados, não para previsão

---

### 🧠 2. K-Means com Variável Target (Uso Supervisionado)
Embora K-Means não use a variável target durante o treino, ele pode ser combinado com modelos supervisionados de três formas:

#### 🔧 A) K-Means como Engenharia de Features
Os clusters são usados como variável categórica adicional (`cluster_id`) no modelo supervisionado.

**Exemplo:**
```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.cluster import KMeans

# K-Means
X_kmeans = df[["idade", "salario", "score_credito"]]
df["cluster"] = KMeans(n_clusters=5).fit_predict(X_kmeans)

# Modelo supervisionado
X_final = df[["idade", "salario", "score_credito", "cluster"]]
y = df["inadimplente"]
modelo = RandomForestClassifier().fit(X_final, y)
```

**Vantagens:**
- Aumenta a capacidade discriminativa do modelo
- Agrupa padrões locais não capturados por modelos lineares
- Reduz necessidade de interação cruzada entre features

**Cuidados:**
- Clusters devem ser recalculados na base de teste com o mesmo modelo KMeans treinado
- Requer avaliação se `cluster_id` é informativo para a variável target
- Se os clusters forem sobreajustados ao conjunto de treino, pode introduzir leakage indireto

## 📊 Comparativo Geral: K-Means Sem vs Com Target

| Aspecto                | K-Means (Sem Target)         | K-Means (Com Target)                        |
|------------------------|------------------------------|---------------------------------------------|
| Tipo de aprendizado    | Não supervisionado           | Híbrido (auxilia modelo supervisionado)     |
| Usa a variável alvo    | ❌ Nunca                     | ✅ Indiretamente (via uso posterior)         |
| Validação              | Interna (silhouette, inertia)| Externa (AUC, accuracy do modelo supervisionado) |
| Objetivo               | Descobrir padrões estruturais| Melhorar predição ou interpretabilidade     |
| Risco de overfitting   | Baixo                        | ✅ Possível se clusters forem sobreajustados |
| Papel nos pipelines    | EDA, compressão, segmentação | Feature engineering, modelagem segmentada   |

---

## 🧠 Conclusão Prática
K-Means não supervisionado é útil para entender a estrutura latente dos dados, fazer exploração e segmentação inicial.

K-Means supervisionado (indiretamente) é poderoso para:
- Adicionar variáveis discriminativas via cluster
- Criar submodelos mais especializados
- Aumentar explicabilidade de padrões em dados tabulares

🎯 A escolha do uso com ou sem target depende do objetivo do projeto: se for segmentar dados para interpretação, use K-Means puro; se for otimizar modelos preditivos, use clusters como feature auxiliar — sempre validando sua utilidade estatística.
