# 🌲 Algoritmos de Clusterização Hierárquica – Guia Ultra-Robusto para Ciência de Dados

A **clusterização hierárquica** é uma abordagem **não supervisionada** que constrói uma **estrutura em árvore** (dendrograma) de agrupamentos. Ao contrário de algoritmos como K-Means, que geram uma partição plana do conjunto de dados, os métodos hierárquicos constroem **níveis aninhados de clusters**, permitindo **exploração multirresolução** da estrutura dos dados — desde agrupamentos globais até subestruturas locais.

Essa característica é particularmente útil em problemas com **ambiguidade na escala dos clusters**, **desconhecimento prévio do número de grupos**, ou **dados com hierarquia natural (ex: taxonomias, documentos, genômica, etc.)**.

---

## 📘 Conceito Teórico

A clusterização hierárquica busca organizar um conjunto de dados em uma **árvore de relações de similaridade**, onde:
- Cada **nó folha** representa uma amostra individual.
- Cada **nó interno** representa um agrupamento (cluster) de amostras.
- O **nível da árvore** reflete o grau de similaridade entre os objetos agrupados.

Existem duas variantes principais:

| Tipo                        | Descrição                                                                 |
|-----------------------------|---------------------------------------------------------------------------|
| **Aglomerativa (bottom-up)**| Inicia com cada ponto em um cluster individual e **funde** clusters iterativamente |
| **Divisiva (top-down)**     | Inicia com todos os pontos em um único cluster e **divide** recursivamente       |

A abordagem aglomerativa é a mais utilizada e suportada por bibliotecas como `scikit-learn`.

---

## 🔧 Características e Definições Formais

### 🚧 Etapas do algoritmo aglomerativo:

1. **Inicialização**: cada amostra inicia como um cluster separado.
2. **Cálculo de distâncias**: calcula-se a distância entre todos os pares de clusters.
3. **Fusão**: os dois clusters mais próximos são fundidos.
4. **Atualização de distâncias**: atualiza-se a matriz de distâncias.
5. **Repetição**: repete-se até todos os pontos estarem em um único cluster.

### 📐 Métodos de ligação (linkage)

Define como calcular a distância entre dois clusters $A$ e $B$:

| Linkage      | Fórmula                                                        | Características                        |
|--------------|----------------------------------------------------------------|-----------------------------------------|
| **Single**   | $\displaystyle \min_{a \in A,\; b \in B} d(a, b)$           | Sensível a outliers, forma longa        |
| **Complete** | $\displaystyle \max_{a \in A,\; b \in B} d(a, b)$           | Mais compacto, propenso a encadeamento  |
| **Average**  | $\displaystyle \frac{1}{\lvert A \rvert \lvert B \rvert} \sum\limits_{a \in A,\; b \in B} d(a, b)$ | Balanceado, menos sensível a ruído      |
| **Ward**     | Minimiza variância total dentro dos clusters após fusão         | Recomendado para dados euclidianos      |

### 🧮 Distâncias suportadas:
- Euclidiana (comum)
- Manhattan
- Cosseno
- Correlação
- Qualquer métrica definida sobre pares (inclusive customizadas)

## 🔬 Diferenças com Outros Algoritmos de Clusterização

| Algoritmo        | Tipo de Cluster | Requer K? | Suporta outliers? | Forma dos Clusters         | Produz hierarquia? |
|------------------|-----------------|-----------|-------------------|----------------------------|--------------------|
| **Hierárquico**  | Qualquer        | ❌ Não     | ⚠️ Parcialmente    | Qualquer forma             | ✅ Sim             |
| **K-Means**      | Esférico        | ✅ Sim     | ❌ Não             | Convexos                   | ❌ Não             |
| **DBSCAN**       | Denso           | ❌ Não     | ✅ Sim             | Arbitrário                 | ❌ Não             |

📌 A clusterização hierárquica **não impõe forma geométrica** aos clusters e permite **exploração em múltiplas resoluções**, diferentemente de algoritmos "planos" como K-Means.

---

## ✅ Vantagens, ❌ Limitações e ⚠️ Armadilhas Comuns

### ✅ Vantagens

- **Não exige número de clusters a priori**
- **Hierarquia de agrupamentos** com múltiplos níveis
- Flexível quanto à **métrica de distância**
- Capaz de **identificar clusters complexos** ou com forma irregular
- Excelente para **análise exploratória** e visualização com dendrogramas

### ❌ Limitações

- **Custo computacional alto**: $O(n^2)$ em tempo e espaço (inviável para n > 10.000 sem otimizações)
- Sensível à escolha da **métrica de distância** e método de linkage
- **Não lida bem com ruído**: todos os pontos são agrupados — não há noção explícita de outlier
- Não suporta atualização incremental (não é “online”)

### ⚠️ Armadilhas comuns

| Erro                              | Consequência                                                            |
|-----------------------------------|-------------------------------------------------------------------------|
| Escolher `single linkage` sem motivo | Forma clusters longos e encadeados, com pouco significado             |
| Usar sem normalizar os dados      | Distâncias distorcidas, agrupamento incorreto                           |
| Tentar aplicar em dados massivos  | Explosão de memória e lentidão insustentável                            |
| Interpretar níveis como “verdade” | Hierarquia é relativa à métrica; não é necessariamente “natural”        |

# 🧩 Métodos de Ligação e Distâncias Suportadas em Agrupamento Hierárquico

Nos algoritmos de **agrupamento hierárquico**, a definição de **proximidade** entre clusters é crítica. Essa proximidade é determinada por dois fatores:

1. 📏 **Métrica de Distância (ou Similaridade)** — como medir a distância entre pontos.
2. 🧷 **Método de Ligação (Linkage)** — como definir a distância entre grupos de pontos (clusters).

Esses dois componentes influenciam fortemente o **formato, profundidade e estabilidade dos clusters** formados.

---

## 📘 Conceito Teórico

O agrupamento hierárquico constrói uma árvore (*dendrograma*) de agrupamentos com base em **fusões sucessivas de pontos ou grupos**, escolhendo a cada passo os dois clusters mais próximos, segundo:

- Uma **métrica de distância** (ex: euclidiana)
- Uma **estratégia de ligação** (ex: complete, single, average)

---

## 📏 🔸 Métricas de Distância Suportadas

As métricas definem a **distância entre duas amostras** (proximidade ponto a ponto). As mais comuns são:

### 🔹 1. Distância Euclidiana

$$
d(x, y) = \sqrt{\sum_{i=1}^n (x_i - y_i)^2}
$$

**Ideal para:**
- Dados **contínuos** e normalizados
- Interpretabilidade geométrica

**Atenção:** Sensível à escala → **Padronizar** os dados (`StandardScaler`) antes de aplicar.

---

### 🔹 2. Distância de Manhattan (Cityblock)

$$
d(x, y) = \sum_{i=1}^n |x_i - y_i|
$$

**Ideal para:**
- Dados com **distribuições assimétricas**
- Modelos mais robustos a outliers

---

### 🔹 3. Distância de Minkowski

$$
d(x, y) = \left( \sum_{i=1}^n |x_i - y_i|^p \right)^{1/p}
$$

- Generaliza Euclidiana (p = 2) e Manhattan (p = 1)
- Permite configurar o **grau de penalização** via o parâmetro $p$

---

### 🔹 4. Distância de Coseno

$$
d_{\text{cos}}(x, y) = 1 - \frac{x \cdot y}{\|x\| \|y\|}
$$

**Ideal para:**
- Dados de **alta dimensionalidade** (ex: NLP, texto vetorizado)
- Casos em que importa mais a **orientação** do vetor do que magnitude

**Atenção:** Não considera magnitude (duas observações paralelas serão próximas mesmo com grande distância euclidiana)

---

### 🔹 5. Correlação (1 - correlação de Pearson)

$$
d(x, y) = 1 - \frac{Cov(x, y)}{\sigma_x \cdot \sigma_y}
$$

**Ideal para:**
- Clustering de **séries temporais**
- Dados com foco em **tendência de variação** e não nos valores absolutos

---

## 🧷 🔸 Métodos de Ligação (Linkage Criteria)

Os métodos de ligação determinam **como calcular a distância entre dois clusters**, considerando os pontos contidos em cada um.

---

### 🔹 1. Ligação Simples (Single Linkage)

$$
D(A, B) = \min \{ d(x, y) \mid x \in A, y \in B \}
$$

➡️ Usa a **menor distância** entre os pontos dos clusters.

**Vantagens:**
- Mantém grupos compactos conectados
- Capta estruturas **não esféricas**

**Armadilhas:**
- Pode formar **"efeito corrente"** (clusters muito alongados)
- Pouco robusto a **outliers**

---

### 🔹 2. Ligação Completa (Complete Linkage)

$$
D(A, B) = \max \{ d(x, y) \mid x \in A, y \in B \}
$$

➡️ Usa a **maior distância** entre os pontos dos clusters.

**Vantagens:**
- Gera clusters **mais compactos e esféricos**
- Resistente ao efeito corrente

**Armadilhas:**
- **Sensível a outliers** (a maior distância domina a fusão)
- Pode separar indevidamente pontos centrais

---

### 🔹 3. Ligação Média (Average Linkage)

$$
D(A, B) = \frac{1}{|A| \cdot |B|} \sum_{x \in A} \sum_{y \in B} d(x, y)
$$

➡️ Considera a **média de todas as distâncias** entre pares de pontos dos clusters.

**Vantagens:**
- Balanceia entre ligação simples e completa
- **Menos sensível a outliers**
- Preserva **forma natural** dos dados

---

### 🔹 4. Ligação Centróide (Centroid Linkage)

$$
D(A, B) = d(\mu_A, \mu_B)
$$

- Onde $\mu_A$ e $\mu_B$ são os **centroides dos clusters** A e B

**Vantagens:**
- Computacionalmente eficiente
- Útil quando se deseja clusters representados por média

**Armadilhas:**
- Pode causar **inversões** no dendrograma (clusters fundidos podem estar mais distantes do que outros níveis inferiores)

---

### 🔹 5. Ligação Ward (Ward's Method)

Minimiza a **variância intra-cluster** após a fusão.

➡️ A fusão é escolhida para **minimizar o aumento do erro quadrático total (SSE)**.

**Vantagens:**
- Gera clusters **homogêneos**
- Excelente para dados com **distribuição esférica**

**Requisitos:**
- Dados devem ser **contínuos**
- Requer uso de **distância euclidiana**

---

## 🧪 Comparativo Prático: Ligação × Formato dos Clusters

| Método   | Formato dos Clusters         | Sensível a Outliers | Comentário                    |
|----------|-----------------------------|---------------------|-------------------------------|
| Single   | Alongados, conectados        | Alta                | Tende a formar cadeias        |
| Complete | Esféricos, compactos         | Média               | Pode supersegmentar           |
| Average  | Intermediário                | Baixa               | Boa escolha padrão            |
| Centroid | Esférico (mas instável)      | Alta                | Pode inverter hierarquia      |
| Ward     | Muito compactos/esféricos    | Média               | Excelente em muitos casos     |

---

## 📊 Visualizações Comuns

- **Dendrogramas**: estrutura hierárquica de fusões de clusters
- **Matriz de distância (heatmap)**: visualização da distância entre pontos
- **Silhouette Plot**: avaliar coesão e separação dos clusters gerados

```python
import scipy.cluster.hierarchy as sch
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_iris

# Dataset
X = load_iris().data
X_std = StandardScaler().fit_transform(X)

# Dendrograma com linkage ward
plt.figure(figsize=(10, 6))
sch.dendrogram(sch.linkage(X_std, method='ward'))
plt.title("Dendrograma - Método Ward")
plt.xlabel("Amostras")
plt.ylabel("Distância Euclidiana")
plt.show()
```

---

## 🧠 Implicações Práticas para Ciência de Dados

- **Modelagem de Segmentação:** Clusters gerados podem ser usados como variáveis (segmentos) em modelos supervisionados.
- **Redução de Dimensionalidade:** Aplicar PCA antes de clustering pode estabilizar métricas e linkage.
- **Detecção de Outliers:** Clusters com poucos pontos ou distantes dos demais podem indicar outliers.
- **Engenharia de Variáveis:** Variáveis como distância ao centroide do cluster podem ser usadas como features.

---

## ⚠️ Armadilhas e Considerações

- O resultado final depende fortemente da combinação **métrica + linkage**.
- Dados não padronizados distorcem distâncias.
- Agrupamento hierárquico não escala bem: custo computacional é $O(n^3)$.

---

## ✅ Recomendações Práticas

| Cenário                              | Métrica + Ligação Ideal                |
|--------------------------------------|----------------------------------------|
| Dados normalizados, esféricos        | Euclidiana + Ward                      |
| Dados de texto (vetores esparsos)    | Cosseno + Média                        |
| Agrupamento de séries temporais      | Correlação + Média ou Ward              |
| Dados com outliers                   | Manhattan + Média                      |
| Segmentação de clientes (RFM, etc.)  | PCA → Euclidiana + Complete ou Ward    |

---

## 📌 Conclusão

A escolha do método de ligação e da métrica de distância afeta profundamente o agrupamento hierárquico. Para modelagem robusta e interpretável, é fundamental:

- Padronizar os dados
- Escolher métrica compatível com o domínio
- Validar os clusters com métricas como *silhouette score*
- Visualizar e interpretar o dendrograma

> 🎯 **Método de ligação + métrica de distância = comportamento geométrico dos clusters. Dominar essa interação é essencial para análises não supervisionadas confiáveis.**
