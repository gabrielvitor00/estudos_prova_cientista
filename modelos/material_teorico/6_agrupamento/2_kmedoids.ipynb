# 💎 K-Medoids – Guia Ultra-Robusto para Clusterização com Resistência a Outliers

O **K-Medoids** é um algoritmo de clusterização **não supervisionada**, semelhante ao K-Means, mas com uma diferença crucial: ao invés de utilizar o **centroide (média vetorial)** como representante do cluster, o K-Medoids usa o **medoide** — um ponto **real do conjunto de dados**, mais robusto a **outliers** e **distribuições não simétricas**. Essa propriedade o torna especialmente útil em situações com ruído, valores extremos ou quando a métrica de similaridade não se ajusta bem à média.

---

## 📘 Conceito Teórico

K-Medoids agrupa os dados em **K clusters**, minimizando a **distância total entre os pontos e o medoide de seu respectivo cluster**. Ao contrário do K-Means, que calcula médias (sensível a outliers), o K-Medoids busca encontrar o ponto **mais central** (medoide) entre os membros de cada grupo — aquele cuja **distância total para os outros membros do cluster é mínima**.

---

## 🔧 Características Principais e Definições Formais

### 🔹 O que é um **Medoide**?

Dado um conjunto $C = \{x_1, x_2, ..., x_n\}$, o **medoide** $x^* \in C$ é definido como:

$$
x^* = \arg\min_{x_i \in C} \sum_{x_j \in C} d(x_i, x_j)
$$

Ou seja, é o ponto **cuja soma de distâncias para todos os outros pontos no cluster é mínima**.

---

### 🔁 Etapas do Algoritmo PAM (Partitioning Around Medoids)

1. **Inicialização**: selecionar K pontos aleatórios como medoides iniciais.
2. **Atribuição**: cada ponto é atribuído ao cluster do medoide mais próximo (baseado em distância, ex: Euclidiana ou Manhattan).
3. **Atualização**: para cada cluster, verifica se a substituição do medoide por outro ponto do cluster reduziria o custo total. Se sim, realiza a troca.
4. **Iteração**: repetir as etapas de atribuição e atualização até não haver mais melhoria.

### 📐 Função de Custo

$$
\text{Total Cost} = \sum_{i=1}^{K} \sum_{x_j \in C_i} d(x_j, m_i)
$$

- $m_i$: medoide do cluster $C_i$
- $d(\cdot, \cdot)$: métrica de distância (pode ser qualquer métrica de dissimilaridade)

## 🧠 Aplicações Práticas em Ciência de Dados

| Cenário Real                              | Utilização do K-Medoids                                                      |
|-------------------------------------------|------------------------------------------------------------------------------|
| **Segmentação de Clientes com outliers**  | Agrupa clientes sem ser distorcido por valores extremos                      |
| **Clusterização de Sinais de Sensores**   | Resistente a picos ou ruídos frequentes                                      |
| **Análise de dados geográficos**          | Usado com distância geodésica ou haversine para clusters reais e robustos    |
| **Agrupamento de séries temporais**       | Pode usar medidas como DTW (Dynamic Time Warping) como distância             |
| **Clusterização com métricas não-euclidianas** | Permite flexibilidade em métricas de dissimilaridade                    |

📌 Exemplo: Em análise de fraude bancária, o K-Medoids pode segmentar clientes sem ser influenciado por transações atípicas de fraude já conhecidas.

---

## 🔬 Diferenças Entre K-Medoids e Algoritmos Relacionados

| Algoritmo         | Representante de Cluster | Requer ponto real? | Sensível a Outliers | Tipo de Distância            |
|-------------------|-------------------------|--------------------|--------------------|------------------------------|
| **K-Means**       | Centróide (média)       | ❌ Não             | ✅ Sim             | Apenas Euclidiana            |
| **K-Medoids**     | Medoide (ponto real)    | ✅ Sim             | ❌ Não             | Qualquer métrica (flexível)  |
| **DBSCAN**        | Densidade local         | ✅ Sim             | ❌ Não             | Qualquer (com cuidado)       |
| **GMM**           | Parâmetros probabilísticos| ❌ Não            | ⚠️ Parcialmente    | Gaussiana (com covariância)  |

📌 K-Medoids é particularmente adequado quando **a média não representa bem a centralidade**, ou **a métrica Euclidiana não é apropriada**.

## ✅ Vantagens, ❌ Limitações e ⚠️ Armadilhas Comuns

### ✅ Vantagens
- Alta **robustez a outliers e ruído**
- Pode ser usado com **qualquer matriz de dissimilaridade** (flexível para dados não vetoriais)
- Medoides são **observações reais**, facilitando interpretação
- Melhor representação da centralidade em dados assimétricos

### ❌ Limitações
- **Custo computacional elevado**: $O(k(n-k)^2)$, pior que K-Means
- Não escala bem com grandes volumes de dados (acima de ~10.000 instâncias)
- **Escolha de K ainda é necessária** (hiperparâmetro fixo)
- Menor disponibilidade em bibliotecas mainstream

### ⚠️ Armadilhas Comuns

| Erro Comum                          | Consequência                                                    |
|-------------------------------------|-----------------------------------------------------------------|
| Aplicar em datasets muito grandes   | Custo computacional inviável sem amostragem                     |
| Usar sem normalização               | Métrica de distância distorcida (caso use Euclidiana)           |
| Definir K sem validação             | Clusters irrelevantes ou redundantes                            |
| Comparar resultados com K-Means     | Pode divergir totalmente se houver outliers ou clusters irregulares |

---

## 📏 Implicações para Modelagem Estatística ou Preditiva

- **Melhor separabilidade entre clusters** pode gerar **features mais significativas** quando usadas em modelos supervisionados
- Por ser mais **estável em relação a ruído**, pode ser útil para:
  - Filtragem de amostras atípicas
  - Clusterização de embeddings em NLP
  - Estudo de padrões anômalos em dados transacionais
- Permite trabalhar com dados **não numéricos vetoriais**, desde que se defina uma matriz de distância adequada (ex: Levenshtein para strings)

## 📌 Conclusão

O **K-Medoids** é uma alternativa **mais robusta, flexível e interpretável** ao K-Means, sendo especialmente útil em dados com **outliers, distribuições assimétricas ou métricas de distância complexas**. Embora seja mais custoso computacionalmente, sua aplicação pode levar a resultados **mais confiáveis e estáveis**, o que é crucial em ciência de dados aplicada.

🎯 Dominar K-Medoids permite ao cientista de dados executar segmentações mais precisas, criar **representações latentes de maior qualidade** e explorar **formas de clusterização além da média** — aproximando a análise da realidade dos dados.
