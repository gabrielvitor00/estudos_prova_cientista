# 🧱 Modelos de Banco de Dados Relacional (RDBMS) – Guia Avançado para Ciência de Dados

Os **modelos relacionais** são a estrutura clássica de armazenamento de dados **estruturados**, sustentando aplicações OLTP e projetos de ciência de dados que requerem **consistência, integridade e normalização**. Este guia cobre conceitos fundamentais e avançados com foco total em aplicações em Ciência de Dados.

---

## 📘 Conceito Teórico

O modelo relacional organiza os dados em **relações (tabelas)**, nas quais cada **linha** representa um registro (ou tupla) e cada **coluna** representa um atributo. A estrutura é fundamentada em **teoria de conjuntos** e **lógica de predicados**.

- Criado por **E. F. Codd** em 1970.

- Cada tabela é **autoconsistente** e **relacionável** a outras via **chaves**.

- Operações realizadas por meio de **álgebra relacional**, acessadas via **SQL**.

---

## 🔧 Elementos Estruturais

| Elemento             | Descrição                                                              |
|:---------------------|:-----------------------------------------------------------------------|
| **Relação (Tabela)** | Conjunto de tuplas organizadas por atributos                           |
| **Tupla**            | Instância de dados (linha)                                              |
| **Atributo**         | Coluna que define uma propriedade da entidade                           |
| **Chave Primária**   | Identificador único da tupla (não nulo, único)                          |
| **Chave Estrangeira**| Referência a uma PK de outra tabela, criando relacionamento             |
| **Domínio**          | Conjunto de valores válidos para um atributo                            |
| **Esquema**          | Estrutura lógica da base (nome das tabelas, colunas, relacionamentos)   |

---

## 🧮 Chaves Compostas

Uma **chave composta** é formada por dois ou mais atributos que, combinados, **identificam unicamente** uma tupla.

### 📌 Exemplo:

```sql
PRIMARY KEY (id_pedido, id_produto)
```

✅ Usado em:
- Modelos associativos (tabelas N:N)
- Relacionamentos sem chave artificial

⚠️ Requer atenção extra para garantir unicidade lógica e evitar duplicações acidentais.

---

## 📏 Normalização

A normalização é o processo de decompor tabelas para:
- Reduzir redundância
- Evitar anomalias de atualização
- Melhorar manutenção e consistência


| Forma Normal | Requisitos                                 | Benefício                        |
|:-------------|:--------------------------------------------|:----------------------------------|
| 1FN          | Elimina grupos repetitivos                  | Estrutura tabular válida          |
| 2FN          | Elimina dependência parcial                 | Dados atomicamente organizados    |
| 3FN          | Elimina dependência transitiva              | Redução de redundância            |
| BCNF         | Toda dependência funcional usa chave candidata | Estrutura ainda mais robusta  |



🔁 Desnormalização pode ser útil em cenários analíticos para reduzir o número de joins.

## ⚙️ Índices

Os índices melhoram o tempo de consulta em colunas frequentemente usadas em:
- Filtros (`WHERE`)
- Junções (`JOIN`)
- Ordenações (`ORDER BY`)

**Tipos comuns:**

| Tipo de Índice | Aplicação                                 |
|:--------------|:-------------------------------------------|
| B-Tree        | Busca exata ou intervalar (padrão em PK)   |
| Hash          | Busca exata (sem range)                    |
| Full-text     | Pesquisa textual                           |
| Bitmap        | Colunas com poucos valores distintos        |


⚠️ Muitos índices podem impactar inserções e atualizações. Use com estratégia.

---

## 🔍 Subconsultas SQL Avançadas

Subconsultas (subqueries) são essenciais para construir consultas complexas e expressar lógica analítica sem múltiplos joins.

**Exemplos:**

- 🧠 Scalar subquery:

```sql
SELECT nome, salario
FROM funcionario
WHERE salario > (
    SELECT AVG(salario) FROM funcionario
)
```

- 📊 Subquery correlacionada:

```sql
SELECT nome
FROM aluno a
WHERE EXISTS (
    SELECT 1 FROM matricula m WHERE m.id_aluno = a.id AND m.ano = 2024
)
```

- 🚫 NOT IN vs NOT EXISTS:
  - `NOT EXISTS` é mais seguro (não falha com NULL)
  - Evite `NOT IN (SELECT ...)` com colunas que podem conter NULL

## 🗺️ Modelagem Dimensional (Data Warehouses)

### 🌟 Modelo Estrela (Star Schema)

| Tabela   | Tipo      | Descrição                                 |
|:---------|:----------|:------------------------------------------|
| Fato     | Central   | Contém medidas (faturamento, qtd, etc)     |
| Dimensão | Periférica| Contém atributos descritivos (tempo, produto, cliente) |

- Simples e eficiente para leitura analítica
- Joins simples e boa performance

### ❄️ Modelo Floco de Neve (Snowflake)
- Dimensões são normalizadas
- Melhora integridade, mas aumenta a complexidade dos joins
- 🔁 Escolha depende de trade-off entre performance e integridade.

## 🚀 Otimização de Consultas

**Técnicas comuns:**
- `EXPLAIN PLAN`: analisa plano de execução da query
- Criação de índices apropriados
- Evitar `SELECT *`
- Utilizar `LIMIT` em consultas exploratórias
- Usar joins explícitos e evitar subqueries desnecessárias

**Boas práticas:**

| Problema                | Solução recomendada                                 |
|:------------------------|:---------------------------------------------------|
| Leitura lenta           | Índices em colunas de filtro/junção                 |
| Alta cardinalidade      | Evite bitmap indexes                                |
| Uso intensivo de joins  | Reduzir colunas selecionadas (`SELECT cols`)        |
| Escaneamento completo   | Criação de índice baseado em análise `EXPLAIN`      |
