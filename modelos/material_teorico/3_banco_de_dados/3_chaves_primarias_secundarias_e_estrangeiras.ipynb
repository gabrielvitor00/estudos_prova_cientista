# 🔑 Chaves Primárias, Estrangeiras e Secundárias – Guia Avançado para Ciência de Dados Relacional

As **chaves** são o núcleo lógico de todo modelo de banco de dados relacional. Elas definem **unicidade, relacionamento e indexação**, assegurando integridade referencial entre entidades e permitindo junções confiáveis entre tabelas. A correta definição e compreensão de **chaves primárias (PK)**, **estrangeiras (FK)** e **secundárias (SK)** é crítica para a integridade de dados e o sucesso de projetos analíticos.

---

## 📘 Conceito Teórico

No modelo relacional:

- Uma **chave primária (PK)** identifica **de forma única** cada tupla em uma relação.
- Uma **chave estrangeira (FK)** estabelece **relacionamentos entre tabelas**, apontando para a chave primária de outra.
- Uma **chave secundária (SK)** é uma coluna que **não identifica unicamente** tuplas, mas é comumente consultada e **indexada** para performance.

Todas derivam do conceito de **dependência funcional**: uma PK determina univocamente o restante da tupla.

---

## 🔧 Elementos e Definições Formais

| Tipo de Chave         | Definição Técnica                                                                              | SQL                                             |
|---------------------- |----------------------------------------------------------------------------------------------|-------------------------------------------------|
| **Primária (PK)**     | Atributo(s) que identificam unicamente uma tupla. Deve ser `NOT NULL` e `UNIQUE`.             | `PRIMARY KEY (col)`                             |
| **Estrangeira (FK)**  | Atributo que referencia a PK de outra tabela. Garante integridade referencial.                | `FOREIGN KEY (col) REFERENCES outra_tabela(pk)` |
| **Secundária (SK)**   | Atributo alternativo usado para busca (pode ser `NULL` ou duplicado). Normalmente indexado.   | `CREATE INDEX ON (col)`                         |

📌 Toda tabela deve possuir **uma e apenas uma PK**. Pode ter múltiplas FKs e múltiplas SKs (índices).

---

## 🧱 Exemplos Práticos

### 1. Chave Primária

```sql
CREATE TABLE cliente (
    id_cliente INT PRIMARY KEY,
    nome VARCHAR(100),
    email VARCHAR(150)
);
```

### 2. Chave Estrangeira

```sql
CREATE TABLE pedido (
    id_pedido INT PRIMARY KEY,
    id_cliente INT,
    FOREIGN KEY (id_cliente) REFERENCES cliente(id_cliente)
);
```

### 3. Chave Secundária (índice auxiliar)

```sql
CREATE INDEX idx_pedido_data ON pedido(data_pedido);
```

## 🧠 Aplicações em Projetos de Ciência de Dados

| Etapa                | Utilização das Chaves                                                                 |
|----------------------|--------------------------------------------------------------------------------------|
| Ingestão de dados    | Chaves identificam duplicatas, inconsistências e ajudam no mapeamento de joins       |
| Engenharia de features | Junções entre tabelas fact/dimensão via PK ↔ FK                                   |
| Validação de integridade | Detecção de anomalias de relacionamento (FK órfãs, registros duplicados)        |

| Construção de datasets | Combinação de tabelas associativas por PKs compostas (ex: relacionamento N:M)     |

---

## 🔍 Diferenças entre Tipos de Chave

| Aspecto             | Chave Primária (PK) | Chave Estrangeira (FK) | Chave Secundária (SK)                |
|---------------------|---------------------|------------------------|--------------------------------------|
| Finalidade          | Unicidade           | Relacionamento entre tabelas | Indexação / performance de consulta |
| Obrigatoriedade     | Obrigatória por tabela | Opcional (mas comum em normalização) | Opcional                |
| Pode conter NULL?   | Não                 | Sim (se relacionamento for opcional) | Sim                      |
| Pode repetir valores?| Não                | Sim                    | Sim                                  |
| Associada a índice? | Sim (índice automático) | Não necessariamente   | Sim (requer CREATE INDEX)            |

---

## ✅ Vantagens no Contexto Relacional
- **Integridade garantida:** PKs e FKs garantem consistência de dados entre entidades.
- **Performance em joins:** joins baseados em PK/FK são otimizáveis pelo SGBD.
- **Evita duplicidade:** PKs e SKs com índice evitam inserções redundantes.
- **Fundamenta modelagem relacional:** permite representar relacionamentos 1:N, N:M, 1:1 de forma formal.

## ⚠️ Limitações e Armadilhas

| Erro Comum                        | Consequência                                         |
|------------------------------------|------------------------------------------------------|
| Criar PK em coluna não única       | Permite registros inconsistentes e invalida modelagem |
| Esquecer FK entre tabelas          | Pode gerar dados órfãos ou joins inválidos           |
| FK sem índice                      | Pode causar lentidão em joins e deleções             |
| SK mal utilizada (sem índice)      | Consultas em larga escala sem filtro eficaz          |
| Chave composta com semântica fraca | Dificulta manutenção e interpretação                 |

---

## 🔄 Implicações para Modelagem Estatística/Preditiva

- **Referencialidade perdida = Feature corrompida:** sem FK, você pode juntar linhas erradas.
- **Duplicidade sem PK = Distorção estatística:** duplicações afetam médias, contagens e modelos.
- **PK como identificador de unidade amostral:** central para dividir corretamente em treino/teste.
- **FK define granularidade:** o tipo de relacionamento define a unidade analítica (ex: cliente vs transação).
- **SKs facilitam segmentações rápidas:** ex: índices em uf, faixa_etaria, categoria melhoram GROUP BY.

## 📌 Conclusão
As chaves primárias, estrangeiras e secundárias são o alicerce da integridade e performance no modelo relacional. Elas não apenas organizam os dados como também definem como eles podem ser confiavelmente combinados, analisados e transformados. Na ciência de dados, sua correta aplicação garante que os datasets usados em modelagem estejam consistentes, auditáveis e sem viés estrutural.

🎯 Dominar o uso de chaves é indispensável para garantir qualidade de dados, confiabilidade estatística e segurança operacional em pipelines analíticos e preditivos.
