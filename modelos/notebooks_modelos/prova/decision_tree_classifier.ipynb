```python
data = './classificacao_Q1.csv'

df = pd.read_csv(data)
```

```python
X = df.drop(['target'], axis=1)
y = df['target']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.33, random_state = 42)

X_train.shape, X_test.shape
```

```python
# import DecisionTreeClassifier

from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score, KFold
from sklearn.metrics import log_loss

# instanciar o modelo DecisionTreeClassifier com o critério índice de Gini

clf_gini = DecisionTreeClassifier(criterion='entropy', max_depth=3, random_state=0)

# KFold cross-validation com 7 folds
kf = KFold(n_splits=7, shuffle=True, random_state=0)

# Avaliando o modelo com cross-validation (usando acurácia)
cv_scores = cross_val_score(clf_gini, X_train, y_train, cv=kf, scoring='accuracy')

print("Acurácias em cada fold:", cv_scores)
print("Acurácia média (cross-validation): {:.4f}".format(cv_scores.mean()))

clf_gini.fit(X_train, y_train)

# Log loss no treino
proba_train = clf_gini.predict_proba(X_train)
logloss_train = log_loss(y_train, proba_train)
print('Log Loss (treino):', round(logloss_train, 4))

# Log loss no teste
proba_test = clf_gini.predict_proba(X_test)
logloss_test = log_loss(y_test, proba_test)
print('Log Loss (teste):', round(logloss_test, 4))
```
