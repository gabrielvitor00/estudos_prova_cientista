```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns 
%matplotlib inline

import sys
import os
import warnings
warnings.filterwarnings('ignore')
```

```python
data = './regressao_Q1.csv'
df = pd.read_csv(data)
```

```python
df.head()
```

```python
X = df.drop(['target'], axis=1)
y = df['target']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.30, random_state = 42)

X_train.shape, X_test.shape
```

```python
from sklearn.linear_model import ElasticNet
from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import mean_squared_error
import numpy as np

# Parâmetros do modelo
alpha = 1.0         # Fator de regularização
l1_ratio = 0.01     # Parâmetro de mistura (0 = Ridge, 1 = Lasso)

kf = TimeSeriesSplit(n_splits=5)

train_sse = []
test_sse = []

for train_idx, test_idx in kf.split(X):
    X_tr, X_te = X.iloc[train_idx], X.iloc[test_idx]
    y_tr, y_te = y.iloc[train_idx], y.iloc[test_idx]
    
    model = ElasticNet(alpha=alpha, l1_ratio=l1_ratio, random_state=42, max_iter=10000)
    model.fit(X_tr, y_tr)
    
    # Previsões
    y_tr_pred = model.predict(X_tr)
    y_te_pred = model.predict(X_te)
    
    # Soma dos quadrados dos erros (SSE)
    sse_tr = np.sum((y_tr - y_tr_pred) ** 2)
    sse_te = np.sum((y_te - y_te_pred) ** 2)
    train_sse.append(sse_tr)
    test_sse.append(sse_te)

# Calcular médias
mean_train_sse = np.mean(train_sse)
mean_test_sse = np.mean(test_sse)

print(f"Média da soma dos quadrados dos erros (SSE) no treino: {mean_train_sse:.2f}")
print(f"Média da soma dos quadrados dos erros (SSE) no teste:  {mean_test_sse:.2f}")

# Exibir SSE por fold
print("\nSSE por fold:")
for i, (tr, te) in enumerate(zip(train_sse, test_sse), 1):
    print(f"Fold {i}: Treino = {tr:.2f} | Teste = {te:.2f}")
```
