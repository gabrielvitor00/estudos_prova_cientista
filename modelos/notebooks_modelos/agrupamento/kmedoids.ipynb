# **Comparação entre K-Medoids e K-Means**

Tanto **K-Means** quanto **K-Medoids** são algoritmos de *clustering particional*, ou seja, dividem os dados em *K* grupos (clusters) buscando minimizar a distância entre os pontos e seus respectivos centros. No entanto, eles diferem em como definem esses centros, na robustez, função objetivo e desempenho computacional.

---

## ⚙️ 1. Centro do Cluster

- **K-Means:** utiliza a **média (mean)** dos pontos como centro. Esse centro pode não corresponder a um ponto real do dataset.
- **K-Medoids:** utiliza a **mediana (medoid)**, ou seja, o ponto mais central dentro do cluster, e **sempre pertence ao dataset**.

---

## 📏 2. Métrica de Distância

- **K-Means:** usa **distância Euclidiana (L2)**.
- **K-Medoids:** aceita **qualquer métrica de dissimilaridade** (Euclidiana, Manhattan, Hamming, etc.).

---

## 🎯 3. Função Objetivo

- **K-Means:** minimiza a **soma dos erros quadráticos** (Sum of Squared Errors – SSE).
- **K-Medoids:** minimiza a **soma das dissimilaridades absolutas** entre os pontos e seus medoides.

---

## 🛡️ 4. Robustez a Outliers

- **K-Means:** sensível a outliers e ruído.
- **K-Medoids:** mais **robusto**, pois usa pontos reais e não amplifica desvios extremos.

---

## ⏱️ 5. Complexidade Computacional

| Algoritmo    | Complexidade             | Observações                         |
|--------------|--------------------------|--------------------------------------|
| **K-Means**  | O(n × k × t × d)         | Mais rápido e adequado para grandes volumes. |
| **K-Medoids**| O(k × (n − k)²) ou O(knt) | Computacionalmente mais custoso.    |

- n = número de pontos  
- k = número de clusters  
- t = número de iterações  
- d = número de dimensões

---

## 🧪 6. Quando Usar

| Cenário                                  | Melhor Escolha     |
|------------------------------------------|---------------------|
| Dados com muitos **outliers**           | ✅ K-Medoids         |
| Dados **categóricos** ou mistos         | ✅ K-Medoids         |
| Grandes volumes de dados                | ✅ K-Means           |
| Alta velocidade é necessária            | ✅ K-Means           |
| Dados bem distribuídos (esféricos)      | ✅ K-Means           |
| Alta robustez e interpretabilidade      | ✅ K-Medoids         |

---

## 🧾 Resumo Comparativo

| Critério                    | **K-Means**               | **K-Medoids**                  |
|----------------------------|----------------------------|--------------------------------|
| Centro do cluster          | Média (não necessariamente real) | Medoide (sempre real)       |
| Métrica de distância       | Apenas Euclidiana          | Qualquer métrica               |
| Função objetivo            | Soma de quadrados          | Soma de dissimilaridades       |
| Robustez a outliers        | Fraca                      | Forte                          |
| Complexidade computacional | Baixa                      | Alta                           |
| Tipos de dados suportados  | Numéricos contínuos        | Numéricos e categóricos        |

---


> 💡 **Resumo:** Use K-Means para grandes volumes de dados numéricos e bem distribuídos. Prefira K-Medoids quando os dados contêm outliers, ruídos ou atributos categóricos.
