# **Normas L1 versus L2**

A regressão Ridge e a regressão Lasso são duas técnicas diferentes usadas para aumentar a robustez da regressão dos mínimos quadrados ordinários contra a multicolinearidade. Ambas tentam minimizar uma função de custo. Essa função é composta por dois termos: o primeiro é a soma dos resíduos quadráticos (RSS), vinda dos mínimos quadrados ordinários; o segundo é um termo de penalização adicional. Esse segundo termo é uma norma L2 na regressão Ridge e uma norma L1 na regressão Lasso.

---

## **Visão Geral**

Vamos analisar as equações. Na regressão dos mínimos quadrados ordinários, minimizamos a seguinte função de custo:

$$\text{Custo} = (y - X\beta)^T(y - X\beta)$$

Esse termo é a **RSS** (soma dos quadrados dos resíduos). Já na regressão Ridge, resolvemos:

$$\text{Custo} = (y - X\beta)^T(y - X\beta) + \lambda \beta^T\beta$$

O termo $\lambda \beta^T\beta$ é uma **norma L2**.

Na regressão Lasso, temos:

$$\text{Custo} = (y - X\beta)^T(y - X\beta) + \lambda |\beta|$$

O termo $\lambda |\beta|$ representa a **norma L1**.

Em um nível mais alto, a principal diferença entre os termos L1 e L2 é que o termo L2 é proporcional ao **quadrado** dos valores de $\beta$, enquanto o termo L1 é proporcional ao **valor absoluto** desses mesmos valores. Essa diferença fundamental é a base para o funcionamento distinto entre Lasso e Ridge. Essa diferença também aparece em outras áreas do aprendizado de máquina, por isso é essencial compreendê-la.

---

## **Definição de uma norma**

Vamos dar um passo atrás e responder: *o que é uma norma?*

Uma norma é uma função matemática aplicada a um vetor (como o vetor $\beta$). A norma mapeia vetores para valores no intervalo $[0, \infty)$. Em aprendizado de máquina, normas são úteis para expressar distâncias.

A definição geral de uma **p-norma** é:

$$||x||_p = \left(\sum_i |x_i|^p\right)^{1/p}$$

### Exemplos comuns:

- **Norma L2 (Euclidiana):**

$$||x||_2 = \sqrt{\sum_i x_i^2} = \sqrt{x_1^2 + x_2^2 + \dots + x_i^2}$$

- **Norma L1 (Taxicab ou Manhattan):**

$$||x||_1 = \sum_i |x_i| = |x_1| + |x_2| + \dots + |x_i|$$

A norma Taxicab é assim chamada porque simula mover-se pelas ruas em ângulos retos, como em Manhattan.

![Taxicab distance](https://upload.wikimedia.org/wikipedia/commons/0/08/Manhattan_distance.svg)

---

## **p-normas e Regularização**

A norma L1 (taxicab) é utilizada na regressão Lasso, enquanto a norma L2 (euclidiana) aparece na regressão Ridge. Podemos reescrever as funções de custo como:

- **Regressão Ridge:**

$$\text{Custo Ridge} = (y - X\beta)^T(y - X\beta) + ||\beta||_2^2$$

- **Regressão Lasso:**

$$\text{Custo Lasso} = (y - X\beta)^T(y - X\beta) + ||\beta||_1$$

Esse processo de adicionar uma norma à função de custo é chamado de **regularização**. Ele ajuda a resolver problemas como **colinearidade** e **complexidade do modelo**.

---

## **Comparações entre normas L1 e L2**

### ✅ Robustez: **L1 > L2**

A L1 é mais robusta a **outliers**, pois penaliza de forma linear. A L2 penaliza quadráticamente, aumentando exageradamente o impacto de valores extremos.

### ✅ Estabilidade: **L2 > L1**

L2 é mais estável a perturbações pequenas nos dados. Esse conceito é oposto à robustez.

### ✅ Número de soluções: **L2 = única, L1 = múltiplas**

A L2 gera uma **única solução** (distância euclidiana única). Já a L1 pode ter **múltiplas soluções equivalentes** (várias formas de andar em Manhattan).

---

## **Comparações de regularizadores L1 e L2**

### 💻 Dificuldade computacional: **L1 > L2**

- **L2** possui **solução analítica fechada**, pois envolve quadrados contínuos.
- **L1** envolve valor absoluto, o que a torna **não diferenciável** e requer métodos iterativos, como **descida coordenada**.

### 🧽 Esparsidade: **L1 > L2**

- A L1 induz **esparsidade**: muitos coeficientes vão a zero.
- A Lasso (L1) possui **seleção de variáveis embutida**.
- A Ridge (L2) apenas encolhe os coeficientes, mas não os zera.

---

## **Resumo**

| Propriedade            | L1 (Lasso)               | L2 (Ridge)              |
|------------------------|--------------------------|--------------------------|
| Penalização            | Valor absoluto           | Quadrado                |
| Solução fechada        | ❌ Não                   | ✅ Sim                  |
| Robustez a outliers    | ✅ Alta                  | ❌ Baixa                |
| Estabilidade           | ❌ Menor                 | ✅ Maior                |
| Esparsidade            | ✅ Induz                 | ❌ Não induz            |
| Seleção de variáveis   | ✅ Sim                   | ❌ Não                  |

> 🔍 **Conclusão:**  
A escolha entre L1 e L2 depende dos objetivos do modelo. L1 é ideal quando se deseja reduzir o número de variáveis (esparsidade), enquanto L2 é mais adequada quando se quer estabilidade e robustez contra colinearidade.

---

## Elastic Net: o melhor dos dois mundos?

A Elastic Net combina os termos de penalização L1 e L2:

$$\text{Custo} = (y - X\beta)^T(y - X\beta) + \lambda_1 ||\beta||_1 + \lambda_2 ||\beta||_2^2$$

Isso permite controle duplo: manter a esparsidade da L1 e a estabilidade da L2.
