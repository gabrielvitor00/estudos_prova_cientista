# 🏔️ Hierarchical Clustering (Agrupamento Hierárquico)

O **Hierarchical Clustering** é uma técnica de aprendizado não supervisionado utilizada para **agrupamento** (clustering) de dados, onde os grupos são formados com base na similaridade, criando uma hierarquia de clusters. O resultado é representado em uma estrutura de árvore chamada **dendrograma**, mostrando relações de proximidade entre os objetos.

---

# 🏔️ Objetivo

Identificar estruturas de agrupamento natural nos dados, sem pré-definir o número de clusters, organizando os objetos em uma hierarquia (do mais geral ao mais específico ou vice-versa).  
Permite análise exploratória, segmentação, detecção de padrões e insights sobre relações entre grupos.

---

## 🧮 Funcionamento e Matemática

- Existem dois métodos principais:
  - **Aglomerativo (bottom-up):** Cada ponto começa como seu próprio cluster. Em cada passo, os dois clusters mais próximos são fundidos, até restar apenas um.
  - **Divisivo (top-down):** Todos os pontos começam em um único cluster e são divididos recursivamente em clusters menores.

- **Medidas de Similaridade:**  
  A distância entre pontos ou clusters pode ser calculada de várias formas:
    $$
    d(x, y) = \sqrt{\sum_{i=1}^n (x_i - y_i)^2}
    $$
    (Distância Euclidiana – a mais comum)

- **Critérios de ligação (Linkage):**
  - **Single linkage:** Distância mínima entre pontos dos dois clusters.
  - **Complete linkage:** Distância máxima entre pontos dos dois clusters.
  - **Average linkage:** Média das distâncias entre todos os pares de pontos dos dois clusters.
  - **Ward linkage:** Minimiza a soma dos quadrados das diferenças dentro de todos os clusters (criterioso para compactar clusters).

---

## ✅ Vantagens do Hierarchical Clustering

- **Não requer número pré-definido de clusters** (pode ser definido visualmente depois).
- Revela a estrutura hierárquica dos dados (multi-níveis).
- Funciona bem com dados pequenos e interpretáveis.
- Flexível: pode ser aplicado a diferentes tipos de dados e métricas de distância.
- Permite fácil visualização via dendrograma.

---

## ⚠️ Desvantagens do Hierarchical Clustering

- **Alto custo computacional** (escalabilidade ruim para bases grandes).
- Sensível a outliers e ruído.
- Não permite "desfazer" fusões (decisões são finais).
- Resultados podem variar conforme a métrica de distância e o linkage escolhidos.
- Difícil de aplicar diretamente em alta dimensionalidade.

---

# ⚖️ Trade-off entre Viés e Variância

- **Linkage agressivo (single):** Clusters podem ser longos e “esticados” (baixa variância, alto viés).
- **Linkage conservador (complete):** Clusters tendem a ser compactos, mais sujeitos a variância.
- O ajuste do linkage permite controlar granularidade e sensibilidade dos clusters.

---

## 🧠 Conceitos-Chave

- **Dendrograma:** Gráfico em árvore que mostra o processo de fusão/divisão dos clusters.
- **Métrica de distância:** Mede proximidade entre pontos ou clusters (euclidiana, Manhattan, etc).
- **Linkage:** Critério para definir distância entre clusters (single, complete, average, ward).
- **Corte do dendrograma:** Definir o número final de clusters com base no ponto de corte.

---

## 🛠️ Como Lidar com o Trade-off

- Escolha o linkage adequado ao padrão dos seus dados (ward para clusters compactos, average para suavidade).
- Considere reduzir dimensionalidade antes (PCA) para bases com muitas features.
- Analise e trate outliers antes do agrupamento.
- Ajuste a altura do corte no dendrograma para obter o número de clusters desejado.

---

## ✅ Data Prep Essencial para Hierarchical Clustering

1. **Normalizar/padronizar os dados** (essencial para distâncias serem comparáveis).
2. **Remover ou tratar outliers** para evitar distorções nos clusters.
3. **Reduzir dimensionalidade** se houver muitas variáveis (PCA, seleção de features).
4. **Escolher métrica de distância e linkage adequados** ao objetivo da análise.

---

## ⚙️ Principais Parâmetros do Hierarchical Clustering

| Parâmetro        | O que faz                                                    |
|------------------|-------------------------------------------------------------|
| `n_clusters`     | Número de clusters desejados (definido após o dendrograma)  |
| `affinity`       | Métrica de distância usada (euclidiana, manhattan, etc.)    |
| `linkage`        | Critério de ligação (single, complete, average, ward)       |
| `distance_threshold` | Permite cortar o dendrograma em um ponto de similaridade |

---

## 📏 Principais Métricas de Avaliação

| Métrica        | Interpretação                                                |
|----------------|-------------------------------------------------------------|
| **Silhouette Score** | Mede coesão/separação dos clusters (quanto maior, melhor) |
| **Davies-Bouldin**  | Índice de similaridade intra/inter cluster (quanto menor, melhor) |
| **Dendrograma**     | Avaliação visual, corte no ponto ideal                  |

---

## 📊 Visualização Intuitiva

- O dendrograma mostra todas as etapas do agrupamento, permitindo identificar o número ótimo de clusters de forma visual.
- Pode ser complementado por projeções 2D (PCA, t-SNE) para entender a separação dos grupos.

---

## 🧪 Dica prática para ciência de dados

- Use Hierarchical Clustering para **análise exploratória** e quando não souber previamente o número de clusters.
- Sempre normalize e trate outliers antes de aplicar.
- Utilize o dendrograma para comunicar insights a públicos não técnicos.
- Para grandes volumes de dados, considere métodos mais escaláveis (ex: K-means).

---

*Obs: Hierarchical Clustering é ferramenta poderosa para explorar relações e estruturas ocultas, servindo como ponto de partida para segmentação e análise de padrões em bases menores ou médias.*
