# 🧭 Support Vector Machine (SVM)

**Support Vector Machine (SVM)** é um algoritmo supervisionado clássico de Machine Learning, projetado para **classificação** e **regressão**. Seu foco principal está em encontrar a **fronteira ótima** de separação entre classes, maximizando a margem (distância) entre os pontos mais próximos de cada classe e a linha ou hiperplano separador.

---

# 🧭 Objetivo

Construir um **hiperplano de separação** que maximize a margem entre as classes, tornando o modelo robusto e com maior capacidade de generalização, mesmo em espaços de alta dimensionalidade.  
Para problemas não linearmente separáveis, utiliza **funções kernel** para projetar os dados em um espaço de maior dimensão onde a separação seja possível.

---

## 🧮 Funcionamento e Matemática

### Margem Máxima

- O SVM encontra o **hiperplano** que separa as classes e maximiza a margem:
    $$
    w \cdot x + b = 0
    $$

- **Vetores de suporte:** São os pontos de dados mais próximos do hiperplano. Eles definem a margem e são críticos para a solução.

- **Função de otimização (classificação linear):**
    $$
    \text{Minimizar} \quad \frac{1}{2} \|w\|^2
    $$
    $$
    \text{Sujeito a:} \quad y_i(w \cdot x_i + b) \geq 1
    $$
    Onde \( w \) é o vetor de pesos, \( b \) é o intercepto, \( y_i \) é o rótulo da amostra \( i \), e \( x_i \) é o vetor de features da amostra \( i \).

### Casos Não-Lineares (Kernel Trick)

- Usa funções kernel (ex: RBF, polinomial, sigmoid) para transformar os dados em um espaço de maior dimensão, permitindo a separação linear nesse novo espaço.

### SVM para Regressão (SVR)

- Minimiza o erro dentro de uma faixa de tolerância (epsilon), penalizando erros fora dela.

---

## ✅ Vantagens do SVM

- **Eficiente em alta dimensionalidade:** Bom para bases com muitas features.
- **Capaz de modelar relações não-lineares** via kernels.
- **Robusto a overfitting** (com margem larga e regularização adequada).
- **Utiliza apenas vetores de suporte** para determinar a fronteira, tornando o modelo eficiente.
- Flexível: pode ser usado para classificação, regressão e até detecção de anomalias.

---

## ⚠️ Desvantagens do SVM

- **Treinamento lento em bases grandes:** Não escala bem para muitos exemplos.
- **Sensível à escolha do kernel e dos hiperparâmetros** (C, gamma, kernel).
- Pouco eficiente em bases muito desbalanceadas.
- Difícil interpretar a importância das features.
- Não fornece probabilidades diretamente (precisa de métodos auxiliares, como Platt scaling).

---

# ⚖️ Trade-off entre Viés e Variância

- **Alto viés:** SVM linear, kernel inadequado ou C baixo (margem larga, pouca flexibilidade).
- **Alta variância:** Kernel complexo, C alto (margem estreita, modelo sensível ao ruído).
- **Controle:** Parâmetro **C** regula o equilíbrio entre margem larga (viés alto) e erros no treino (variância alta).  
  Parâmetros do kernel (ex: **gamma** para RBF) também afetam o trade-off.

---

## 🧠 Conceitos-Chave

- **Margem máxima:** Distância entre o hiperplano e os vetores de suporte.
- **Kernel trick:** Mapeia dados para espaços de maior dimensão para separação não-linear.
- **Vetores de suporte:** Pontos de dados que definem a margem e são críticos para a solução.
- **Parâmetro C:** Controla o grau de penalização dos erros (regularização).
- **Gamma:** Controla a influência de pontos individuais em kernels RBF e polinomial.

---

## 🛠️ Como Lidar com o Trade-off

- Ajustar **C** para controlar regularização.
- Escolher o **kernel** adequado ao problema (linear, polinomial, RBF, sigmoid).
- Ajustar **gamma** em kernels RBF/polinomial para definir influência local/global.
- Normalizar/padronizar features para melhorar performance dos kernels.
- Usar validação cruzada para buscar os melhores hiperparâmetros.

---

## ✅ Data Prep Essencial para SVM

1. **Normalizar ou padronizar variáveis** (SVM é sensível à escala).
2. **Tratar valores ausentes** (necessário, SVM não aceita missing).
3. **Remover variáveis irrelevantes** para acelerar o treino.
4. **Reduzir dimensionalidade** (PCA, seleção de features) se houver muitas features redundantes.
5. **Balancear as classes** (se necessário, para evitar viés no ajuste).

---

## ⚙️ Principais Parâmetros do SVM

| Parâmetro   | O que faz                                                         |
|-------------|-------------------------------------------------------------------|
| `C`         | Penalidade de erro (quanto maior, menor margem, mais variância)   |
| `kernel`    | Tipo de kernel (linear, poly, rbf, sigmoid)                       |
| `degree`    | Grau do polinômio (para kernel poly)                              |
| `gamma`     | Influência de pontos próximos/distantes (kernels rbf, poly)       |
| `coef0`     | Constante em kernels poly e sigmoid                               |
| `class_weight` | Ajuste para classes desbalanceadas                          |

---

## 📏 Principais Métricas de Avaliação

| Métrica        | Interpretação                                        |
|----------------|-----------------------------------------------------|
| **Acurácia**   | Proporção de acertos                                |
| **Precisão**   | Proporção de positivos previstos corretamente       |
| **Recall**     | Proporção de positivos reais identificados          |
| **F1-Score**   | Média harmônica entre precisão e recall             |
| **ROC AUC**    | Capacidade de separação entre classes               |
| **Erro epsilon** | (SVR) Erro dentro da faixa de tolerância          |

---

## 📊 Visualização Intuitiva

- O SVM linear pode ser visualizado como uma linha (ou hiperplano) que separa as classes, maximizando a distância dos pontos mais próximos.
- Com kernels, a fronteira pode se tornar altamente não-linear e adaptativa ao formato dos dados.

---

## 🧪 Dica prática para ciência de dados

- Use SVM para datasets com **muitas features** e poucos exemplos.
- Sempre normalizar os dados.
- Escolha e ajuste do kernel são cruciais para bom desempenho.
- Em bases muito grandes, avalie alternativas mais escaláveis (árvores, boosting, redes).
- Se precisar de probabilidades, use `probability=True` ou técnicas pós-modelo.

---

*Obs: SVM foi referência em competições tabulares até a popularização dos métodos de boosting, permanecendo um modelo sólido, especialmente em cenários de alta dimensionalidade e dados complexos.*
