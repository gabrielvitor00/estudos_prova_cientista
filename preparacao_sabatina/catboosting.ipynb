# 🐱 CatBoost (Categorical Boosting)

**CatBoost** é um framework de Gradient Boosting desenvolvido pela Yandex, projetado para trabalhar de forma eficiente com **variáveis categóricas** e para entregar alta performance em tarefas de classificação e regressão em dados tabulares, mesmo em conjuntos de dados complexos, desbalanceados e com valores ausentes.

---

# 🐱 Objetivo

Fornecer um modelo de boosting robusto, preciso e rápido, com **suporte nativo e eficiente a variáveis categóricas**, reduzindo a necessidade de pré-processamento e evitando overfitting causado por codificação inadequada das categorias.

---

## 🧮 Funcionamento e Matemática

### Características-Chave do CatBoost

- **Codificação Ordenada de Categóricas:** CatBoost transforma variáveis categóricas em números usando uma estratégia de codificação ordenada (target statistics), minimizando o risco de overfitting por vazamento de alvo (“target leak”).
- **Node-wise Growth (Best-first split):** Árvores crescem expandindo o nó com maior ganho de informação, assimétricas e adaptativas.
- **Boosting simétrico:** Constrói árvores binárias simétricas, o que acelera a inferência e melhora a generalização.
- **Manejo avançado de valores ausentes:** Suporte nativo, sem necessidade de imputação.
- **Randomização adicional:** Usa esquemas específicos de randomização para evitar overfitting e aumentar robustez.

### Processo de Treinamento

- Constrói árvores sequencialmente, cada uma focando nos resíduos do ensemble anterior, como todo método de Gradient Boosting.
- Categóricas são transformadas durante o treinamento sem necessidade de one-hot ou label encoding manual.
- Regularização embutida para evitar overfitting.

---

## ✅ Vantagens do CatBoost

- **Suporte nativo, rápido e eficiente a variáveis categóricas.**
- Reduz risco de overfitting com codificação ordenada exclusiva.
- **Fácil de usar:** menos pré-processamento e ajuste fino necessários.
- **Performance competitiva** com XGBoost e LightGBM em muitos cenários tabulares.
- Resistente a desbalanceamento de classes e valores ausentes.
- Permite explainability (`Pool`, SHAP, visualização de importâncias).

---

## ⚠️ Desvantagens do CatBoost

- **Consome mais memória** para conjuntos muito grandes que LightGBM.
- **Treinamento mais lento** que LightGBM para alguns cenários, principalmente em CPU.
- Algumas opções de hiperparâmetros avançados menos flexíveis que XGBoost/LightGBM.
- A comunidade, documentação e exemplos ainda são menores que dos concorrentes.

---

# ⚖️ Trade-off entre Viés e Variância

- **Viés:** Baixo, devido à capacidade das árvores adaptativas e codificação ordenada.
- **Variância:** Controlada por regularização interna, profundidade de árvore (`depth`) e taxa de aprendizado (`learning_rate`).
- **Ajuste:** Parâmetros como `iterations`, `depth`, `learning_rate` e `l2_leaf_reg` regulam o equilíbrio.

---

## 🧠 Conceitos-Chave

- **Ordered Target Statistics:** Codificação que previne data leak ao calcular estatísticas apenas em observações anteriores na ordem do dataset.
- **Symmetric Tree:** Todas as folhas no mesmo nível, acelera a inferência.
- **Pool:** Estrutura de dados interna otimizada para CatBoost.
- **Handling de missing values:** Feito nativamente sem necessidade de imputação manual.
- **Suporte a múltiplos tipos de tarefa:** Classificação, regressão, ranking, multi-rótulo, etc.

---

## 🛠️ Como Lidar com o Trade-off

- Ajuste a profundidade das árvores (`depth`) para evitar overfitting em dados ruidosos.
- Reduza `learning_rate` e aumente `iterations` para aprendizagem mais estável.
- Utilize parâmetros de regularização como `l2_leaf_reg`.
- Use early stopping para prevenir overfitting.
- Especifique corretamente variáveis categóricas (parâmetro `cat_features`).

---

## ✅ Data Prep Essencial para CatBoost

1. **Informar variáveis categóricas** explicitamente (não faça one-hot!).
2. **Tratar valores ausentes** é opcional, pois o CatBoost lida internamente.
3. **Remover variáveis irrelevantes** pode acelerar e melhorar o modelo.
4. **Separar dados em treino/validação/teste.**
5. **Balancear classes** se necessário (`class_weights`).

---

## ⚙️ Principais Parâmetros do CatBoost

| Parâmetro             | O que faz                                                      |
|-----------------------|----------------------------------------------------------------|
| `iterations`          | Número de árvores (estimadores)                                |
| `learning_rate`       | Taxa de aprendizado (shrinkage)                                |
| `depth`               | Profundidade máxima das árvores                                |
| `l2_leaf_reg`         | Regularização L2 das folhas                                    |
| `cat_features`        | Lista de colunas categóricas                                   |
| `loss_function`       | Função de perda (logloss, MAE, RMSE, etc.)                     |
| `random_seed`         | Semente de aleatoriedade                                       |
| `early_stopping_rounds` | Early stopping baseado na validação                        |

---

## 📏 Principais Métricas de Avaliação

| Métrica        | Interpretação                                        |
|----------------|-----------------------------------------------------|
| **Acurácia**   | Proporção de acertos (classificação)                |
| **Precisão**   | Proporção de positivos previstos corretamente       |
| **Recall**     | Proporção de positivos reais identificados          |
| **F1-Score**   | Média harmônica entre precisão e recall             |
| **ROC AUC**    | Capacidade do modelo em separar classes             |
| **MAE/MSE/RMSE** | Métricas de erro para regressão                   |
| **Log Loss**   | Erro das probabilidades previstas (classificação)   |

---

## 📊 Visualização Intuitiva

- Permite visualizar importâncias das variáveis, efeitos de variáveis categóricas e interpretar predições com métodos SHAP integrados.
- Estrutura de árvores simétricas pode ser plotada para debugging e explicação.

---

## 🧪 Dica prática para ciência de dados

- Use CatBoost quando seu dataset tiver **muitas variáveis categóricas** e missing values.
- Economize tempo em pré-processamento: informe as categorias, e o CatBoost faz o resto.
- Combine com validação cruzada e ajuste de `learning_rate`, `depth`, e regularização para evitar overfitting.
- Ideal para competição de dados tabulares quando interpretabilidade e facilidade de uso são diferenciais.

---

*Obs: CatBoost é referência em boosting para dados ricos em variáveis categóricas, tornando-se padrão de fato em muitos problemas do mundo real.*
