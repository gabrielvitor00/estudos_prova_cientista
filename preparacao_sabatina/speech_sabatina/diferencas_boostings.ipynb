**Diferenças Entre Boostings**

Os principais algoritmos de boosting – AdaBoost, Gradient Boosting, XGBoost, LightGBM e CatBoost – compartilham o objetivo de combinar vários modelos simples para formar um modelo mais robusto, porém apresentam diferenças técnicas e funcionais relevantes.

AdaBoost ajusta pesos das amostras, focando nos exemplos mal classificados, e é especialmente sensível a outliers. O Gradient Boosting tradicional constrói modelos sequenciais corrigindo resíduos dos anteriores, sem grandes otimizações para escala ou velocidade.

XGBoost inovou ao introduzir regularização explícita, tratamento de missing values, paralelização e construção de árvores node-wise, tornando-se referência em performance para dados tabulares. LightGBM trouxe como diferencial a velocidade, usando histogramas para discretizar features e crescimento level-wise, além do suporte eficiente a variáveis categóricas.

CatBoost destaca-se por tratar variáveis categóricas nativamente, usando codificação ordenada para evitar vazamento de alvo e overfitting, e por construir árvores simétricas que aceleram a inferência.

A escolha do algoritmo depende do formato dos dados, presença de variáveis categóricas, tamanho do conjunto e da necessidade de balancear velocidade, interpretabilidade e performance.
