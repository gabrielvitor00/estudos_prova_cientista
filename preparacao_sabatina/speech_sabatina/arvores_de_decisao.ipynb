**Árvores de Decisão**

Árvores de decisão são algoritmos supervisionados usados para classificação e regressão. O modelo constrói uma estrutura em formato de árvore onde, a cada nó, faz uma pergunta sobre os dados e separa os exemplos com base em uma regra de decisão baseada em impureza. Cada divisão busca maximizar a pureza dos grupos resultantes, utilizando critérios como Gini, Entropia ou MSE, dependendo da tarefa. O processo segue de forma recursiva até alcançar folhas puras ou até atender critérios de parada definidos.

Entre as vantagens das árvores de decisão estão a interpretabilidade – é fácil visualizar o caminho de decisão para qualquer predição –, a capacidade de lidar com dados numéricos e categóricos e a resistência a outliers. No entanto, árvores profundas podem sofrer overfitting, capturando ruídos do treino, enquanto árvores rasas podem não capturar padrões importantes, gerando underfitting. Técnicas como poda e limitação da profundidade ajudam a regular esse equilíbrio entre viés e variância.

A escolha do critério de divisão e dos hiperparâmetros como profundidade máxima e número mínimo de exemplos nas folhas impacta diretamente o desempenho e a robustez do modelo. Árvores de decisão são amplamente usadas como base para métodos de ensemble como Random Forest e Gradient Boosting.

Em resumo, árvores de decisão oferecem um modelo flexível, interpretável e útil para diferentes contextos, mas precisam de regularização para evitar overfitting em bases ruidosas.
