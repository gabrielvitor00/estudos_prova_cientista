**K-Modes**

K-Modes é um algoritmo de agrupamento não supervisionado projetado para trabalhar com dados categóricos. Ele é uma extensão do K-Means, mas ao invés de calcular médias, cada cluster é representado pelo modo, ou seja, pela categoria mais frequente em cada atributo. O algoritmo atribui cada ponto ao cluster cujo modo é mais semelhante, recalculando os modos a cada iteração até estabilizar as atribuições.

A principal vantagem do K-Modes é lidar naturalmente com variáveis categóricas, sem exigir transformações numéricas ou métricas de distância baseadas em valores contínuos. É eficiente para segmentação em bases pequenas e médias, mas pode ser sensível à inicialização dos modos e à alta cardinalidade de atributos. O número de clusters K precisa ser definido antecipadamente.

K-Modes é uma escolha indicada para análises de segmentação quando todas as variáveis do conjunto de dados são qualitativas.
