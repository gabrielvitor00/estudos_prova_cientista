**DBSCAN**

DBSCAN é um algoritmo de agrupamento não supervisionado baseado em densidade, projetado para identificar clusters de qualquer formato e detectar automaticamente outliers. O funcionamento é guiado por dois parâmetros principais: o raio de vizinhança ($\varepsilon$) e o número mínimo de pontos necessários para formar um cluster (min_samples). Um ponto é considerado núcleo de cluster se houver pelo menos min_samples dentro do raio $\varepsilon$; pontos periféricos que estejam próximos de núcleos também são incluídos nos clusters, enquanto pontos isolados são classificados como ruído.

Uma das maiores vantagens do DBSCAN é não exigir a definição prévia do número de clusters e conseguir identificar agrupamentos com formatos complexos, além de detectar outliers de forma natural. Ele é robusto para clusters de diferentes formatos e densidades, desde que os parâmetros estejam bem ajustados. No entanto, DBSCAN é sensível à escolha de $\varepsilon$ e min_samples, pode ter desempenho limitado em dados de alta dimensionalidade e não é indicado para dados puramente categóricos.

DBSCAN é ideal para análise exploratória, detecção de anomalias e quando não há clareza sobre a quantidade ou formato dos clusters presentes nos dados.
