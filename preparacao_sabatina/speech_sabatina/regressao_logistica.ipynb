**Regressão Logística**

A regressão logística é um modelo supervisionado utilizado para prever a probabilidade de ocorrência de um evento binário, como “sim” ou “não”. Ao invés de buscar uma linha reta como na regressão linear, o modelo aprende uma função em formato de “S”, chamada sigmoide, que transforma qualquer valor real em uma probabilidade entre zero e um.

O funcionamento é simples: cada variável de entrada recebe um peso, que indica o quanto ela influencia a chance do evento ocorrer. O modelo soma os efeitos dessas variáveis e passa o resultado por uma função que comprime esse valor para o intervalo de probabilidade. Se o resultado for acima de um certo limite, geralmente 0,5, o modelo prevê uma classe; se for abaixo, prevê a outra.

Entre as vantagens da regressão logística estão a facilidade de interpretação dos pesos das variáveis, a capacidade de trabalhar com probabilidades diretamente e a estabilidade em problemas com relações lineares entre variáveis e o logit da resposta. No entanto, ela pode ter dificuldade em capturar padrões não lineares e é sensível a valores extremos. O modelo é muito usado como baseline em classificação binária e em contextos onde a transparência das decisões é fundamental.

Resumindo, a regressão logística é uma escolha sólida para previsão de eventos binários, oferece probabilidades como saída e permite interpretar facilmente o efeito de cada variável no resultado.
