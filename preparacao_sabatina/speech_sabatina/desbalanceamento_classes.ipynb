**Desbalanceamento de Classes**

O desbalanceamento de classes ocorre quando uma ou mais classes estão sub-representadas em relação às demais, o que é comum em aplicações como detecção de fraude ou doenças raras. Esse cenário leva modelos a priorizarem a classe majoritária, resultando em baixo recall ou precisão para a classe minoritária e métricas como acurácia tornam-se enganosas.

Para lidar com esse problema, existem abordagens diretas nos dados e nos algoritmos. Entre as técnicas de amostragem estão o oversampling, como o SMOTE, que cria exemplos sintéticos da classe minoritária, e o undersampling, que reduz o número de exemplos da classe majoritária. Também é possível combinar ambos para balancear o conjunto de treino.

Outra estratégia é ajustar o algoritmo, usando hiperparâmetros como class_weight em modelos lineares, árvores, ensemble ou boosting. Isso faz com que o erro na minoria seja mais penalizado durante o treino, forçando o modelo a dar mais atenção a essa classe. Também pode-se adotar funções de custo customizadas ou técnicas como focal loss em redes neurais.

Além disso, é fundamental mudar as métricas de avaliação: priorize F1-score, recall, precisão, ROC-AUC e PR-AUC, em vez de acurácia. O ajuste do limiar de decisão e o uso de validação cruzada estratificada completam o ciclo de boas práticas.

Resumindo, lidar com desbalanceamento exige abordagem múltipla: balancear os dados, ajustar o modelo e adotar métricas mais informativas para garantir bom desempenho em todas as classes.
