**K-Means**

K-Means é um algoritmo de agrupamento não supervisionado utilizado para segmentar dados numéricos em K clusters distintos. O processo começa com a seleção aleatória de K centróides, que representam cada cluster. Em seguida, cada ponto é atribuído ao cluster cujo centróide está mais próximo, com base na distância euclidiana. Após a atribuição, os centróides são recalculados como a média dos pontos de cada grupo. Esse ciclo de atribuição e atualização continua até que os centróides não mudem mais ou até atingir o limite de iterações.

O principal benefício do K-Means é a simplicidade e eficiência para bases médias e grandes. Entretanto, é sensível à inicialização dos centróides, outliers e clusters de formatos não esféricos. O número de clusters K deve ser definido previamente, normalmente com métodos como o elbow ou silhouette score. O algoritmo também exige normalização das variáveis para garantir que todas contribuam igualmente.

K-Means é bastante usado como método inicial de segmentação em análises exploratórias, sendo eficiente quando os clusters são bem definidos e os dados são contínuos.
