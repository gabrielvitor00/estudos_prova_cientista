**Random Forest**

Random Forest é um algoritmo de ensemble supervisionado baseado no princípio do bagging (bootstrap aggregating). Ele constrói múltiplas árvores de decisão, onde cada árvore é treinada com uma amostra aleatória do conjunto de dados (com reposição, via bootstrap). Além disso, a cada split em cada árvore, apenas um subconjunto aleatório das variáveis é considerado para escolha da divisão. Essa combinação de amostragem de dados e de features gera diversidade entre as árvores, reduzindo correlação e melhorando a generalização do modelo.

A predição final é feita por votação majoritária nas árvores (para classificação) ou pela média (para regressão). O uso do bagging permite que a Random Forest reduza a variância dos modelos individuais, tornando-se muito menos propensa a overfitting do que uma árvore única, ainda que mantenha baixo viés. É especialmente robusta a outliers, ruídos e a dados com muitas variáveis irrelevantes, já que árvores diferentes capturam diferentes padrões.

Por outro lado, Random Forest perde em interpretabilidade comparado a uma árvore única e pode demandar mais memória e processamento, especialmente em bases muito grandes ou com muitas árvores. Entre os hiperparâmetros mais importantes estão o número de árvores, o número de variáveis consideradas por split, a profundidade máxima e o tamanho mínimo das folhas.

O trade-off entre viés e variância pode ser ajustado aumentando ou reduzindo o número de árvores e a profundidade máxima. O método é amplamente usado como baseline robusto e para problemas onde se deseja alta performance sem a necessidade de tunning extremo de parâmetros.
