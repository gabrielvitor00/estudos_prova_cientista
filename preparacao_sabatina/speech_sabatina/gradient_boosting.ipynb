**Gradient Boosting**

Gradient Boosting é um método de ensemble supervisionado que constrói modelos sequencialmente, onde cada novo modelo tenta corrigir os erros dos anteriores. O mecanismo central é ajustar, a cada passo, um novo modelo (geralmente uma árvore de decisão) aos resíduos dos modelos anteriores, usando gradientes da função de perda escolhida. Ao final, as previsões são combinadas somando os resultados ponderados de cada árvore.

O Gradient Boosting é altamente flexível, permite usar diferentes funções de perda (como erro quadrático, log-loss ou customizadas) e oferece bom desempenho tanto em classificação quanto em regressão. Os principais hiperparâmetros são o número de árvores, learning rate, profundidade das árvores e subsample, que controlam o equilíbrio entre viés e variância.

O método é sensível ao overfitting se houver árvores muito profundas ou learning rate alto, mas técnicas como shrinkage, early stopping e regularização ajudam a mitigar esse risco. Diferentemente de bagging, o boosting constrói modelos dependentes, cada um focando onde o anterior falhou.

Gradient Boosting é base para algoritmos avançados como XGBoost, LightGBM e CatBoost, sendo um dos métodos mais usados para dados tabulares estruturados.
