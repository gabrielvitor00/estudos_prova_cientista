**K-Medoids**

K-Medoids é um algoritmo de agrupamento não supervisionado semelhante ao K-Means, porém com maior robustez a outliers. Em vez de usar a média para representar o centro do cluster, o K-Medoids utiliza um ponto real do conjunto de dados, chamado medoid, como referência central de cada grupo. O processo consiste em escolher K medoids iniciais, atribuir cada ponto ao medoid mais próximo e iterativamente atualizar os medoids para minimizar a soma das distâncias dentro dos clusters.

Essa abordagem permite trabalhar com diferentes métricas de distância e reduz o impacto de pontos extremos, já que os medoids não são influenciados por valores discrepantes como a média seria. K-Medoids é mais indicado para bases pequenas ou médias devido ao maior custo computacional em relação ao K-Means, especialmente em grandes volumes de dados.

K-Medoids é útil quando se deseja segmentação mais robusta e interpretável, especialmente em cenários com dados ruidosos ou onde é fundamental que o centro do cluster seja sempre um ponto existente na base.
