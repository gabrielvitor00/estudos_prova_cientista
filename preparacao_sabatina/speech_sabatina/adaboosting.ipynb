**AdaBoosting**

AdaBoost, ou Adaptive Boosting, é um método de ensemble supervisionado que combina vários classificadores fracos para formar um classificador forte. O funcionamento central do AdaBoost se baseia em treinar uma sequência de modelos simples – geralmente stumps, que são árvores com apenas um split – onde cada novo modelo foca nos exemplos que os anteriores erraram. O algoritmo ajusta pesos das amostras: exemplos mal classificados recebem mais peso, tornando-se mais relevantes para os próximos modelos. O resultado final é uma votação ponderada desses modelos fracos, onde aqueles com maior acurácia têm maior influência na decisão.

As vantagens do AdaBoost incluem sua simplicidade, sua capacidade de melhorar rapidamente o desempenho de modelos básicos, e o fato de ser pouco propenso a overfitting quando se usa classificadores realmente fracos. Por outro lado, o ponto sensível do AdaBoost está nos outliers e ruídos: como o algoritmo foca obsessivamente nos erros, pode acabar superajustando o ruído, principalmente se houver muitos dados problemáticos.

AdaBoost pode ser adaptado para tarefas de regressão, mas seu uso mais comum é em classificação binária. Entre os parâmetros importantes estão o número de estimadores e o learning rate, que controla a influência de cada modelo individual.

Em resumo, o AdaBoost é um método eficiente para turbinar modelos simples, principalmente quando as classes são bem separáveis e não há tanto ruído nos dados.
