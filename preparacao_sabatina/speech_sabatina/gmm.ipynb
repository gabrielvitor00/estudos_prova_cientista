**GMM (Gaussian Mixture Models)**

GMM, ou Gaussian Mixture Models, é um método probabilístico para clustering e modelagem de densidade. O modelo representa os dados como uma soma ponderada de distribuições gaussianas multivariadas, cada uma correspondendo a um cluster. Ao contrário de métodos como K-Means, o GMM permite que cada ponto pertença parcialmente a vários clusters, atribuindo uma probabilidade de associação a cada componente.

O treinamento é feito pelo algoritmo Expectation-Maximization, que alterna entre calcular as probabilidades de cada ponto pertencer a cada cluster (E-step) e atualizar os parâmetros das gaussianas (M-step) para maximizar a verossimilhança dos dados. Os principais parâmetros são o número de componentes, o tipo de matriz de covariância e critérios de convergência.

GMM é flexível para clusters de formas elípticas e sobrepostas, permite modelar incerteza e detectar anomalias com base nas probabilidades, além de estimar densidades em problemas de modelagem estatística. Entre as limitações estão a necessidade de definir o número de componentes, a sensibilidade à inicialização, baixa robustez a outliers e menor performance em alta dimensionalidade sem regularização.

Em resumo, GMM é indicado quando se espera clusters não esféricos ou sobrepostos, sendo uma abordagem mais flexível e estatística do que o K-Means.
