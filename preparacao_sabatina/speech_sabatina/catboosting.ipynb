**CatBoosting**

CatBoost é um algoritmo de gradient boosting voltado para classificação, regressão e ranking, com diferencial no tratamento eficiente de variáveis categóricas. O CatBoost emprega técnicas próprias de codificação ordenada das variáveis categóricas, evitando vazamento de informação e reduzindo overfitting associado ao tratamento convencional dessas variáveis.

O algoritmo constrói árvores de decisão simétricas, em que todas as folhas do mesmo nível são expandidas simultaneamente. Isso traz ganhos em robustez e velocidade na inferência. O processo de boosting ocorre de maneira tradicional: as árvores subsequentes corrigem os erros residuais dos modelos anteriores, e cada árvore recebe um peso ajustado pelo learning rate.

Entre as vantagens do CatBoost estão o suporte nativo e eficiente a variáveis categóricas, robustez frente a dados desbalanceados e facilidade de uso – exige menos pré-processamento. Além disso, é competitivo em performance com outros métodos de boosting como XGBoost e LightGBM, especialmente em dados tabulares com muitos atributos qualitativos.

Como ponto de atenção, CatBoost pode consumir mais memória em conjuntos de dados muito grandes e ser mais lento que o LightGBM em alguns cenários, especialmente em CPU. Os principais parâmetros incluem o número de árvores, profundidade, learning rate e regularização das folhas.

Resumindo, CatBoost é recomendado sempre que o dataset contém muitas variáveis categóricas e quando se deseja um modelo robusto, de fácil uso e alta performance sem necessidade de pré-processamento complexo.
