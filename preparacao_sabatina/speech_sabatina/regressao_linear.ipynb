**Regressão Linear**

A regressão linear é um modelo supervisionado usado para prever valores numéricos a partir de uma ou mais variáveis de entrada. O modelo busca encontrar a linha reta ou plano que melhor se ajusta aos dados, de forma que a soma dos quadrados das diferenças entre os valores reais e os previstos seja a menor possível.

O funcionamento básico é o seguinte: para cada variável independente, o modelo calcula um peso, chamado coeficiente, que indica o quanto aquela variável impacta na previsão. Existe também um termo fixo, chamado intercepto, que representa o valor previsto quando todas as variáveis de entrada são zero. O objetivo é encontrar os valores desses coeficientes que fazem com que as previsões fiquem o mais próximo possível dos valores observados.

A principal vantagem da regressão linear está na sua simplicidade e transparência – é fácil interpretar o quanto cada variável influencia o resultado. Por outro lado, o modelo só captura relações lineares e pode ser bastante sensível a outliers ou a presença de variáveis altamente correlacionadas. Quando a relação entre as variáveis não é linear, o modelo pode apresentar alto viés, ou seja, errar sistematicamente. Já em bases pequenas ou muito ajustadas, pode ocorrer o overfitting, mas isso é menos comum em modelos lineares.

Em síntese, a regressão linear serve como referência interpretável para previsão de variáveis numéricas, sendo útil tanto em análise exploratória quanto como baseline em projetos de modelagem.
