**SVM (Support Vector Machine)**

SVM, ou Máquina de Vetores de Suporte, é um modelo supervisionado usado tanto para classificação quanto para regressão. O objetivo do SVM é encontrar a melhor fronteira possível que separa as classes, maximizando a margem entre essa fronteira e os pontos mais próximos de cada classe, conhecidos como vetores de suporte.

Quando os dados não podem ser separados linearmente, o SVM utiliza funções chamadas kernels para transformar os dados em um espaço de maior dimensão, tornando possível encontrar uma fronteira mais adequada.

Um aspecto central do SVM é o controle do equilíbrio entre viés e variância, que é feito principalmente por dois parâmetros: o valor de C e o tipo de kernel. O parâmetro C determina o quanto o modelo aceita erros de classificação em troca de uma margem mais ampla. Um valor de C baixo prioriza uma margem maior e resulta em maior viés e menor variância, tornando o modelo mais simples e menos sensível ao ruído. Já um C alto reduz o viés, buscando separar perfeitamente as classes no treino, mas pode aumentar a variância e levar ao overfitting. A escolha do kernel também influencia esse equilíbrio – kernels mais complexos aumentam a capacidade de ajuste, podendo elevar a variância se usados sem regularização adequada.

Para contornar overfitting ou underfitting em SVM, é importante testar diferentes valores de C, experimentar tipos de kernel conforme o padrão dos dados e sempre normalizar as variáveis de entrada. O uso de validação cruzada para escolha dos hiperparâmetros é fundamental para garantir generalização.

SVM é robusto para dados com fronteiras bem definidas, funciona tanto para problemas lineares quanto não lineares, e é amplamente utilizado quando se busca precisão e boa capacidade de generalização.
