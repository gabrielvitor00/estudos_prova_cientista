**KNN (K-Nearest Neighbors)**

KNN é um algoritmo supervisionado simples e intuitivo, usado para classificação e regressão. Para fazer uma predição, o algoritmo identifica os K exemplos mais próximos do ponto a ser classificado, com base em uma métrica de distância, normalmente a euclidiana. Na classificação, a classe mais frequente entre os vizinhos é escolhida como resultado; na regressão, a média dos valores dos vizinhos é utilizada.

KNN não faz suposições sobre a distribuição dos dados e não possui etapa de treinamento explícita, já que todas as decisões são baseadas nos dados de treino armazenados. O desempenho é sensível à escolha de K, à escala das variáveis e ao aumento da dimensionalidade, podendo ser lento na predição em grandes bases. Por outro lado, é fácil de interpretar e eficaz para problemas com fronteiras de decisão complexas, quando os dados estão bem normalizados e não são muito volumosos.

KNN é útil como baseline e em problemas onde a simplicidade e a interpretabilidade local são desejadas.
