**Hierarchical Clustering**

Hierarchical Clustering é uma técnica de agrupamento não supervisionado que organiza os dados em uma hierarquia de clusters, representada graficamente por um dendrograma. O método pode ser aglomerativo (inicia com cada ponto em um cluster e vai agrupando os mais próximos) ou divisivo (começa com todos em um cluster e vai dividindo recursivamente).

O agrupamento é guiado por medidas de distância (como euclidiana ou Manhattan) e critérios de ligação (single, complete, average, ward), que determinam como calcular a proximidade entre clusters. A hierarquia permite visualizar relações entre grupos em diferentes níveis de granularidade e escolher o número de clusters ideal cortando o dendrograma em diferentes alturas.

Hierarchical Clustering não exige pré-definir o número de clusters e é interpretável visualmente. Contudo, tem custo computacional alto para grandes conjuntos, é sensível a outliers e a decisão de fusão não pode ser desfeita. O método é útil em análise exploratória, especialmente para segmentação ou descoberta de padrões onde a estrutura hierárquica é relevante.
