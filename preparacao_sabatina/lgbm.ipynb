# 💡 LightGBM (Light Gradient Boosting Machine)

**LightGBM** é um framework open source de boosting baseado em árvores, altamente otimizado para velocidade e escalabilidade. Ele implementa o método de **Gradient Boosting**, trazendo avanços no tipo de split, uso eficiente de memória e suporte nativo a grandes volumes de dados e variáveis categóricas.

---

# 💡 Objetivo

Fornecer um modelo de Gradient Boosting rápido, eficiente e escalável para grandes conjuntos de dados tabulares, capaz de entregar alta performance preditiva em tarefas de classificação e regressão, sem comprometer a velocidade de treinamento ou a memória.

---

## 🧮 Funcionamento e Matemática

### Diferenças-chave do LightGBM

- **Histograma-based:** Ao invés de calcular splits exatos em valores contínuos, LightGBM discretiza os valores das features em bins (histogramas), acelerando os cálculos e reduzindo uso de memória.
- **Level-wise Growth (Layered Split):** Por padrão, LightGBM cresce as árvores de forma “level-wise”, expandindo todos os nós de um nível antes de avançar para o próximo, gerando árvores mais balanceadas e rápidas de treinar.
- **Suporte nativo a variáveis categóricas:** Pode manipular colunas categóricas sem necessidade de one-hot encoding.
- **Gradient-based One-Side Sampling (GOSS):** Técnica de amostragem que prioriza exemplos com maiores gradientes, acelerando o treinamento.
- **Exclusive Feature Bundling (EFB):** Agrupa features mutuamente exclusivas para reduzir a dimensionalidade sem perder informação.

---

## ✅ Vantagens do LightGBM

- **Extremamente rápido** em grandes datasets, com uso eficiente de CPU e RAM.
- **Suporta alta dimensionalidade** e bases com milhões de exemplos e features.
- **Melhor performance que outros métodos em muitos cenários tabulares.**
- **Lida nativamente com variáveis categóricas.**
- Permite ajustar o tipo de crescimento das árvores (`growth_policy`).

---

## ⚠️ Desvantagens do LightGBM

- **Mais suscetível a overfitting** se não for ajustado cuidadosamente (árvores podem ser profundas).
- Pode ser menos interpretável do que modelos lineares ou até que Random Forest.
- Para datasets muito pequenos, pode não ter vantagem sobre modelos tradicionais.
- **Menos robusto a valores ausentes** do que CatBoost em alguns cenários.

---

# ⚖️ Trade-off entre Viés e Variância

- **Viés:** Baixo, pois permite árvores profundas e flexíveis.
- **Variância:** Pode ser alto se muitos estimadores ou árvores muito profundas forem usados.
- Parâmetros como `num_leaves`, `max_depth`, `min_data_in_leaf` e `learning_rate` são essenciais para regularização.

---

## 🧠 Conceitos-Chave

- **Histograma-based Split:** Transforma valores contínuos em bins para ganho de eficiência.
- **Level-wise (Layered):** Crescimento das árvores camada a camada (padrão), podendo ser alterado para node-wise (`leaf-wise`) para ajuste mais agressivo.
- **GOSS:** Seleciona exemplos “difíceis” para treinamento, acelerando aprendizado.
- **EFB:** Junta features esparsas e exclusivas, reduzindo dimensionalidade.
- **Suporte nativo a categóricas:** Reconhece, trata e faz split em variáveis categóricas diretamente.

---

## 🛠️ Como Lidar com o Trade-off

- Ajustar `num_leaves` (número máximo de folhas): folhas demais aumentam variância.
- Controlar profundidade (`max_depth`) e amostras mínimas em folha (`min_data_in_leaf`).
- Regularizar via `min_split_gain`, `bagging_fraction`, `feature_fraction`.
- Diminuir `learning_rate` e aumentar `num_boost_round` para aprendizagem mais robusta.
- Usar validação cruzada ou early stopping.

---

## ✅ Data Prep Essencial para LightGBM

1. **Tratar valores ausentes** (LightGBM lida, mas tratamento pode ajudar).
2. **Informar variáveis categóricas** ao modelo (não fazer one-hot, apenas indicar os índices das colunas categóricas).
3. **Remover variáveis irrelevantes** para ganho de performance e interpretação.
4. **Separar treino/validação/teste.**
5. **Balancear classes** se necessário (`scale_pos_weight`).

---

## ⚙️ Principais Parâmetros do LightGBM

| Parâmetro              | O que faz                                                                  |
|------------------------|----------------------------------------------------------------------------|
| `num_leaves`           | Número máximo de folhas por árvore (controla complexidade)                 |
| `max_depth`            | Profundidade máxima da árvore                                              |
| `learning_rate`        | Taxa de aprendizado (shrinkage)                                            |
| `n_estimators`/`num_boost_round` | Número de árvores no ensemble                                 |
| `min_data_in_leaf`     | Mínimo de exemplos por folha                                               |
| `feature_fraction`     | Fração de features usada em cada árvore                                    |
| `bagging_fraction`     | Fração de exemplos usada em cada árvore (subsampling)                      |
| `objective`            | Tipo de tarefa: `binary`, `multiclass`, `regression`, etc.                 |
| `boosting_type`        | Tipo de boosting: `gbdt`, `dart`, `rf`, etc.                               |
| `categorical_feature`  | Índices das colunas categóricas                                            |
| `growth_policy`        | Define se o split é level-wise (`'tree'`, padrão) ou node-wise (`'lossguide'`)|

---

## 📏 Principais Métricas de Avaliação

| Métrica        | Interpretação                                        |
|----------------|-----------------------------------------------------|
| **Acurácia**   | Proporção de acertos (classificação)                |
| **Precisão**   | Proporção de positivos previstos corretamente       |
| **Recall**     | Proporção de positivos reais identificados          |
| **F1-Score**   | Média harmônica entre precisão e recall             |
| **ROC AUC**    | Capacidade do modelo em separar classes             |
| **MAE/MSE/RMSE** | Métricas de erro para regressão                   |
| **Log Loss**   | Erro das probabilidades previstas (classificação)   |

---

## 📊 Visualização Intuitiva

- **Histograma de importâncias:** Permite identificar as features mais relevantes.
- **Curva de aprendizado:** Acompanhe performance vs. número de iterações.
- **Árvore individual:** Pode ser visualizada, mas a floresta é mais complexa.

---

## 🧪 Dica prática para ciência de dados

- Use LightGBM para grandes volumes tabulares, com muitas features ou valores categóricos.
- Sempre faça validação cruzada e ajuste de hiperparâmetros (principalmente `num_leaves`, `learning_rate`, `min_data_in_leaf`).
- Para evitar overfitting, combine early stopping com regularização e ajuste dos parâmetros de complexidade.

---

*Obs: LightGBM é frequentemente campeão em competições de ciência de dados por aliar rapidez, escalabilidade e alto desempenho preditivo em bases tabulares.*
