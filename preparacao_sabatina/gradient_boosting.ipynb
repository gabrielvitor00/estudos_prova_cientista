# 🚀 Gradient Boosting

**Gradient Boosting** é uma poderosa técnica de ensemble supervisionado, utilizada tanto para **classificação** quanto para **regressão**. Diferente do Random Forest, onde as árvores são construídas em paralelo e de forma independente, no Gradient Boosting cada árvore é construída **sequencialmente** para corrigir os erros do conjunto anterior, aprimorando iterativamente o desempenho do modelo.

---

# 🚀 Objetivo

Construir um modelo forte a partir da soma de vários modelos fracos (árvores pequenas), onde cada nova árvore **corrige os erros residuais** (diferença entre previsão atual e valor real) das anteriores. O objetivo é **minimizar uma função de custo** (loss) de forma iterativa, otimizando o desempenho preditivo.

---

## 🧮 Funcionamento e Matemática

### Processo de Treinamento

1. **Inicialização:**
   - O modelo começa com uma predição simples (por exemplo, a média do target no caso de regressão).

2. **Iterativamente:**
   - Em cada iteração, calcula-se o **resíduo** (erro) entre o valor real e a predição do modelo atual.
   - Uma nova árvore é treinada para prever esse resíduo (o gradiente da função de perda em relação à predição).
   - O modelo é atualizado somando a predição desta nova árvore, ponderada por um fator de aprendizado (`learning_rate`).
   - Repete-se o processo por um número definido de árvores (`n_estimators`).

3. **Agregação das Árvores:**
   - A predição final é a soma das predições de todas as árvores, cada uma corrigindo os erros das anteriores.

### Otimização

- O método utiliza **gradiente descendente** para minimizar a função de custo escolhida (ex: log loss para classificação, MSE para regressão).
- Pode ser usado com várias funções de custo, tornando-se altamente flexível.

---

### Tipos de Split: Level-wise vs. Node-wise

O método pelo qual as árvores são construídas no Gradient Boosting afeta tanto a eficiência quanto a capacidade de ajuste do modelo. Existem dois estilos principais:

- **Node-wise (Best-first):**
  - Cada nova divisão é feita no nó que mais reduz a função de custo (loss), independentemente do nível em que ele está na árvore.
  - As árvores crescem de maneira “irregular”, podendo ficar desbalanceadas.
  - **Utilizado por padrão em frameworks clássicos de Gradient Boosting**, como `scikit-learn`, **XGBoost** (default), e **CatBoost**.
  - **Vantagens:** Árvores mais adaptativas e profundas em regiões onde há mais complexidade.
  - **Desvantagens:** Pode favorecer overfitting em dados ruidosos, se não for bem regularizado.

- **Level-wise (Layered):**
  - Todos os nós de um mesmo nível são divididos simultaneamente antes de passar para o próximo nível.
  - Resulta em árvores mais balanceadas, com expansão “uniforme” por camada.
  - **Utilizado como padrão em frameworks como LightGBM** (mas pode ser ajustado).
  - **Vantagens:** Mais rápido e eficiente para grandes bases; tende a regularizar melhor.
  - **Desvantagens:** Pode perder poder preditivo em regiões muito complexas dos dados.

**Resumo prático:**  
- **Node-wise (Best-first):** Foca onde há mais erro, cresce a árvore “para baixo”, tornando-a assimétrica/adaptativa.  
- **Level-wise (Layered):** Cresce todas as folhas por camada, resultando em árvores mais simétricas e geralmente mais rasas.

> O tipo de split é um diferencial entre frameworks de boosting, influenciando velocidade, consumo de memória e risco de overfitting.

| Algoritmo         | Tipo de Split Padrão | Parametrizável?             | Observação                                         |
|-------------------|---------------------|-----------------------------|----------------------------------------------------|
| **scikit-learn GradientBoosting** | Node-wise (Best-first)  | Não                        | Implementação clássica, sempre node-wise           |
| **XGBoost**       | Node-wise (Best-first)  | Não                        | Árvores assimétricas, foca no maior ganho          |
| **CatBoost**      | Node-wise (Best-first)  | Não                        | Simples, sempre node-wise                          |
| **LightGBM**      | Level-wise (Layered)    | Sim                        | Parametrizável com `growth_policy` para node-wise  |
| **HistGradientBoosting (sklearn)** | Node-wise (Best-first) | Não                        | Implementação histogram-based, sempre node-wise    |

---

## ✅ Vantagens do Gradient Boosting

- **Alto poder preditivo** (um dos algoritmos mais competitivos em ML tabular).
- Pode ser usado com diferentes funções de perda (classificação, regressão, quantile, etc.).
- Pode lidar com diferentes tipos de dados, inclusive dados tabulares complexos.
- Permite controlar o aprendizado via parâmetros como `learning_rate` e profundidade das árvores.
- Fornece importâncias das variáveis.

---

## ⚠️ Desvantagens do Gradient Boosting

- **Mais sensível a overfitting** se não for ajustado cuidadosamente.
- **Mais lento** para treinar do que Random Forest (árvores são sequenciais).
- Precisa de ajuste fino de hiperparâmetros para melhor desempenho.
- Menos interpretável devido à complexidade do ensemble.
- Menos robusto a outliers em comparação com Random Forest.

---

# ⚖️ Trade-off entre Viés e Variância

- **Viés:** Tipicamente baixo, pois cada árvore corrige os erros da anterior.
- **Variância:** Pode ser alta se o número de árvores ou profundidade for muito grande.
- **Controle:** Parâmetros como `learning_rate`, `n_estimators` e `max_depth` ajudam a ajustar o equilíbrio.

---

## 🧠 Conceitos-Chave

- **Boosting:** Técnica onde modelos são treinados sequencialmente, cada um focando nos erros do anterior.
- **Learning Rate:** Fator de ajuste do quanto cada nova árvore contribui na correção dos erros. Valores mais baixos tornam o modelo mais robusto, porém exigem mais árvores.
- **Shrinkage:** Nome alternativo ao learning rate, refere-se ao encolhimento do passo de atualização.
- **Subsample:** Fração dos dados usada para treinar cada árvore, pode reduzir variância (stochastic gradient boosting).

---

## 🛠️ Como Lidar com o Trade-off

- **Ajustar learning rate:** Valores mais baixos previnem overfitting, mas exigem mais árvores.
- **Limitar profundidade das árvores (max_depth):** Evita que cada árvore ajuste demais ao ruído.
- **Utilizar subsampling (subsample < 1):** Aumenta a robustez contra overfitting.
- **Regularização adicional:** Parâmetros como `min_samples_leaf`, `max_features` e penalizações L1/L2.

---

## ✅ Data Prep Essencial para Gradient Boosting

1. **Tratar valores ausentes** (a maioria das implementações pode lidar, mas tratamento prévio é recomendado).
2. **Codificar variáveis categóricas** (one-hot ou ordinal, dependendo do framework).
3. **Remover variáveis irrelevantes** para melhorar performance e interpretação.
4. **Analisar e, se necessário, balancear as classes** (pode ajustar via parâmetro ou sample_weight).
5. **Separar dados em treino e teste**.
6. **Padronização/normalização** normalmente não é obrigatória, mas pode ser útil dependendo dos dados.

---

## ⚙️ Principais Parâmetros do Gradient Boosting

| Parâmetro           | O que faz                                                       |
|---------------------|-----------------------------------------------------------------|
| `n_estimators`      | Número de árvores (iterações de boosting)                       |
| `learning_rate`     | Taxa de aprendizado (impacto de cada árvore na predição final)  |
| `max_depth`         | Profundidade máxima de cada árvore                              |
| `subsample`         | Fração dos dados usados para treinar cada árvore                |
| `min_samples_leaf`  | Mínimo de amostras em cada folha                                |
| `loss`              | Função de custo a ser minimizada                                |
| `max_features`      | Número máximo de features consideradas em cada split            |
| `random_state`      | Semente para reprodutibilidade                                  |

## 📏 Principais Métricas de Avaliação

### Classificação

| Métrica        | Interpretação                                        |
|----------------|-----------------------------------------------------|
| **Acurácia**   | Proporção de acertos                                |
| **Precisão**   | Proporção de positivos previstos corretamente       |
| **Recall**     | Proporção de positivos reais identificados          |
| **F1-Score**   | Média harmônica entre precisão e recall             |
| **ROC AUC**    | Capacidade do modelo em separar classes             |
| **Log Loss**   | Erro das probabilidades previstas                   |

### Regressão

| Métrica        | Interpretação                                        |
|----------------|-----------------------------------------------------|
| **MAE**        | Erro médio absoluto                                 |
| **MSE**        | Erro quadrático médio                               |
| **RMSE**       | Raiz do erro quadrático médio                       |
| **R²**         | Proporção da variância explicada                    |

---

## 📊 Visualização Intuitiva

- Gradient Boosting constrói árvores pequenas que, sozinhas, são fracas, mas, ao serem somadas sequencialmente, vão **corrigindo os erros residuais** e formando um modelo muito poderoso.
- Permite visualizar a curva de aprendizagem (learning curve) para ajustar hiperparâmetros e evitar overfitting.

---

## 🧪 Dica prática para ciência de dados

- Use Gradient Boosting para problemas tabulares complexos que exigem alto desempenho, especialmente quando Random Forest não atinge os resultados desejados.
- Faça validação cruzada e ajuste de hiperparâmetros (principalmente `learning_rate` e `n_estimators`).
- Frameworks populares incluem **scikit-learn GradientBoosting**, **XGBoost**, **LightGBM** e **CatBoost**, que trazem otimizações adicionais.

---

*Obs: Gradient Boosting é o “coração” de muitos modelos campeões em competições de dados tabulares (Kaggle, etc.). Use com cautela em bases pequenas, pois o risco de overfitting pode ser maior.*
