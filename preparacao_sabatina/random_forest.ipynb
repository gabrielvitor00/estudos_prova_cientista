# 🌲 Random Forest

**Random Forest** é um algoritmo de Machine Learning supervisionado do tipo ensemble, que combina múltiplas **árvores de decisão** para produzir previsões mais robustas, precisas e estáveis, tanto para **classificação** quanto para **regressão**. Ele reduz o risco de overfitting e melhora a generalização em relação a uma única árvore.

---

# 🌲 Objetivo

Construir uma coleção (“floresta”) de árvores de decisão independentes, onde cada árvore contribui com um “voto” (classificação) ou média (regressão) para a predição final. O objetivo é **agregar múltiplos modelos fracos** para criar um modelo forte.

---

## 🧮 Funcionamento e Matemática

### Processo de Treinamento

1. **Bootstrap Aggregating (Bagging):**
   - Cada árvore é treinada em uma amostra aleatória dos dados originais (com reposição).
   - Isso gera diversidade entre as árvores, tornando-as menos correlacionadas entre si.

2. **Seleção Aleatória de Features:**
   - Em cada divisão (split) de cada árvore, um subconjunto aleatório de features é selecionado para determinar a melhor divisão.
   - Isso força as árvores a aprenderem padrões diferentes e reduz a dependência de features dominantes.

3. **Agregação das Predições:**
   - **Classificação:** A predição final é feita por votação majoritária (“majority voting”).
   - **Regressão:** A predição é a média das saídas das árvores.

---

## ✅ Vantagens do Random Forest

- **Alta performance preditiva** em comparação com uma única árvore.
- **Robusto a outliers** e ruído nos dados.
- **Reduz risco de overfitting**, pois cada árvore é “ruidosa”, mas a média tende a generalizar bem.
- Pode lidar com variáveis categóricas e numéricas sem necessidade de pré-processamento intenso.
- Estima automaticamente a importância das features (feature importance).

---

## ⚠️ Desvantagens do Random Forest

- **Menos interpretável** do que uma árvore única (floresta difícil de explicar).
- Pode ser computacionalmente custoso (tempo e memória) para grandes bases e muitas árvores.
- Tende a não performar tão bem para dados muito esparsos ou de alta dimensionalidade sem ajuste fino.
- Menos eficiente para tarefas onde interpretabilidade é crítica.

---

# ⚖️ Trade-off entre Viés e Variância

- **Viés:** Moderado, pois cada árvore tem alto viés, mas o ensemble reduz esse efeito.
- **Variância:** Reduzida drasticamente pelo processo de bagging e votação.
- **Resultado:** Random Forest atinge ótimo equilíbrio, proporcionando boa generalização em muitos cenários.

---

## 🧠 Conceitos-Chave

- **Bagging:** Técnica de ensemble baseada em amostragem e agregação.
- **Votação Majoritária:** Estratégia de decisão baseada na maioria das árvores (classificação).
- **Feature Importance:** Mede a contribuição de cada feature para as divisões na floresta.
- **Out-of-Bag Error (OOB):** Estimativa de erro feita usando amostras não vistas por cada árvore durante o treinamento, sem necessidade de conjunto de validação separado.

---

## 🛠️ Como Lidar com o Trade-off

- **Aumentar o número de árvores (n_estimators):** Reduz variância, mas aumenta custo computacional.
- **Ajustar profundidade máxima das árvores (max_depth):** Controla o viés/variância.
- **Controlar número de features em cada split (max_features):** Influencia diversidade das árvores.
- **Utilizar OOB error:** Para validação rápida e ajuste de hiperparâmetros sem separar conjunto de teste.

---

## ✅ Data Prep Essencial para Random Forest

1. **Tratar valores ausentes** (algumas implementações lidam automaticamente, mas é recomendável tratamento prévio).
2. **Codificar variáveis categóricas** (árvore aceita tanto dummies quanto ordinais).
3. **Remover variáveis irrelevantes** para performance e interpretação.
4. **Analisar balanceamento das classes** (class_weight pode ser ajustado).
5. **Separar dados em treino e teste**.
6. **Padronização/normalização geralmente não é obrigatória.**

---

## ⚙️ Principais Parâmetros do Random Forest

| Parâmetro         | O que faz                                                           |
|-------------------|---------------------------------------------------------------------|
| `n_estimators`    | Número de árvores na floresta                                       |
| `criterion`       | Métrica de split: `gini`, `entropy`, `mse`, `mae`                   |
| `max_depth`       | Profundidade máxima de cada árvore                                  |
| `max_features`    | Número de features a serem consideradas em cada split               |
| `min_samples_leaf`| Mínimo de amostras em cada folha                                    |
| `bootstrap`       | Se usa amostragem com reposição                                     |
| `oob_score`       | Se calcula o erro Out-of-Bag para validação                         |
| `class_weight`    | Ajusta peso das classes (problemas desbalanceados)                  |
| `random_state`    | Semente para reprodutibilidade                                      |

---

## 📏 Principais Métricas de Avaliação

### Classificação

| Métrica        | Interpretação                                        |
|----------------|-----------------------------------------------------|
| **Acurácia**   | Proporção de acertos                                |
| **Precisão**   | Proporção de positivos previstos corretamente       |
| **Recall**     | Proporção de positivos reais identificados          |
| **F1-Score**   | Média harmônica entre precisão e recall             |
| **ROC AUC**    | Capacidade do modelo em separar classes             |

### Regressão

| Métrica        | Interpretação                                        |
|----------------|-----------------------------------------------------|
| **MAE**        | Erro médio absoluto                                 |
| **MSE**        | Erro quadrático médio                               |
| **RMSE**       | Raiz do erro quadrático médio                       |
| **R²**         | Proporção da variância explicada                    |

---

## 📊 Visualização Intuitiva

- O Random Forest pode ser visto como uma “votação democrática” entre várias árvores: cada árvore individual pode ser fraca e cometer erros, mas o conjunto tende a acertar pela média/maioria.
- Permite analisar **importância das variáveis**, ajudando na interpretação do modelo, mesmo que a floresta em si seja opaca.

---

## 🧪 Dica prática para ciência de dados

- Utilize Random Forest como baseline para classificação e regressão: geralmente oferece bom desempenho com pouco ajuste de hiperparâmetros.
- Explore o atributo `.feature_importances_` para identificar variáveis relevantes.
- Caso performance seja crítica, ajuste hiperparâmetros como `n_estimators`, `max_depth` e `max_features` via validação cruzada.
- Use o erro OOB para avaliação rápida sem precisar de split de validação separado.

---

*Obs: Para ainda mais performance, considere técnicas de boosting (ex: Gradient Boosting, XGBoost), que refinam ainda mais os resultados, porém com maior custo computacional e menor interpretabilidade.*
