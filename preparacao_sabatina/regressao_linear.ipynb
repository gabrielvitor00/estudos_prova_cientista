# 📈 Regressão Linear

A **regressão linear** **é um algoritmo de Machine Learning supervisionado**, sendo uma das técnicas mais clássicas e fundamentais em estatística e ciência de dados utilizado para **modelar relações entre variáveis contínuas**. Ela modela a relação entre uma variável dependente (target) e uma ou mais variáveis independentes (features) assumindo que essa relação é linear.

---

# 📈 Objetivo
Encontrar uma reta (ou hiperplano) que melhor se ajusta aos dados, minimizando a distância entre os valores reais e os valores previstos. (Resíduos)

---

## 🧮 Fórmula Matemática

### Regressão Linear Simples (uma variável)

$$
\hat{y} = \beta_0 + \beta_1 x
$$

- $\hat{y}$: valor previsto  
- $x$: variável independente  
- $\beta_0$: intercepto (valor de $y$ quando $x = 0$)  
- $\beta_1$: coeficiente angular (variação de $y$ para cada unidade de $x$)

---

### Regressão Linear Múltipla (múltiplas variáveis)

$$
\hat{y} = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \ldots + \beta_p x_p
$$

ou, em forma vetorial:

$$
\hat{y} = \mathbf{X} \boldsymbol{\beta}
$$

---

## ⚙️ Como o Modelo Aprende?

O modelo encontra os coeficientes $\beta$ que **minimizam o erro quadrático médio (MSE)**:

$$
MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$

A otimização pode ser feita por:

- **Gradiente Descendente**
- **Método dos Mínimos Quadrados (álgebra linear):**

$$
\boldsymbol{\beta} = (\mathbf{X}^T \mathbf{X})^{-1} \mathbf{X}^T \mathbf{y}
$$

---

## ✅ Vantagens da Regressão Linear

- **Simplicidade e Interpretabilidade**  
  Fácil de entender e interpretar, especialmente útil para análises exploratórias e relatórios executivos.

- **Rápida de Treinar**  
  Baixo custo computacional mesmo em grandes volumes de dados.

- **Base para Modelos Mais Avançados**  
  Serve como referência inicial para benchmarking de modelos mais complexos.

- **Coeficientes Explicativos**  
  Os coeficientes $\beta$ mostram a magnitude e direção do impacto de cada variável independente.

- **Funcionamento Determinístico**  
  Dado o mesmo conjunto de dados, sempre produz os mesmos resultados (sem aleatoriedade).

---

## ⚠️ Desvantagens da Regressão Linear

- **Suposição de Linearidade**  
  Assume que a relação entre as variáveis independentes e dependente é linear, o que nem sempre é verdade.

- **Sensível a Outliers**  
  Valores extremos podem distorcer os coeficientes e as previsões.

- **Multicolinearidade**  
  Correlações fortes entre variáveis independentes podem tornar os coeficientes instáveis.

- **Heterocedasticidade**  
  Violações da homocedasticidade (variância constante dos erros) afetam a validade dos testes estatísticos.

- **Não Capta Relações Complexas**  
  Não consegue modelar interações não lineares ou relações complexas entre as variáveis sem engenharia adicional.

- **Requer Pressupostos Fortes**  
  Para ser estatisticamente confiável, precisa atender a pressupostos como:
  - Normalidade dos resíduos
  - Independência dos erros
  - Homocedasticidade
  - Não multicolinearidade

---

*Obs: Quando esses pressupostos não são atendidos, técnicas como regressão polinomial, regularização (Ridge/Lasso) ou modelos não lineares podem ser alternativas mais apropriadas.*

# ⚖️ Trade-off entre Viés e Variância

O **trade-off entre viés e variância** descreve o equilíbrio necessário entre **simplicidade** e **complexidade** de um modelo para obter **boa generalização** em dados nunca vistos.

---

## 🧠 Conceitos-Chave

- **Viés (Bias)**  
  Erro introduzido por suposições incorretas no modelo. Reflete o **quanto o modelo está distante da realidade**.  
  - Modelos **simples demais** (como regressão linear para um problema não linear) têm **alto viés**.

- **Variância (Variance)**  
  Refere-se à **sensibilidade do modelo aos dados de treinamento**. Um modelo com alta variância aprende até os **ruídos** dos dados.  
  - Modelos **complexos demais** (como árvores profundas ou redes neurais sem regularização) têm **alta variância**.

---

## 🔁 Trade-off

- Aumentar a **complexidade** do modelo → **Reduz o viés**, mas **aumenta a variância**.
- Simplificar o modelo → **Aumenta o viés**, mas **reduz a variância**.

O **objetivo é encontrar o ponto ótimo**, onde o erro total é **mínimo**, mesmo que o modelo **não seja perfeito**.

---

## 🛠️ Como Lidar com o Trade-off

### Estratégias para **reduzir variância (overfitting)**:
- Usar modelos mais simples
- Cross-validation (ex: k-fold)
- Regularização: **Ridge, Lasso, Elastic Net**
- Aumentar quantidade de dados
- Early stopping (em redes neurais)
- Pruning (em árvores de decisão)

### Estratégias para **reduzir viés (underfitting)**:
- Usar modelos mais complexos
- Adicionar novas features
- Reduzir regularização
- Transformar variáveis (interações, polinomiais, etc.)

---

## 📊 Visualização Intuitiva

Imaginando o desempenho do modelo como um gráfico:

- No eixo **x**: complexidade do modelo  
- No eixo **y**: erro de validação

O **erro de validação** inicialmente cai (reduzindo o viés), atinge um ponto ótimo, e depois **sobe** (aumentando a variância).

---

## 🧪 Dica prática para ciência de dados

Utilize **curvas de aprendizado**, **cross-validation** e análise de **resíduos** para identificar se o seu modelo está sofrendo mais de **alto viés** ou **alta variância**, e então aplique as técnicas apropriadas para corrigir.

- **Alto Viés:** O modelo é **simples demais** → use **mais features** ou **modelos mais flexíveis**.
- **Alta Variância:** O modelo é **complexo demais** → use **regularização**, **menos features** ou **mais dados**.

Equilibrar essas estratégias é essencial para alcançar **bom desempenho preditivo** e **generalização**.

---

## ✅ Data Prep Essencial para Regressão Linear

1. **Tratar valores ausentes**
   - Imputação (média, mediana, moda) ou remoção

2. **Remover ou tratar outliers**
   - Boxplot, z-score, IQR, winsorização, log

3. **Codificar variáveis categóricas**
   - One-hot encoding (evitar dummy trap)

4. **Padronizar ou normalizar variáveis**
   - Especialmente se usar regularização (Ridge/Lasso)

5. **Verificar e corrigir multicolinearidade**
   - Matriz de correlação e VIF

6. **Selecionar variáveis relevantes**
   - Eliminar irrelevantes e redundantes (ex: via RFE ou Lasso)

7. **Transformar variáveis não lineares**
   - Log, raiz, quadrado, interações, polinômios

8. **Analisar os resíduos (após o ajuste)**
   - Homocedasticidade, normalidade, média ≈ 0

9. **Separar dados em treino e teste**
   - Ex: 80/20 + validação cruzada (k-fold)

## ⚙️ Principais Parâmetros da Regressão Linear

| Parâmetro         | O que faz                                                                 |
|-------------------|---------------------------------------------------------------------------|
| `fit_intercept`   | Se `True`, calcula o intercepto $\beta_0$; se `False`, força a origem (0) |
| `normalize`       | (Descontinuado) Normalizava os dados antes do ajuste                      |
| `copy_X`          | Se `True`, faz cópia dos dados de entrada (evita sobrescrever)            |
| `n_jobs`          | Número de CPUs usadas (em tarefas paralelas como cross-validation)        |
| `positive`        | Se `True`, força os coeficientes $\beta$ a serem positivos                |

---

## 📏 Principais Métricas de Avaliação para Regressão Linear

| Métrica        | Interpretação                                                                 |
|----------------|--------------------------------------------------------------------------------|
| **MAE**        | Erro médio absoluto: média dos erros em valor absoluto                        |
| **MSE**        | Erro quadrático médio: penaliza mais os grandes erros                         |
| **RMSE**       | Raiz do MSE: tem mesma unidade do target                                      |
| **R² (R-quadrado)** | Proporção da variância explicada pelo modelo $(0 \leq R^2 \leq 1)$          |
| **Adjusted R²**| Ajusta o $R^2$ pelo número de variáveis no modelo                             |
| **MAPE**       | Erro percentual absoluto médio: expressa erro em termos percentuais           |
| **MSLE**       | MSE sobre valores log-transformados (útil para targets positivos e escalas grandes) |
