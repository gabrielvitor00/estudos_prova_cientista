# 🟠 DBSCAN (Density-Based Spatial Clustering of Applications with Noise)

**DBSCAN** é um algoritmo de agrupamento não supervisionado baseado em densidade, capaz de identificar clusters de formato arbitrário e detectar outliers (ruído) automaticamente. Muito útil para bases em que o número de clusters não é conhecido e/ou há ruído significativo.

---

# 🟠 Objetivo

Identificar regiões densas de pontos no espaço de features, agrupando-os em clusters, enquanto pontos isolados são classificados como ruído. O algoritmo **não exige pré-definir o número de clusters** e é especialmente útil para encontrar clusters de forma irregular.

---

## 🧮 Funcionamento e Matemática

- **Parâmetros principais:**
  - $\varepsilon$ (**epsilon**): raio de vizinhança considerado para agrupar pontos.
  - **min_samples:** número mínimo de pontos para formar uma região densa (núcleo de cluster).

- **Tipos de pontos:**
  - **Core points:** têm pelo menos `min_samples` vizinhos dentro de $\varepsilon$.
  - **Border points:** têm menos de `min_samples` vizinhos, mas estão na vizinhança de um core point.
  - **Noise points:** não pertencem a nenhum cluster (outliers).

- **Etapas do algoritmo:**
  1. Para cada ponto, verifica se ele é um core point.
  2. Expande o cluster a partir de cada core point, incluindo todos os pontos conectados por densidade.
  3. Pontos que não pertencem a nenhum cluster são rotulados como ruído.

---

## ✅ Vantagens do DBSCAN

- Não requer definir o número de clusters antecipadamente.
- Identifica clusters de qualquer formato (não só esféricos).
- Detecta automaticamente outliers e pontos de ruído.
- Robusto a clusters de densidades variadas (em certos cenários).
- Não sensível à ordem dos dados.

---

## ⚠️ Desvantagens do DBSCAN

- Sensível à escolha dos parâmetros $\varepsilon$ e `min_samples`.
- Difícil de ajustar para dados com densidades muito variadas.
- Performance ruim em dados de alta dimensionalidade (maldição da dimensionalidade).
- Não lida bem com dados categóricos puros.

---

# ⚖️ Trade-off entre Viés e Variância

- **$\varepsilon$ pequeno:** Muitos pontos como ruído, muitos clusters pequenos (alta variância).
- **$\varepsilon$ grande:** Poucos clusters, risco de misturar grupos distintos (alto viés).
- Ajustar bem os parâmetros é fundamental para capturar a estrutura real dos dados.

---

## 🧠 Conceitos-Chave

- **Epsilon ($\varepsilon$):** Raio do entorno de cada ponto.
- **min_samples:** Número mínimo de pontos para formar um cluster.
- **Core/Border/Noise points:** Classificação fundamental do DBSCAN.
- **Clusters arbitrários:** Capaz de identificar formas complexas e não esféricas.

---

## 🛠️ Como Lidar com o Trade-off

- Utilizar o gráfico k-distance plot para escolher um bom valor de $\varepsilon$.
- Ajustar `min_samples` conforme a densidade esperada dos clusters.
- Padronizar as variáveis para evitar viés devido à escala.
- Reduzir dimensionalidade para melhorar desempenho e definição dos clusters.

---

## ✅ Data Prep Essencial para DBSCAN

1. **Normalizar ou padronizar as variáveis**.
2. **Reduzir dimensionalidade** (PCA, t-SNE) se necessário.
3. **Remover outliers extremos** (opcional, DBSCAN já detecta outliers, mas pode facilitar ajuste).
4. **Escolher métrica de distância apropriada** ao tipo de dado (Euclidiana é padrão).

---

## ⚙️ Principais Parâmetros do DBSCAN

| Parâmetro     | O que faz                                            |
|---------------|------------------------------------------------------|
| `eps`         | Raio máximo da vizinhança ($\varepsilon$)            |
| `min_samples` | Número mínimo de pontos para formar um cluster        |
| `metric`      | Métrica de distância (Euclidiana, Manhattan, etc.)    |
| `algorithm`   | Algoritmo de busca de vizinhança (ball_tree, kd_tree, etc.) |
| `leaf_size`   | Parâmetro para eficiência dos algoritmos de busca     |

---

## 📏 Principais Métricas de Avaliação

| Métrica            | Interpretação                                     |
|--------------------|---------------------------------------------------|
| **Silhouette Score** | Coesão/separação dos clusters                   |
| **Homogeneidade**  | Similaridade dentro dos clusters                  |
| **Número de clusters detectados** | Quantidade e qualidade dos agrupamentos |
| **Porcentagem de ruído** | Proporção de pontos classificados como outliers |

---

## 📊 Visualização Intuitiva

- Projeção dos clusters encontrados em 2D/3D com destaque para outliers.
- k-distance plot para definir $\varepsilon$.
- Comparação entre clusters de diferentes formas/densidades.

---

## 🧪 Dica prática para ciência de dados

- Use DBSCAN para identificar clusters de formas irregulares, detectar outliers e quando não souber o número de clusters a priori.
- Sempre normalize e reduza a dimensionalidade para melhorar resultados.
- Teste múltiplos valores de $\varepsilon$ e `min_samples` usando o gráfico k-distance.
- Prefira métodos alternativos (OPTICS, HDBSCAN) em bases com densidade muito variável.

---

*Obs: DBSCAN é referência em agrupamento não supervisionado para bases com clusters de formas complexas e presença de ruído, sendo ótima opção para análise exploratória robusta.*
