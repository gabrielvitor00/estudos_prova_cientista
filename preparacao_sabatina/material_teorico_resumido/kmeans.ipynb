# 🔵 K-Means Clustering

O **K-Means** é um algoritmo de aprendizado não supervisionado, muito utilizado para **agrupamento** (clustering) de dados numéricos. Ele visa particionar o conjunto de dados em “K” clusters, de modo que cada ponto pertença ao grupo cujo centro (média) está mais próximo.

---

# 🔵 Objetivo

Agrupar dados em K grupos distintos, minimizando a soma das distâncias quadráticas dos pontos ao centroide de seu cluster.

---

## 🧮 Funcionamento e Matemática

- **Inicialização:** Seleciona K centróides iniciais (aleatórios ou via métodos como k-means++).
- **Atribuição:** Cada ponto é atribuído ao cluster cujo centróide está mais próximo (distância Euclidiana, padrão).
- **Atualização:** Recalcula os centróides como a média dos pontos de cada cluster.
- **Iteração:** Repete atribuição e atualização até os centróides estabilizarem (ou atingir limite de iterações).

- **Função objetivo:**
    $$
    \underset{C}{\operatorname{arg\,min}} \sum_{i=1}^{K} \sum_{x \in C_i} \| x - \mu_i \|^2
    $$
    Onde \( C_i \) é o cluster i, \( \mu_i \) é o centróide do cluster i.

---

## ✅ Vantagens do K-Means

- Simples, rápido e eficiente para bases grandes e médias.
- Escalável para grandes volumes de dados.
- Fácil de interpretar e implementar.
- Muito útil como método inicial de segmentação.

---

## ⚠️ Desvantagens do K-Means

- Requer pré-definir o número de clusters (K).
- Sensível a outliers e pontos ruidosos.
- Só lida bem com dados numéricos e esféricos.
- Não é robusto para clusters de tamanhos/densidades muito diferentes.
- Resultados podem variar conforme a inicialização.

---

# ⚖️ Trade-off entre Viés e Variância

- **K pequeno:** Clusters grandes, alto viés, baixo ajuste.
- **K grande:** Clusters pequenos, risco de overfitting/local patterns (alta variância).
- Escolher o K ideal (Elbow method, Silhouette Score) é crítico.

---

## 🧠 Conceitos-Chave

- **Centróide:** Média dos pontos de um cluster.
- **Distância Euclidiana:** Padrão para medir proximidade.
- **Elbow method:** Técnica gráfica para escolher K ideal.
- **Cluster esférico:** Formato natural para K-Means funcionar bem.

---

## 🛠️ Como Lidar com o Trade-off

- Normalizar dados antes de aplicar K-Means.
- Testar diferentes valores de K.
- Usar inicialização k-means++ para melhores centróides iniciais.
- Rodar múltiplas vezes para evitar mínimos locais.

---

## ✅ Data Prep Essencial para K-Means

1. **Normalizar/padronizar dados**.
2. **Remover outliers**.
3. **Selecionar features relevantes**.
4. **Reduzir dimensionalidade** se necessário.

---

## ⚙️ Principais Parâmetros do K-Means

| Parâmetro    | O que faz                                      |
|--------------|-----------------------------------------------|
| `n_clusters` | Número de clusters (K)                         |
| `init`       | Método de inicialização dos centróides         |
| `n_init`     | Número de rodadas com diferentes inicializações|
| `max_iter`   | Máximo de iterações por rodada                 |
| `tol`        | Tolerância para convergência                   |

---

## 📏 Principais Métricas de Avaliação

| Métrica            | Interpretação                                |
|--------------------|----------------------------------------------|
| **Inertia**        | Soma das distâncias quadráticas intra-cluster|
| **Silhouette Score** | Coesão/separação dos clusters             |
| **Davies-Bouldin** | Índice de qualidade de agrupamento           |

---

## 📊 Visualização Intuitiva

- Projeção dos clusters em 2D/3D, colorindo pontos por grupo.
- Gráfico do método Elbow para escolha de K.
- Plot dos centróides em destaque.

---

## 🧪 Dica prática para ciência de dados

- K-Means é ótimo para dados contínuos e bases não muito ruidosas.
- Sempre normalize as variáveis.
- Use como baseline antes de métodos mais complexos.
- Combine com redução de dimensionalidade para visualização.

---

*Obs: K-Means é um dos métodos de clustering mais populares, mas deve ser aplicado com cuidado a bases categóricas ou com outliers.*
