# ⚡ AdaBoost (Adaptive Boosting)

**AdaBoost (Adaptive Boosting)** é um algoritmo de ensemble supervisionado focado em **classificação** (originalmente) e adaptado para regressão, que constrói uma sequência de classificadores simples (“fracos”, geralmente stumps – árvores de decisão com uma única divisão). Cada novo classificador concentra-se em corrigir os erros dos anteriores, atribuindo mais peso às amostras classificadas erroneamente.

---

# ⚡ Objetivo

Criar um modelo forte combinando vários modelos fracos, onde cada etapa do treinamento dá **maior importância (peso)** às amostras que foram incorretamente classificadas anteriormente. O objetivo é **reduzir o erro de classificação geral** iterativamente.

---

## 🧮 Funcionamento e Matemática

### Processo de Treinamento

1. **Inicialização dos Pesos:**
   - Todos os exemplos do treino recebem peso igual.

2. **Treinamento Sequencial:**
   - Um classificador fraco (ex: stump) é treinado a cada iteração.
   - O desempenho desse classificador é avaliado. Amostras classificadas incorretamente recebem aumento de peso.
   - O próximo classificador é treinado considerando os pesos atualizados.

3. **Combinação dos Classificadores:**
   - Cada classificador recebe um peso na votação, proporcional à sua acurácia.
   - O modelo final é uma votação ponderada dos classificadores fracos.

### Atualização dos Pesos

- O erro de cada classificador é calculado com base nos pesos atuais.
- Amostras erradas recebem maior peso para as próximas iterações, forçando o modelo a focar onde errou mais.

### Predição Final

- **Classificação:** Votação ponderada pelos pesos de cada classificador.
- **Regressão:** Média ponderada das predições.

---

## ✅ Vantagens do AdaBoost

- **Melhora substancialmente o desempenho** de classificadores fracos.
- **Simples de implementar** e eficaz em muitos problemas.
- **Menos suscetível a overfitting** do que outros ensembles (com modelos fracos bem controlados).
- Pode ser combinado com diferentes algoritmos base (árvores, regressão logística, etc.).

---

## ⚠️ Desvantagens do AdaBoost

- **Muito sensível a outliers** e dados ruidosos (pois aumenta peso de erros, pode focar em ruídos).
- Menos eficiente para grandes volumes de dados comparado a métodos mais modernos (como Gradient Boosting).
- Requer dados balanceados; classes muito desbalanceadas podem prejudicar o desempenho.
- Menos flexível em relação ao controle de hiperparâmetros avançados.

---

# ⚖️ Trade-off entre Viés e Variância

- **Viés:** Reduzido iterativamente ao focar nos exemplos difíceis.
- **Variância:** Baixa a moderada, pois os classificadores são simples e o método combina múltiplos modelos.
- **Atenção:** AdaBoost pode aumentar a variância se utilizar modelos-base muito complexos.

---

## 🧠 Conceitos-Chave

- **Stump:** Árvore de decisão com apenas um split (modelo fraco clássico).
- **Peso das amostras:** Amostras erradas ganham mais peso, forçando o algoritmo a focar onde está errando.
- **Votação Ponderada:** Classificadores mais precisos têm maior influência no resultado final.
- **Taxa de erro:** Mede o desempenho de cada modelo-base e determina sua influência no ensemble.

---

## 🛠️ Como Lidar com o Trade-off

- Usar modelos-base simples (stumps) para evitar overfitting.
- Limitar o número de estimadores (`n_estimators`) para controlar variância.
- Remover outliers e tratar dados ruidosos antes do ajuste.
- Balancear as classes quando necessário.

---

## ✅ Data Prep Essencial para AdaBoost

1. **Tratar valores ausentes** antes do treinamento.
2. **Codificar variáveis categóricas** se o modelo-base exigir.
3. **Remover outliers** para não distorcer o treinamento.
4. **Balancear as classes** caso haja grande desbalanceamento.
5. **Separar dados em treino e teste** para avaliação justa.

---

## ⚙️ Principais Parâmetros do AdaBoost

| Parâmetro         | O que faz                                                       |
|-------------------|-----------------------------------------------------------------|
| `n_estimators`    | Número de classificadores fracos (stumps, por padrão)           |
| `base_estimator`  | Tipo de modelo-base usado (ex: DecisionTreeClassifier)          |
| `learning_rate`   | Peso de contribuição de cada classificador                      |
| `random_state`    | Semente para reprodutibilidade                                  |

---

## 📏 Principais Métricas de Avaliação

| Métrica        | Interpretação                                        |
|----------------|-----------------------------------------------------|
| **Acurácia**   | Proporção de acertos                                |
| **Precisão**   | Proporção de positivos previstos corretamente       |
| **Recall**     | Proporção de positivos reais identificados          |
| **F1-Score**   | Média harmônica entre precisão e recall             |
| **ROC AUC**    | Capacidade do modelo em separar classes             |

---

## 📊 Visualização Intuitiva

- AdaBoost constrói uma sequência de modelos simples, onde cada novo modelo “corrige” os erros dos anteriores, tornando o ensemble cada vez mais forte.
- O modelo final é, na prática, uma votação ponderada baseada no desempenho dos classificadores fracos.

---

## 🧪 Dica prática para ciência de dados

- AdaBoost é excelente para melhorar rapidamente a performance de modelos simples em conjuntos de dados tabulares.
- Se notar sensibilidade a outliers, experimente modelos-base mais robustos ou limpe os dados antes do ajuste.
- Use como baseline ou em competições quando interpretabilidade e velocidade forem importantes.

---

*Obs: AdaBoost marcou a origem do conceito moderno de boosting, sendo precursor de métodos mais sofisticados como Gradient Boosting e XGBoost.*
