# 🟣 K-Medoids (PAM, Partitioning Around Medoids)

**K-Medoids** é uma variante robusta do K-Means para **clustering**, utilizada quando há outliers, ruído, ou quando não se quer usar médias como centróides. O cluster é representado por um “medoid” (exemplo real mais central), e não pela média dos pontos.

---

# 🟣 Objetivo

Agrupar os dados em K clusters minimizando a soma das distâncias dos pontos ao medoid do seu cluster, escolhendo sempre um ponto real como representante.

---

## 🧮 Funcionamento e Matemática

- **Inicialização:** Escolhe K medoids (pontos reais do dataset).
- **Atribuição:** Cada ponto é atribuído ao cluster cujo medoid está mais próximo (por distância definida).
- **Atualização:** Para cada cluster, testa substituir o medoid por outro ponto do cluster para reduzir o custo total.
- **Iteração:** Repete atribuição e atualização até não haver mais trocas vantajosas.

- **Função objetivo:**
    $$
    \underset{M}{\operatorname{arg\,min}} \sum_{i=1}^{K} \sum_{x \in C_i} d(x, m_i)
    $$
    Onde \( m_i \) é o medoid do cluster \( C_i \) e \( d \) é a métrica de distância.

---

## ✅ Vantagens do K-Medoids

- Mais robusto a outliers do que K-Means.
- Medoids são sempre pontos reais do dataset.
- Pode ser usado com qualquer métrica de distância (não apenas Euclidiana).
- Funciona melhor em dados com clusters de forma/densidade diferente.

---

## ⚠️ Desvantagens do K-Medoids

- Mais lento que K-Means (custo computacional elevado em bases grandes).
- Ainda requer definir K previamente.
- Não escala tão bem para conjuntos de dados muito grandes.

---

# ⚖️ Trade-off entre Viés e Variância

- **Menor variância que K-Means** em datasets com outliers.
- Para bases pequenas/médias, geralmente melhor estabilidade que K-Means.
- Para bases muito grandes, pode ser impraticável.

---

## 🧠 Conceitos-Chave

- **Medoid:** Ponto real mais representativo do cluster.
- **PAM (Partitioning Around Medoids):** Algoritmo clássico para K-Medoids.
- **Métrica de distância:** Pode ser Euclidiana, Manhattan, ou personalizada.

---

## 🛠️ Como Lidar com o Trade-off

- Use para bases com muitos outliers ou variáveis não numéricas.
- Ajuste K via métodos gráficos (Elbow, Silhouette Score).
- Considere amostragem se a base for grande.

---

## ✅ Data Prep Essencial para K-Medoids

1. **Normalizar/padronizar variáveis** (se necessário).
2. **Remover outliers grossos, se possível**.
3. **Selecionar K cuidadosamente**.
4. **Escolher a métrica de distância apropriada** ao domínio dos dados.

---

## ⚙️ Principais Parâmetros do K-Medoids

| Parâmetro    | O que faz                                 |
|--------------|------------------------------------------|
| `n_clusters` | Número de clusters (K)                   |
| `metric`     | Métrica de distância                     |
| `init`       | Método de inicialização dos medoids      |
| `max_iter`   | Máximo de iterações                      |
| `random_state` | Semente para reprodutibilidade        |

---

## 📏 Principais Métricas de Avaliação

| Métrica            | Interpretação                                |
|--------------------|----------------------------------------------|
| **Custo total**    | Soma das distâncias aos medoids              |
| **Silhouette Score** | Coesão/separação dos clusters             |
| **Davies-Bouldin** | Índice de qualidade de agrupamento           |

---

## 📊 Visualização Intuitiva

- Visualização semelhante ao K-Means, mas medoids são sempre pontos reais.
- Análise da estabilidade dos medoids frente a outliers.

---

## 🧪 Dica prática para ciência de dados

- Prefira K-Medoids a K-Means quando houver muitos outliers ou variáveis categóricas (com distância apropriada).
- Utilize em bases pequenas/médias e para análise exploratória robusta.

---

*Obs: K-Medoids é alternativa mais robusta ao K-Means e excelente para segmentação onde a média não é representativa.*
