# ⚖️ Desbalanceamento de Classes: Como Lidar e Corrigir

O **desbalanceamento de classes** ocorre quando uma ou mais classes estão sub-representadas em relação a outras. Esse cenário é muito comum em aplicações reais (fraude, churn, doenças raras, etc.), mas pode levar a **modelos enviesados**, incapazes de identificar corretamente a classe minoritária.

---

## 🧩 Por que é um problema?

- Modelos tendem a "ignorar" a classe rara e superestimar a majoritária.
- Métricas tradicionais (acurácia) podem mascarar desempenho ruim na minoria.
- Reduz capacidade de generalização, levando a alto viés para a classe majoritária.

---

## 🛠️ Técnicas para Lidar com Desbalanceamento

### 1. **Amostragem dos Dados**

#### a) Oversampling (Aumento da minoria)
- **Random Oversampling:** Replica aleatoriamente exemplos da classe minoritária até balancear.
- **SMOTE (Synthetic Minority Over-sampling Technique):** Cria exemplos sintéticos da classe minoria interpolando entre exemplos reais.
- **ADASYN, Borderline-SMOTE:** Variações para gerar amostras sintéticas de modo mais controlado.

#### b) Undersampling (Redução da maioria)
- **Random Undersampling:** Remove aleatoriamente exemplos da classe majoritária até balancear.
- **Cluster Centroids:** Usa centroides de clusters da maioria para representar a base reduzida.
- **Tomek Links/ENN:** Remove exemplos ambíguos na fronteira entre classes.

#### c) Oversample + Undersample Combinados
- Usar ambas técnicas pode reduzir overfitting e melhorar generalização.

---

### 2. **Ajuste de Algoritmos (Hiperparâmetros e Função de Custo)**

- **Class Weights / Sample Weights:**
    - Muitos algoritmos aceitam o parâmetro `class_weight='balanced'` (ex: Logistic Regression, Random Forest, SVM, XGBoost, CatBoost, LightGBM).
    - Penaliza mais os erros na classe minoritária, forçando o modelo a “prestar atenção” nela.
    - Pode-se também customizar os pesos manualmente.

- **Alteração da função de custo:**
    - Custom loss (ex: focal loss em redes neurais), onde erros em exemplos raros são mais penalizados.
    - Ajustar hiperparâmetros de regularização, threshold de decisão, ou adicionar penalidades.

---

### 3. **Ajustes no Fluxo de Modelagem**

- **Mudar métrica de avaliação:**  
    - Prefira **precision, recall, F1-score, ROC AUC, PR AUC** ao invés de acurácia simples.
    - Use matriz de confusão para interpretar o resultado.
- **Ajustar limiar de decisão:**  
    - Alterar o threshold de classificação para favorecer recall/precision da classe de interesse.
- **Validação estratificada:**  
    - Sempre usar cross-validation estratificada, garantindo proporções similares nas folds de treino/validação.

---

## 🧪 Boas Práticas e Dicas para Ciência de Dados

- Sempre **analise a distribuição das classes** antes de modelar.
- Use **pipelines** para aplicar oversampling/undersampling **dentro do CV**, evitando data leak.
- Teste o impacto de diferentes técnicas e ajuste conforme o domínio do problema.
- Explique os resultados considerando a classe minoritária: F1, recall, PR-AUC são mais informativos.

---

## 📦 Exemplos práticos de aplicação

| Técnica                   | Onde aplicar                   |
|---------------------------|-------------------------------|
| `SMOTE`, `RandomOverSampler` | Pacote `imblearn` (scikit-learn) |
| `class_weight`            | `LogisticRegression`, `RandomForestClassifier`, `SVC`, `XGBoost`, `LightGBM`, `CatBoost` |
| Custom loss               | Redes neurais (Keras, PyTorch, TensorFlow)   |

---

> **Resumo:**  
Desbalanceamento de classes é uma das principais causas de falha em modelos reais.  
Combinar amostragem adequada, ajuste de pesos, métrica correta e análise de threshold é a chave para alta performance em contextos desbalanceados!
