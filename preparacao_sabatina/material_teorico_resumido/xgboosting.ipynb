# 🏆 XGBoost (Extreme Gradient Boosting)

**XGBoost** é uma das bibliotecas de boosting mais populares e poderosas para tarefas de classificação e regressão em dados tabulares. Implementa o algoritmo de **Gradient Boosting**, trazendo otimizações de velocidade, regularização e robustez, sendo destaque em competições e aplicações práticas de ciência de dados.

---

# 🏆 Objetivo

Criar modelos altamente preditivos e eficientes combinando múltiplas árvores de decisão sequenciais, onde cada nova árvore corrige os erros das anteriores. O foco é entregar **máxima performance com escalabilidade**, robustez a overfitting e flexibilidade para diferentes tipos de tarefa.

---

## 🧮 Funcionamento e Matemática

### Otimizações-Chave do XGBoost

- **Construção node-wise (best-first):** Expande sempre o nó com maior redução de erro, formando árvores assimétricas e profundas em regiões complexas dos dados.
- **Booster sequencial:** Cada árvore é treinada para prever o gradiente da função de perda dos resíduos do ensemble anterior.
- **Suporte a regularização explícita:** Aplica penalização L1 e L2 nos coeficientes das árvores, reduzindo overfitting.
- **Split histogram-based:** Opção para acelerar splits em grandes volumes de dados.
- **Handle de missing values:** Gerencia valores ausentes automaticamente durante o split.

### Fórmula simplificada da otimização:

A cada iteração, o modelo minimiza a função objetivo:
$$
Obj = L(y, \hat{y}) + \Omega(f)
$$
Onde \( L \) é a função de perda (ex: log loss, mse) e \( \Omega(f) \) é o termo de regularização (complexidade da árvore).

---

## ✅ Vantagens do XGBoost

- **Alta performance e acurácia** em problemas tabulares.
- **Treinamento rápido** e eficiente (paralelização, histogram split).
- **Forte regularização** (L1, L2, max_depth, min_child_weight) para controlar overfitting.
- Suporta missing values e variáveis categóricas (em versões recentes).
- Flexível para diferentes funções de perda (classificação, regressão, ranking, quantile, etc.).
- Oferece métricas de importância das features.
