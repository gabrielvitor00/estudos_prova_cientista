# 🟡 GMM (Gaussian Mixture Models)

**GMM (Gaussian Mixture Models)** é um algoritmo de aprendizado não supervisionado para **clustering** e **modelagem de densidade**, que representa a distribuição dos dados como uma combinação (mistura) de várias distribuições Gaussianas (normais), permitindo capturar clusters com formatos elípticos e sobrepostos.

---

# 🟡 Objetivo

Modelar os dados como uma soma ponderada de distribuições normais multivariadas, onde cada componente Gaussiano representa um “cluster” e cada ponto possui uma probabilidade de pertencer a cada cluster (soft clustering).

---

## 🧮 Funcionamento e Matemática

- **Modelo:** Cada cluster é representado por uma distribuição Gaussiana parametrizada por média ($\mu_k$), covariância ($\Sigma_k$) e peso ($\pi_k$).
- **Probabilidade de pertencimento:** Cada ponto tem uma probabilidade associada a cada cluster, ao invés de pertencer estritamente a apenas um grupo.
- **Função densidade da mistura:**
    $$
    p(x) = \sum_{k=1}^K \pi_k \mathcal{N}(x | \mu_k, \Sigma_k)
    $$
    Onde $\mathcal{N}$ é a densidade Gaussiana multivariada.

- **Estimativa dos parâmetros:** Feita via algoritmo de Expectation-Maximization (EM):
  - **E-step:** Calcula as probabilidades de cada ponto pertencer a cada componente (responsabilidades).
  - **M-step:** Atualiza os parâmetros das distribuições (média, covariância, pesos) maximizando a verossimilhança.

---

## ✅ Vantagens do GMM

- **Clustering suave (soft):** Permite que pontos pertençam parcialmente a mais de um cluster.
- **Flexível:** Capta clusters elípticos, alongados ou sobrepostos.
- **Modelagem de densidade:** Além de clustering, pode estimar a densidade dos dados.
- **Estatisticamente interpretável:** Baseia-se em probabilidade, permite inferências e detecção de anomalias.

---

## ⚠️ Desvantagens do GMM

- Requer pré-definir o número de componentes (clusters).
- **Sensível à inicialização e a mínimos locais.**
- Não lida bem com alta dimensionalidade ou features muito correlacionadas sem regularização.
- Supõe que os dados realmente vêm de uma mistura de Gaussianas (nem sempre verdadeiro).
- Não é robusto a outliers extremos (afeta as médias e covariâncias).

---

# ⚖️ Trade-off entre Viés e Variância

- **Poucos componentes:** Alto viés (subestima complexidade, mistura clusters diferentes).
- **Muitos componentes:** Alta variância (overfitting, cada ponto pode virar um cluster).
- **Regularização da covariância** pode ajudar a evitar instabilidade para dados de alta dimensão.

---

## 🧠 Conceitos-Chave

- **Componente Gaussiano:** Cada cluster é uma Gaussiana multivariada.
- **Soft clustering:** Associação probabilística entre ponto e cluster.
- **Covariância:** Pode ser “full” (elíptico, qualquer direção), “diag” (elipses alinhadas com eixos), ou “spherical” (esférico).
- **Log-verossimilhança:** Usada para avaliar o ajuste do modelo.
- **Algoritmo EM:** Expectation-Maximization, alterna entre atribuição probabilística e atualização dos parâmetros.

---

## 🛠️ Como Lidar com o Trade-off

- Escolher o número de componentes (K) por métodos gráficos (AIC, BIC, curva elbow).
- Regularizar covariância para evitar problemas de ajuste em alta dimensão.
- Testar diferentes tipos de covariância.
- Rodar múltiplas inicializações para evitar mínimos locais.

---

## ✅ Data Prep Essencial para GMM

1. **Normalizar/padronizar as features** (importante para distância e ajuste da Gaussiana).
2. **Remover outliers extremos**.
3. **Reduzir dimensionalidade** se necessário (PCA, etc.).
4. **Selecionar features relevantes**.

---

## ⚙️ Principais Parâmetros do GMM

| Parâmetro         | O que faz                                          |
|-------------------|---------------------------------------------------|
| `n_components`    | Número de componentes Gaussianos (clusters)       |
| `covariance_type` | Tipo de matriz de covariância (`'full'`, `'diag'`, `'tied'`, `'spherical'`) |
| `init_params`     | Inicialização (`'kmeans'` ou `'random'`)          |
| `max_iter`        | Máximo de iterações do EM                         |
| `tol`             | Tolerância para convergência                      |
| `reg_covar`       | Regularização da covariância                      |
| `random_state`    | Semente para reprodutibilidade                    |

---

## 📏 Principais Métricas de Avaliação

| Métrica             | Interpretação                                 |
|---------------------|-----------------------------------------------|
| **Log-likelihood**  | Verossimilhança do modelo nos dados           |
| **AIC/BIC**         | Critérios de informação para escolha de K      |
| **Silhouette Score** | Coesão/separação dos clusters                |
| **Homogeneidade**   | Grau de uniformidade interna dos clusters      |

---

## 📊 Visualização Intuitiva

- Projeção dos clusters (via PCA/t-SNE) com elipses representando as covariâncias.
- Gráficos de log-verossimilhança/AIC/BIC por número de componentes.
- Plot de “responsabilidades” mostrando o grau de pertencimento dos pontos a cada cluster.

---

## 🧪 Dica prática para ciência de dados

- Use GMM quando espera clusters não esféricos ou sobrepostos.
- Sempre normalize as features e teste diferentes inicializações/covariâncias.
- Utilize AIC/BIC para evitar overfitting ao escolher o número de componentes.
- Interprete as probabilidades de pertencimento para análises de incerteza, segmentação e detecção de anomalias.

---

*Obs: GMM é ferramenta estatística avançada para clustering e modelagem de densidade, recomendada quando a estrutura dos dados sugere misturas suaves e não-linearidades nos agrupamentos.*
