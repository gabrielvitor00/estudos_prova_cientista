# 🌳 Árvores de Decisão

As **árvores de decisão** são algoritmos supervisionados de Machine Learning, usados tanto para **classificação** quanto para **regressão**. Elas modelam decisões em forma de árvore, realizando divisões sucessivas dos dados a partir de regras baseadas nas features, até chegar a um resultado (classe ou valor).

---

# 🌳 Objetivo

Construir um modelo em formato de árvore, onde cada nó interno representa uma decisão baseada em uma feature, cada ramo é um resultado possível dessa decisão, e cada folha representa uma predição final (classe ou valor numérico).

---

## 🧮 Funcionamento e Matemática

### Estrutura da Árvore

- **Nó raiz (root):** onde a primeira divisão dos dados ocorre.
- **Nós internos:** representam decisões baseadas em condições das features (ex: `idade > 30`).
- **Folhas:** resultado final (classe no caso de classificação, valor no caso de regressão).

### Critérios de Divisão

O objetivo das árvores de decisão é **aumentar a pureza** dos subconjuntos a cada divisão, ou seja, criar nós onde os exemplos pertençam majoritariamente a uma única classe (classificação) ou sejam o mais homogêneos possível (regressão). Para isso, são utilizadas métricas específicas:

#### **Classificação**

- **Gini (Impureza de Gini):**
  - Mede a probabilidade de um item ser classificado incorretamente se for rotulado aleatoriamente segundo a distribuição das classes no nó.
  - **Fórmula:**
    $$
    Gini = 1 - \sum_{i=1}^C p_i^2
    $$
    Onde \(p_i\) é a proporção de elementos da classe \(i\) no nó.
  - **Interpretação prática:** Quanto menor o Gini, mais puro o nó (tendendo a 0 quando todos os exemplos são da mesma classe).

- **Entropia:**
  - Mede o grau de desordem (ou incerteza) das classes em um nó.
  - **Fórmula:**
    $$
    Entropia = -\sum_{i=1}^C p_i \log_2(p_i)
    $$
  - **Interpretação prática:** Entropia é 0 se o nó é puro (só tem uma classe), e atinge o máximo quando todas as classes estão igualmente distribuídas.
  - **O que significa na prática?** Alta entropia = alta desordem, baixa entropia = alto grau de pureza.

- **Ganho de Informação:**
  - Mede a redução da entropia após uma divisão — ou seja, **quanto uma feature contribui para tornar os nós mais puros**.
  - **Fórmula:**
    $$
    Ganho\ de\ Informação = Entropia_{\text{pai}} - \sum_{k=1}^{K} \frac{N_k}{N} Entropia_k
    $$
    Onde \(N\) é o número de amostras no nó pai, \(N_k\) no filho \(k\).
  - **Interpretação prática:** Quanto maior o ganho de informação, mais eficiente foi a divisão em reduzir a incerteza das classes.

#### **Regressão**

- **MSE (Mean Squared Error):**
  - Calcula o erro médio quadrático dentro de cada folha.
  - **Fórmula:**
    $$
    MSE = \frac{1}{N} \sum_{i=1}^{N} (y_i - \bar{y})^2
    $$
    Onde \(\bar{y}\) é o valor médio dos exemplos no nó.
  - **Interpretação prática:** Quanto menor o MSE, mais homogêneos (parecidos) são os valores dos exemplos naquela folha.

- **MAE (Mean Absolute Error):**
  - Mede o erro absoluto médio entre os valores reais e a média do nó.
  - **Fórmula:**
    $$
    MAE = \frac{1}{N} \sum_{i=1}^{N} |y_i - \bar{y}|
    $$
  - **Interpretação prática:** O MAE é menos sensível a outliers do que o MSE, refletindo o desvio médio absoluto dos valores em relação à média do nó. Nó com baixo MAE indica alta homogeneidade.

---

### Algoritmo de Treinamento

1. Começa no nó raiz com todos os dados.
2. Para cada feature e ponto de corte, calcula a métrica de impureza.
3. Seleciona a divisão que gera os subconjuntos mais “puros”.
4. Repete o processo recursivamente em cada ramo até atingir um critério de parada (profundidade máxima, mínimo de amostras, pureza total, etc.).

---

## ✅ Vantagens das Árvores de Decisão

- **Fácil interpretação e visualização**
- Lida naturalmente com variáveis categóricas e numéricas
- Não requer normalização ou padronização de dados
- Capaz de modelar relações não lineares e interações complexas
- Pode lidar com dados faltantes (dependendo da implementação)

---

## ⚠️ Desvantagens das Árvores de Decisão

- **Tendência ao overfitting** (árvores profundas)
- Instáveis a pequenas variações nos dados (alta variância)
- Podem gerar árvores muito complexas se não houver poda/limite
- Resultados menos robustos se comparados a ensembles (Random Forest, Boosting)
- Não são boas para extrapolar fora dos dados vistos

---

# ⚖️ Trade-off entre Viés e Variância

- Árvores **rasas**: alto viés, baixa variância (underfitting)
- Árvores **profundas**: baixo viés, alta variância (overfitting)
- O **ponto ótimo** é buscado ajustando hiperparâmetros (poda, max_depth, min_samples_leaf, etc.)

---

## 🧠 Conceitos-Chave

- **Poda (Pruning):** Reduzir a complexidade removendo ramos desnecessários após o treinamento, prevenindo overfitting.
- **Feature Importance:** Avaliação da importância de cada variável baseada em sua contribuição nas divisões.
- **Split:** Critério para dividir os dados (Gini, Entropia, MSE).
- **Folha:** Nó final da árvore que representa a predição.

---

## 🛠️ Como Lidar com o Trade-off

- **Limitar profundidade máxima (max_depth)**
- Definir mínimo de amostras por folha (min_samples_leaf)
- Realizar **poda** pós-treinamento
- Usar técnicas de ensemble para reduzir variância (Random Forest, Gradient Boosting)

---

## ✅ Data Prep Essencial para Árvores de Decisão

1. **Tratar valores ausentes** (árvore pode lidar, mas é recomendável)
2. **Codificar variáveis categóricas** (árvore lida com dummies e, em alguns casos, com rótulos)
3. **Remover variáveis irrelevantes** (para clareza e desempenho)
4. **Separar em treino e teste**
5. **Análise de balanceamento das classes** (em problemas de classificação)

---

## ⚙️ Principais Parâmetros das Árvores de Decisão

| Parâmetro              | O que faz                                                                |
|------------------------|--------------------------------------------------------------------------|
| `criterion`            | Métrica usada para divisão (`'gini'`, `'entropy'`, `'mse'`, etc.)        |
| `max_depth`            | Profundidade máxima da árvore                                            |
| `min_samples_split`    | Mínimo de amostras para dividir um nó                                   |
| `min_samples_leaf`     | Mínimo de amostras em cada folha                                        |
| `max_features`         | Número máximo de features avaliadas por divisão                         |
| `random_state`         | Semente para reprodutibilidade                                          |

---

## 📏 Principais Métricas de Avaliação

### Classificação

| Métrica        | Interpretação                                        |
|----------------|-----------------------------------------------------|
| **Acurácia**   | Proporção de acertos                                |
| **Precisão**   | Proporção de positivos previstos corretamente       |
| **Recall**     | Proporção de positivos reais identificados          |
| **F1-Score**   | Média harmônica entre precisão e recall             |
| **Matriz de Confusão** | Mostra acertos/erros por classe            |

### Regressão

| Métrica        | Interpretação                                        |
|----------------|-----------------------------------------------------|
| **MAE**        | Erro médio absoluto                                 |
| **MSE**        | Erro quadrático médio                               |
| **RMSE**       | Raiz do erro quadrático médio                       |
| **R²**         | Proporção da variância explicada                    |

---

## 📊 Visualização Intuitiva

Árvores podem ser representadas graficamente, mostrando como cada decisão divide os dados até chegar às folhas. Essa visualização é útil para explicar predições a públicos não técnicos.

---

## 🧪 Dica prática para ciência de dados

- Árvores de decisão são excelentes como modelo explicativo inicial.
- Para maior desempenho preditivo, utilize técnicas de ensemble como **Random Forest** ou **Gradient Boosting**.
- Analise a “feature importance” para identificar quais variáveis são realmente relevantes no problema.

---

*Obs: Árvores de decisão, embora intuitivas, devem ser usadas com cautela em produção, preferencialmente dentro de ensembles, para evitar overfitting e instabilidade.*
