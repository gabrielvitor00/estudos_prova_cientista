# CRISP-DM (Cross Industry Standard Process for Data Mining)

O CRISP-DM é um framework consagrado para condução de projetos de ciência de dados, composto por seis fases principais, estruturando o trabalho de ponta a ponta de forma iterativa e adaptável:

**1. Compreensão do Negócio**  
Definir claramente o objetivo do projeto e entender as necessidades do negócio. Traduza o problema de negócio em um problema de ciência de dados.

**2. Compreensão dos Dados**  
Coletar, descrever, explorar e avaliar a qualidade dos dados. Identificar padrões iniciais, inconsistências, outliers e variáveis relevantes.

**3. Preparação dos Dados**  
Selecionar, limpar, transformar e formatar os dados para modelagem. Envolve tratamento de valores ausentes, criação de variáveis, normalização e seleção de atributos.

**4. Modelagem**  
Escolher, treinar e ajustar algoritmos. Comparar modelos, otimizar hiperparâmetros, validar desempenho com métricas adequadas.

**5. Avaliação**  
Verificar se os modelos atendem às necessidades do negócio. Avaliar métricas quantitativas e qualitativas, além de garantir que não há riscos ou limitações críticas.

**6. Implantação**  
Colocar a solução em produção ou entregar resultados acionáveis ao negócio. Inclui planejamento, documentação e monitoramento da entrega final.

O CRISP-DM é iterativo: ao final de cada etapa, pode ser necessário retornar a fases anteriores conforme novas descobertas ou ajustes de escopo. O framework é flexível e aplicável a projetos de qualquer área, promovendo organização, rastreabilidade e alinhamento entre equipes técnicas e de negócio.
