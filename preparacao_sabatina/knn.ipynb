# 🤝 KNN (K-Nearest Neighbors)

**KNN (K-Nearest Neighbors)** é um algoritmo supervisionado simples e intuitivo usado para **classificação** e **regressão**. Ele prevê o rótulo de um exemplo novo analisando os “K” exemplos mais próximos no espaço de features, baseando-se em alguma medida de distância.

---

# 🤝 Objetivo

Classificar (ou prever) um novo ponto analisando os “K” exemplos mais próximos do conjunto de treino e usando uma votação majoritária (classificação) ou média dos valores (regressão) para determinar o resultado.

---

## 🧮 Funcionamento e Matemática

- **Distância:** Para cada ponto novo, calcula-se a distância entre ele e todos os exemplos do conjunto de treino.
- **Seleção dos K vizinhos:** Seleciona-se os K pontos mais próximos (menor distância).
- **Classificação:** O rótulo é escolhido pela maioria entre os K vizinhos (votação).
- **Regressão:** O valor é a média (ou mediana) dos valores dos K vizinhos.

- **Fórmula da distância (Euclidiana, padrão):**
    $$
    d(x, x') = \sqrt{\sum_{i=1}^{n} (x_i - x'_i)^2}
    $$
    Onde \(x\) é o vetor do novo exemplo e \(x'\) é um exemplo do treino.

- Também pode-se usar outras métricas: Manhattan, Minkowski, distância de Mahalanobis, etc.

---

## ✅ Vantagens do KNN

- **Extremamente simples e fácil de entender.**
- Não faz suposições sobre a distribuição dos dados (não paramétrico).
- Bom desempenho em bases pequenas e/ou com limites de decisão não lineares.
- Adapta-se facilmente a mudanças nos dados (não há fase de treinamento).
- Útil para problemas de classificação e regressão.

---

## ⚠️ Desvantagens do KNN

- **Lento na predição:** O cálculo da distância para todos os pontos é custoso para grandes bases.
- **Sensível à escala das variáveis:** Features com escalas diferentes podem distorcer o cálculo das distâncias.
- **Desempenho ruim em bases muito grandes ou muito esparsas.**
- Não lida bem com dados de alta dimensionalidade (maldição da dimensionalidade).
- Pode ser sensível a outliers e a valores faltantes.

---

# ⚖️ Trade-off entre Viés e Variância

- **K pequeno:** Baixo viés, alta variância (modelo muito adaptativo, sensível ao ruído e overfitting).
- **K grande:** Alto viés, baixa variância (modelo mais estável, mas pode ignorar padrões locais e underfittar).
- **Escolha ótima de K:** Deve ser feita por validação cruzada.

---

## 🧠 Conceitos-Chave

- **K:** Número de vizinhos considerados para a votação ou média.
- **Peso dos vizinhos:** Pode-se ponderar vizinhos por distância (mais próximos têm maior influência).
- **Métrica de distância:** Euclidiana, Manhattan, Minkowski, etc.
- **Normalização:** Fundamental para garantir que todas as features contribuam de modo equilibrado.

---

## 🛠️ Como Lidar com o Trade-off

- Usar validação cruzada para escolher o valor ideal de **K**.
- Testar diferentes métricas de distância.
- Normalizar ou padronizar as variáveis.
- Considerar ponderação por distância.
- Remover outliers e tratar valores faltantes.

---

## ✅ Data Prep Essencial para KNN

1. **Normalizar/padronizar features** (essencial!).
2. **Remover outliers** que podem distorcer o resultado.
3. **Tratar valores ausentes** (imputação ou exclusão).
4. **Selecionar features relevantes** para evitar ruído.
5. **Reduzir dimensionalidade** em bases com muitas variáveis (PCA, seleção de features).

---

## ⚙️ Principais Parâmetros do KNN

| Parâmetro          | O que faz                                              |
|--------------------|-------------------------------------------------------|
| `n_neighbors` (K)  | Número de vizinhos usados na decisão                  |
| `metric`           | Métrica de distância (`'euclidean'`, `'manhattan'`, etc.) |
| `weights`          | `'uniform'` (todos iguais) ou `'distance'` (ponderados)|
| `algorithm`        | Estratégia de busca (`'auto'`, `'ball_tree'`, `'kd_tree'`, `'brute'`) |
| `p`                | Parâmetro da distância de Minkowski (p=1 Manhattan, p=2 Euclidiana) |

---

## 📏 Principais Métricas de Avaliação

| Métrica        | Interpretação                                        |
|----------------|-----------------------------------------------------|
| **Acurácia**   | Proporção de acertos (classificação)                |
| **Precisão**   | Proporção de positivos previstos corretamente       |
| **Recall**     | Proporção de positivos reais identificados          |
| **F1-Score**   | Média harmônica entre precisão e recall             |
| **ROC AUC**    | Capacidade do modelo em separar classes             |
| **MAE/MSE/RMSE** | Métricas de erro para regressão                   |

---

## 📊 Visualização Intuitiva

- KNN define regiões de decisão altamente flexíveis; as fronteiras podem ser complexas e seguir o formato real dos dados.
- A escolha de K e a normalização têm impacto direto na fronteira de decisão.

---

## 🧪 Dica prática para ciência de dados

- Sempre normalize os dados antes de usar KNN.
- Use K pequeno para capturar padrões locais, mas com cuidado para não overfittar.
- K grande é melhor para bases ruidosas, mas pode “diluir” padrões importantes.
- Em bases grandes ou de alta dimensionalidade, considere métodos alternativos ou otimize via algoritmos de busca.
- Útil como baseline simples e para problemas onde interpretabilidade local é importante.

---

*Obs: KNN é um excelente algoritmo introdutório para classificação/regressão, mas em bases muito grandes ou de alta dimensionalidade, modelos baseados em árvores ou redes podem ser preferíveis.*
