# 🟢 K-Modes Clustering

**K-Modes** é um algoritmo específico para **agrupamento de dados categóricos**. Ele é uma extensão do K-Means, substituindo a média pelo **modo** (categoria mais frequente) e usando uma métrica de dissimilaridade para variáveis qualitativas.

---

# 🟢 Objetivo

Agrupar dados categóricos em K clusters, minimizando a dissimilaridade total dentro de cada cluster, onde cada cluster é representado pelo modo (categoria mais comum) de cada feature.

---

## 🧮 Funcionamento e Matemática

- **Inicialização:** Seleciona K modos iniciais (categorias mais frequentes ou aleatórias).
- **Atribuição:** Cada ponto é atribuído ao cluster cujo modo é mais semelhante (menor dissimilaridade).
- **Atualização:** Para cada cluster, recalcula o modo de cada feature.
- **Iteração:** Repete atribuição e atualização até convergência.

- **Métrica de dissimilaridade:**
    $$
    d(X, Y) = \sum_{i=1}^m \delta(x_i, y_i)
    $$
    Onde \( \delta(x_i, y_i) = 0 \) se \( x_i = y_i \), senão 1.

---

## ✅ Vantagens do K-Modes

- Específico para dados categóricos.
- Simples e eficiente para bases pequenas/médias.
- Representatividade clara do cluster pelo modo de cada feature.
- Não exige transformação das variáveis em numéricas.

---

## ⚠️ Desvantagens do K-Modes

- Requer definir K previamente.
- Pode convergir para mínimos locais (dependente da inicialização).
- Não funciona para dados numéricos.
- Pode ser sensível a features com muitos níveis (altas cardinalidades).

---

# ⚖️ Trade-off entre Viés e Variância

- **K pequeno:** Clusters amplos, pode ignorar padrões locais (alto viés).
- **K grande:** Clusters pequenos, risco de overfitting (alta variância).
- Inicialização dos modos pode impactar resultados.

---

## 🧠 Conceitos-Chave

- **Modo:** Categoria mais frequente em cada coluna do cluster.
- **Dissimilaridade categórica:** Soma de discordâncias entre categorias.
- **Cardinalidade:** Número de níveis distintos em cada variável.

---

## 🛠️ Como Lidar com o Trade-off

- Escolher K por métodos gráficos ou validados (Silhouette Score adaptado).
- Rodar múltiplas inicializações para evitar mínimos locais.
- Considerar agrupar variáveis categóricas com muitas categorias.

---

## ✅ Data Prep Essencial para K-Modes

1. **Tratar valores ausentes** (imputar, remover).
2. **Atenção à cardinalidade alta:** Variáveis com muitos níveis podem distorcer resultados.
3. **Selecionar features relevantes**.
4. **Padronizar categorias (nomes, ortografia, etc.)**.

---

## ⚙️ Principais Parâmetros do K-Modes

| Parâmetro    | O que faz                                    |
|--------------|----------------------------------------------|
| `n_clusters` | Número de clusters (K)                       |
| `init`       | Método de inicialização dos modos            |
| `n_init`     | Número de inicializações diferentes          |
| `max_iter`   | Máximo de iterações                          |
| `cat_dissim` | Tipo de métrica de dissimilaridade           |

---

## 📏 Principais Métricas de Avaliação

| Métrica            | Interpretação                                |
|--------------------|----------------------------------------------|
| **Custo total**    | Soma das dissimilaridades dentro do cluster  |
| **Silhouette Score** | Adaptado para dissimilaridade categórica   |
| **Homogeneidade**  | Grau de similaridade dentro dos clusters     |

---

## 📊 Visualização Intuitiva

- Projeção dos clusters (via redução de dimensionalidade).
- Tabela resumo dos modos de cada cluster para interpretação.

---

## 🧪 Dica prática para ciência de dados

- Use K-Modes quando todas as variáveis são categóricas e a interpretação do cluster pelo “modo” for importante.
- Pode ser combinado com K-Means/K-Prototypes em bases mistas.

---

*Obs: K-Modes é referência para clustering de dados categóricos, mas depende da definição adequada de dissimilaridade e do tratamento de cardinalidade.*
