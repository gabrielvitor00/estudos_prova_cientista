# 🔄 Regressão Logística

A **regressão logística** é um algoritmo supervisionado de Machine Learning voltado para **problemas de classificação**, especialmente binária (duas classes). Ao contrário da regressão linear, que prevê valores contínuos, a logística modela a **probabilidade de um evento ocorrer**, atribuindo o resultado a uma das classes.

---

# 🔄 Objetivo

Modelar a probabilidade de ocorrência de uma classe alvo (ex: sucesso/fracasso, positivo/negativo), transformando a saída linear em uma escala entre 0 e 1, **usando a função sigmoide**.

---

## 🧮 Fórmula Matemática

### Logística Binária

A relação predita é:

$$
P(y=1|X) = \sigma(z) = \frac{1}{1 + e^{-z}}
$$

Onde:

- \( z = \beta_0 + \beta_1 x_1 + \ldots + \beta_p x_p \)
- \( \sigma(z) \): função sigmoide, transforma qualquer valor real em probabilidade.

> O modelo retorna a probabilidade de pertencer à classe 1. Para decidir a classe, compara-se o valor a um threshold (tipicamente 0.5).

---

## ⚙️ Como o Modelo Aprende?

Os coeficientes \( \beta \) são ajustados para **maximizar a verossimilhança** (Maximum Likelihood Estimation), e não para minimizar o erro quadrático.

A função de custo é o **Log Loss** (também chamada de Cross-Entropy):

$$
J(\beta) = -\frac{1}{n} \sum_{i=1}^n \left[y_i \log(\hat{p}_i) + (1 - y_i) \log(1 - \hat{p}_i)\right]
$$

Onde \( \hat{p}_i \) é a probabilidade prevista para a amostra \( i \).

---

## ✅ Vantagens da Regressão Logística

- **Simplicidade e interpretabilidade**
- Fornece probabilidades diretas e odds para tomada de decisão.
- Rápida de treinar, mesmo em grandes bases.
- Serve como baseline para comparação de modelos complexos.
- Possui extensões para multiclasse (one-vs-rest, softmax).

---

## ⚠️ Desvantagens da Regressão Logística

- **Limitação a relações lineares** entre variáveis e o logit.
- Não modela interações não lineares sem engenharia adicional.
- Menor performance para fronteiras complexas de decisão.
- Sensível a outliers e à multicolinearidade.
- Requer grande volume de dados para situações de classes muito desbalanceadas.

---

# ⚖️ Trade-off entre Viés e Variância

A regressão logística é **simples**, portanto:
- **Alto viés** (pode não capturar padrões complexos)
- **Baixa variância** (menos tendência a overfitting)

Com regularização (L1/L2), pode-se aumentar o viés e reduzir a variância ainda mais.

---

## 🧠 Conceitos-Chave

- **Sigmoide:** Função que converte valores contínuos para o intervalo (0, 1).
- **Logit:** Logaritmo da razão de chances, \( \log\left(\frac{p}{1-p}\right) \).
- **Odds:** Razão da probabilidade de sucesso sobre fracasso.
- **Regularização:** Penaliza coeficientes grandes para reduzir overfitting (L1 = Lasso, L2 = Ridge).

---

## 🛠️ Como Lidar com o Trade-off

- **Regularização:** Usar penalização L1/L2 para evitar que o modelo fique "sensível" ao ruído.
- **Cross-validation:** Avaliar estabilidade do modelo em diferentes subconjuntos.
- **Engenharia de features:** Adicionar variáveis, transformar relações, tratar outliers.

---

## ✅ Data Prep Essencial para Regressão Logística

1. **Tratar valores ausentes**
   - Imputação (mediana, moda) ou remoção

2. **Tratar outliers**
   - Boxplot, log, winsorização

3. **Codificar variáveis categóricas**
   - One-hot, ordinal, dummy

4. **Padronizar/normalizar variáveis**
   - Especialmente útil com regularização

5. **Verificar multicolinearidade**
   - Matriz de correlação, VIF

6. **Balanceamento de classes**
   - Subamostragem, superamostragem, SMOTE

7. **Separar dados em treino e teste**
   - Ex: 80/20 + validação cruzada

---

## ⚙️ Principais Parâmetros da Regressão Logística

| Parâmetro         | O que faz                                                         |
|-------------------|-------------------------------------------------------------------|
| `penalty`         | Tipo de regularização: `'l1'`, `'l2'`, `'elasticnet'`, `'none'`   |
| `C`               | Inverso da força de regularização (quanto menor, mais regularização) |
| `solver`          | Algoritmo de otimização usado (`'liblinear'`, `'saga'`, `'lbfgs'`, etc.) |
| `max_iter`        | Número máximo de iterações para convergência                      |
| `class_weight`    | Peso das classes (lida com desbalanceamento)                      |
| `fit_intercept`   | Ajusta o intercepto no modelo                                     |

---

## 📏 Principais Métricas de Avaliação para Regressão Logística

| Métrica        | Interpretação                                                                           |
|----------------|----------------------------------------------------------------------------------------|
| **Acurácia**   | Proporção de previsões corretas                                                        |
| **Precisão**   | Proporção de positivos previstos que são realmente positivos                           |
| **Recall**     | Proporção de positivos reais que foram corretamente identificados                      |
| **F1-Score**   | Média harmônica entre precisão e recall, útil para classes desbalanceadas              |
| **ROC AUC**    | Área sob a curva ROC, mede separação entre classes                                     |
| **Log Loss**   | Mede o erro da previsão probabilística; valores menores são melhores                   |

---

## 📊 Visualização Intuitiva

A saída do modelo é sempre uma **probabilidade**. Para problemas binários, o gráfico da função sigmoide mostra a transição suave de 0 para 1.

---

## 🧪 Dica prática para ciência de dados

Utilize regressão logística como **primeiro modelo** para classificação, interpretando coeficientes para entender o impacto de cada variável.  
Se o desempenho for insuficiente, avance para modelos mais complexos ou crie features não lineares.

---

*Obs: Para problemas multiclasse, utilize estratégias como One-vs-Rest (OvR) ou regressão logística multinomial (softmax).*
